## 1. Variables and Data Types

# 2. Conditional Statements

Conditional statements allow a program to make decisions
using if, elif, and else.

```python
marks = 75

if marks >= 90:
    print("Grade A")
elif marks >= 60:
    print("Grade B")
else:
    print("Needs improvement")
```

```python
name = "Alex"
age = 21
height = 5.8
is_student = True

print("Name:", name)
print("Age:", age)
print("Height:", height)
print("Student:", is_student)
```

## 3. Loops

Loops repeat a block of code. A for loop is useful
when iterating over a sequence of items.
Cell 7 — Code: For loop

```python
for number in range(1, 6):
    print(number)
```

## 4. Functions

Functions are reusable blocks of code. They can accept
parameters and return results

## 5. Lists and Dictionaries

A list stores an ordered collection of values.
A dictionary stores data as key-value pairs

```python
fruits = ["apple", "banana", "mango"]
student = {"name": "Alex", "marks": 85}

print("Fruits:", fruits)
print("First fruit:", fruits[0])
print("Student:", student)
print("Marks:", student["marks"])
```

Exception Handling

Exception handling helps a program respond to errors
without terminating unexpectedly.

```python
try:
    number = int("123")
    print("Number:", number)
except ValueError:
    print("Invalid number")
```
