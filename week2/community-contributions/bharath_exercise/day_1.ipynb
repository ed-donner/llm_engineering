```python
import os
from openai import OpenAI
import ollama
from IPython.display import Markdown, display

ollama_client = OpenAI(
    base_url="http://localhost:11434/v1",
    api_key="ollama"
)

```

```python
llama_1_model = "llama3.2"
llama_2_model = "llama3.2"
llama_3_model = "llama3.2"

llama_1_system = "You are a chatbot who is very argumentative; \
you disagree with anything in the conversation and you challenge everything, in a snarky way."

llama_2_system = "You are a very polite, courteous chatbot. You try to agree with \
everything the other person says, or find common ground. If the other person is argumentative, \
you try to calm them down and keep chatting."

llama_3_system = "You are a chatbot who is very hilarious ; \
you taunt the other chatbot and you challenge everything, in a snarky way."

llama_1_messages = ["Hi"]
llama_2_messages = ["Hi there"]
llama_3_messages = ["hello"]
```

```python
def call_llama_1():
    messages = [{"role": "system", "content": llama_1_system}]
    for llama_1, llama_2,llama_3 in zip(llama_1_messages, llama_2_messages,llama_3_messages):
        messages.append({"role": "assistant", "content": llama_1})
        messages.append({"role": "user", "content": llama_2})
        messages.append({"role": "assistant", "content": llama_3})
    response = ollama_client.chat.completions.create(model=llama_1_model, messages=messages)
    return response.choices[0].message.content
```

```python
def call_llama_2():
    messages = [{"role": "system", "content": llama_2_system}]
    for llama_1, llama_2_message,llama_3 in zip(llama_1_messages, llama_2_messages,llama_3_messages):
        messages.append({"role": "assistant", "content": llama_1})
        messages.append({"role": "user", "content": llama_2_message})
        messages.append({"role": "assistant", "content": llama_3})
    messages.append({"role": "user", "content": llama_1_messages[-1]})
    response = ollama_client.chat.completions.create(model=llama_2_model, messages=messages)
    return response.choices[0].message.content

```

```python
def call_llama_3():
    messages = [{"role": "system", "content": llama_3_system}]
    for llama_1, llama_2,llama_3_message in zip(llama_1_messages, llama_2_messages,llama_3_messages):
        messages.append({"role": "assistant", "content": llama_1})
        messages.append({"role": "user", "content": llama_2})
        messages.append({"role": "assistant", "content": llama_3_message})
    messages.append({"role": "user", "content": llama_2_messages[-1]})
    response = ollama_client.chat.completions.create(model=llama_3_model, messages=messages)
    return response.choices[0].message.content
```

```python
llama_1_messages = ["Hi there"]
llama_2_messages = ["Hi"]
llama_3_messages = ["hello"]

display(Markdown(f"### Llama_1:\n{llama_1_messages[0]}\n"))
display(Markdown(f"### Llama_2:\n{llama_2_messages[0]}\n"))
display(Markdown(f"### Llama_3:\n{llama_3_messages[0]}\n"))

for i in range(5):
    llama_1_next = call_llama_1()
    display(Markdown(f"### Llama_1:\n{llama_1_next}\n"))
    llama_1_messages.append(llama_1_next)
    llama_2_next = call_llama_2()
    display(Markdown(f"### Llama_2:\n{llama_2_next}\n"))
    llama_2_messages.append(llama_2_next)
    llama_3_next = call_llama_3()
    display(Markdown(f"### Llama_3:\n{llama_3_next}\n"))
    llama_3_messages.append(llama_3_next)
```
