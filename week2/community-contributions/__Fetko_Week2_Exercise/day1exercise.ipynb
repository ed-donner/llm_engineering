```python
# imports
import os
import importlib
import utility
from dotenv import load_dotenv
from openai import OpenAI
from IPython.display import Markdown, display

importlib.reload(utility)
from utility import call_model_for_conversation
```

```python
# initialize and constants

MODEL_GPT = 'gpt-6-luna'
MODEL_ANTHROPIC = 'claude-sonnet-5-5'
MODEL_GOOGLE = 'gemini-3.8-flash'

load_dotenv(override=True)
openai_api_key = os.getenv('OPENAI_API_KEY')
anthropic_api_key = os.getenv('ANTHROPIC_API_KEY')
google_api_key = os.getenv('GOOGLE_API_KEY')

anthropic_url = "https://api.anthropic.com/v1/"
gemini_url = "https://generativelanguage.googleapis.com/v1beta/openai/"

if openai_api_key and openai_api_key.startswith('sk-proj-') and len(openai_api_key)>10:
    print("OPENAI_API_KEY key looks good so far")
else:
    print("There might be a problem with your OPENAI_API_KEY key? Please visit the troubleshooting notebook!")

if anthropic_api_key and anthropic_api_key.startswith('sk-ant-usr-') and len(anthropic_api_key)>10:
    print("ANTHROPIC_API_KEY key looks good so far")
else:
    print("There might be a problem with your ANTHROPIC_API_KEY key? Please visit the troubleshooting notebook!")

if google_api_key and google_api_key.startswith('AQ.') and len(google_api_key)>10:
    print("GOOGLE_API_KEY key looks good so far")
else:
    print("There might be a problem with your GOOGLE_API_KEY key? Please visit the troubleshooting notebook!")

```

```python
# set up clients
gpt_client = OpenAI()
anthropic_client = OpenAI(api_key=anthropic_api_key, base_url=anthropic_url)
gemini_client = OpenAI(api_key=google_api_key, base_url=gemini_url) 
```

```python
# defining system messages

alex_system_message = """
    You are Alex, an expert Senior Frontend Developer. Your main expertise
    lies in the React framework.

    You strongly support your opinions with confidence, logic, and technical depth.
    You challenge weak arguments in moderate way but professionally.
    Your responses must stay focused on the topic.

    Rules:
    - Give powerful and intelligent arguments
    - Be concise and impactful
    - Speak like a highly experienced expert

    You are debating with Blake and Charlie.
"""

blake_system_message = """
    You are Blake, an expert Senior Frontend Developer. Your main expertise
    lies in the Angular framework.

    You strongly support your opinions with confidence, logic, and technical depth.
    You challenge weak arguments in aggresive way but professionally.
    Your responses must stay focused on the topic.

    Rules:
    - Give powerful and intelligent arguments
    - Be concise and impactful
    - Speak like a highly experienced expert

    You are debating with Alex and Charlie.
"""

charlie_system_message = """
    You are Charlie, an expert Senior Frontend Developer. Your main expertise
    lies in the Vue framework.

    You strongly support your opinions with confidence, logic, and technical depth.
    You challenge weak arguments in peacefull way but professionally.
    Your responses must stay focused on the topic.

    Rules:
    - Give powerful and intelligent arguments
    - Be concise and impactful
    - Speak like a highly experienced expert

    You are debating with Alex and Charlie.
"""

conversation = """
Alex: Hi everyone.
Blake: Hello Alex! Nice to meet you.
Charlie: what's up buddy let's do some discussion about "Which is the best Frontend Framework in the moment?".
"""
```

```python
# invoke conversation
display(Markdown("## Three Model Conversation:\n"))

for i in range(5):

    alex_answer = call_model_for_conversation(gpt_client, MODEL_GPT, alex_system_message, conversation)
    conversation += f"\nAlex: {alex_answer}"

    display(Markdown(f"### Alex\n{alex_answer}\n"))

    blake_answer = call_model_for_conversation(anthropic_client, MODEL_ANTHROPIC, blake_system_message, conversation)
    conversation += f"\nBlake: {blake_answer}"

    display(Markdown(f"### Blake\n{blake_answer}\n"))

    charlie_answer = call_model_for_conversation(gemini_client, MODEL_GOOGLE, charlie_system_message, conversation)
    conversation += f"\nCharlie: {charlie_answer}"

    display(Markdown(f"### Charlie\n{charlie_answer}\n"))
```
