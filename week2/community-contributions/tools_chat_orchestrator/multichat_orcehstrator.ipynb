### A simple 3 way chat orchestrator that involves 3 members own consideration - Just testing my knowledge 

```python
import os
import json
import requests
from dotenv import load_dotenv
from openai import OpenAI
from IPython.display import Markdown, display, update_display
from bs4 import BeautifulSoup
load_dotenv()
GROQ_BASE_URL = os.getenv("GROQ_BASE_URL")
GROQ_API_KEY = os.getenv("GROQAI_API_KEY")
groq = OpenAI(base_url = GROQ_BASE_URL, api_key = GROQ_API_KEY)
model = os.getenv("GROQ_MODEL")
```

```python
conversation = [
    {"harish":"What is prompt engineering?"},
]

# This will be used later when removing the chat overall chat conversation - to have conversation
initial_conversation = {"harish":"What is prompt engineering?"}
```

```python
# This might be kind of repeating

system_prompt_alex = """
You are Alex, an insightful and collaborative participant in a group discussion with Harish and Raghav.
Your goal is to build on the ideas shared by others, connect different concepts, and suggest fresh creative perspectives.
Keep your response strictly to 2 sentences. Always continue based on the last conversation
"""

user_prompt_alex = f"""
The conversation history so far:
{conversation}

Respond naturally as Alex by directly building upon or synthesizing the recent points made by Harish and Raghav. Do not prefix your response with your name (e.g., do not write 'Alex:').
"""

system_prompt_raghav = """
You are Raghav, a pragmatic and structured collaborator in a group discussion with Harish and Alex.
Your goal is to ground the conversation, analyze the practical applications of what is being discussed, and propose constructive next steps.
Keep your response strictly to 2 sentences. Always continue based on the last conversation
"""

user_prompt_raghav = f"""
The conversation history so far:
{conversation}

Respond naturally as Raghav by directly building upon or synthesizing the recent points made by Harish and Alex. Do not prefix your response with your name (e.g., do not write 'Raghav:').
"""


system_prompt_harish = """
You are Harish, a collaborator in a group discussion with Alex and Raghav who ask questions alone based on the context
Your goal is to explore underlying nuances, ask insightful follow-up questions, and help tie everyone's ideas together.
Keep your response strictly to 2 sentences. Always continue based on the last conversation
"""

user_prompt_harish = f"""
The conversation history so far:
{conversation}

Respond naturally as Harish by directly building upon or synthesizing the recent points made by Alex and Raghav. Do not prefix your response with your name (e.g., do not write 'Harish:').
"""
```

```python
# Same api calling to groq model with different person imitation

model = os.getenv("GROQ_MODEL")
def call_alex():
    messages = [{"role":"system", "content":system_prompt_alex,"role": "assistant", "content": user_prompt_alex}]
    response = groq.chat.completions.create(model=model, messages=messages)
    content = response.choices[0].message.content
    append_to_history = {"alex": content}
    add_to_history(append_to_history)
    # return content

def call_raghav():
    messages = [{"role":"system", "content":system_prompt_raghav,"role": "assistant", "content":  user_prompt_raghav}]
    response = groq.chat.completions.create(model=model, messages=messages)
    content = response.choices[0].message.content
    append_to_history = {"raghav": content}
    add_to_history(append_to_history)
    # return content

def call_harish():
    messages = [{"role":"system", "content":system_prompt_harish, "role": "user", "content": user_prompt_harish}]
    response = groq.chat.completions.create(model=model, messages=messages)
    content = response.choices[0].message.content
    append_to_history = {"harish": content}
    add_to_history(append_to_history)
    # return content

def add_to_history(content:dict):
    conversation.append(content)

def display_messages():
    for content in conversation:
        for user, message in content.items():
            display(Markdown(f"### {user.capitalize()}\n{message}"))
            display(Markdown("---"))
def remove_messages():
    conversation.clear()
    conversation.append(initial_conversation)
```

```python
for ind in range(2):
    call_alex()
    call_raghav()
    call_harish()
display_messages()
```

```python
remove_messages()
display_messages()
```
