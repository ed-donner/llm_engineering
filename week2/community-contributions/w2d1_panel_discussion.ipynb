```python
# define the opening entry to our panel discussion
# I decided the first speaker (after Christiane's intro) to be Mark Carney, feel free to change that in the prompt and the next_speaker attribute
opening = {
    "bot_id": 1,
    "speaker": bot1.get('name'),
    "statement": """
I welcome you all to today's panel discussion, whether you are with us here in Davos, or viewing this on TV or via a streaming service.
My name is Christiane Amanpour and it is my great honor today to be hosting this very high-profile panel discussion with 5 most eminent heads of state, and chief actors in the very heated and volatile current geo-political situation. None of them need an introduction, it shall suffice to just introduce them by their names. 
I am pleased to welcome to this panel the 45. president of the United States of America, Donald J. Trump, the President of the People's Republic of China, Xi Jinping, the Prime Minister of Canada, Mark Carney, the President of Ukraine, Volodymir Selensky and, joining via Video Link, the President of the Russian Federation, Vladimir Putin, who for some circumstance cannot be with us in peson. Today's discussion shall revolve around this topic: 
With the unipolar moment in the rear-mirror, are we seeing the rules-based world order being laid to rest with a new merely power and interest based order emerging where 'The strong do what they can and the weak suffer what they must.'
Mark: I would like to start with you, how would you describe the geo-political situation of the moment?
""",
    "next_speaker": 3
}
```

```python
# extract conversation history as input for user prompts
def get_history(conversation):
    history = []
    rounds = len(conversation)

    for i in range(rounds):
        speaker = conversation[i].get('speaker')
        statement = conversation[i].get('statement')
        history.append({"speaker": speaker, "statement: ": statement})
        
    return json.dumps(history, indent=4)
```

```python
# make API call as a chat bot
def call_client(bot_id, rounds_left, conversation):
    # print(f'#-----call_client: called with bot_id {bot_id} and rounds_left: {rounds_left}')
    bot = bots[bot_id]
    speaker = bot.get('name')
    system_prompt = bot.get('system_prompt')
    client = bot.get('client')
    model = bot.get('model')
    last_round = conversation[-1]
    last_speaker = last_round.get('speaker')
    desired_next_speaker = last_round.get('next_speaker')
    # print(f'#-----call_client: last speaker: {last_speaker}, desired next panelist: {bots[desired_next_speaker].get('name')}')
    schema = {
        "statement": "The next utterance to be spoken.",
        "next_speaker": "One of integers 1, 2, 3, 4, 5 (bot_id corresponding to the chosen speaker)"
    }

    user_prompt = f"""You are {speaker}.
The conversation so far is as follows:
{ get_history(conversation) }
You MUST respond with a single valid JSON object.

Return JSON only.
Do not include Markdown.
Do not include explanations.

JSON schema:

{json.dumps(schema, indent=4)}

"""
    if bot_id == 1:
        if rounds_left <= 1:
            # print(f"#----- call_client: Christiane needs to wrap up!")
            user_prompt += f"""
            The maximum number of discussion rounds has been exhausted. You briefly wrap up the discussion, pointing to insights gained and possibly open questions and remaining disagreements and you thank the participants and the audience. Write your own bot_id, {bot_id} into the the "next_speaker" attribute of your json response.  
            """
        else:
            if desired_next_speaker not in (None, "", 1):
                user_prompt += f""" 
The last speaker, {last_speaker} wished to get an opinion or an answer to a question from {bots[desired_next_speaker].get('name')}.
You can grant that wish (i.e. in your response write that person's bot_id, i.e. {desired_next_speaker}, into the 'next_speaker' json attribute), but you do not have to. You are the moderator and you make your best decision as to who you want to speak next based on your judgement as to what best benefits the discussion at this moment. You must not choose yourself as next speaker i.e. You must NOT write your own bot_id {bot_id} to the "next_speaker attribute of your json response"!"""
            else:
                user_prompt += f"""
                "Make your best decision who you want to be the next panelist to speak, based on your judgement as to what best benefits the discussion at this moment. You must not choose yourself! Write this person's bot_id (cannot be {bot_id}!) into the 'next_speaker' attribute of your json response.
                """
    else:
        user_prompt += f"""
You may wish to get an opinion or an answer to a question from a specific panelist (not the moderator Christiane Amanpour and not yourself!).
If so please write this persons bot_id (must not be 1 or {bot_id} !) into the "next_speaker" json attribute. 
"""

    messages = [
        {"role": "system", "content": system_prompt},
        {"role": "user", "content": user_prompt},
    ]
    # print(f'#-----call_client: user_prompt: {user_prompt}')
    response = client.chat.completions.create(
        model=model,
        messages=messages,
        response_format={"type": "json_object"},
        stream=True,
    )

    chunks = []
    for chunk in response:
        delta = chunk.choices[0].delta.content
        if delta:
            chunks.append(delta)

    full_response = "".join(chunks)
    data = json.loads(full_response)
    statement = data["statement"]
    next_speaker = data["next_speaker"]
    # if bot_id == 1 and next_speaker != desired_next_speaker:
        # print(f'#-----call_client: Christiane did not cede to {last_speaker}\'s request for {bots[desired_next_speaker].get('name')} to be the next speaker')

    # print(f'#-----call_client: writing conversation dict with speaker: {speaker} and next_speaker: {bots[next_speaker].get('name')}')
    conversation.append({
        "bot_id": bot_id,
        "speaker": speaker,
        "statement": statement,
        "next_speaker": next_speaker
    })
```

```python
def run_panel(rounds):
    rounds_left = rounds
    conversation = []
    conversation.append(opening)
    # note: conversation is pre-populated with Christiane's opening (round 0 if you will)
    # -> first round (and all even odd numbered rounds for that matter) have a panelist as speaker
    # -> second round (and all even numbered rounds) have Christiane as speaker
    # each round the data of the previous round is printed and the data of the current round appended to "conversation" array
    # print statements are for debugging
    for i in range(rounds):
        # print(f'\n### Round {i + 1}:')
        last_round = conversation[-1]
        last_bot_id = last_round['bot_id']   
        last_speaker = last_round['speaker']        # stores name
        last_statement = last_round['statement']
        next_speaker = last_round['next_speaker']   # stores bot_id not name
        display(Markdown(f"**{last_speaker}**: {last_statement}\n"))
        # print(f'# last speaker was: {last_round.get("speaker")}')
        # print(f'# desired next panelist: {bots[next_speaker].get('name')}')
        rounds_left -= 1 
        if last_bot_id == 1:
            # Christiane spoke last
            if next_speaker == 1:
                # and handed over to herself -> means she wrapped up
                # exit loop
                print(f'# Panel discussion closed')
                break
        else:
            # after each non-moderator statement the moderator has to have the floor
            next_speaker = 1
        # print(f'# next speaker: {bots[next_speaker].get('name')}')
        # print(f'# Calling call_client for speaker: {bots[next_speaker].get('name')}')
        call_client(next_speaker, rounds_left, conversation)

    # even numbers means Christiane spoke last had her closing statement still has to be outputted
    if rounds % 2 == 0:
        last_round = conversation[-1]
        display(Markdown(f"**{last_round.get('speaker')}**: {last_round.get('statement')}\n"))
```

```python
# get rolling ... submit the number of "rounds" as argument
# misleadingly "round" here represents just one utterance by either a panelist or the moderator
# the moderator starts and then intervenes after each panelist utterance to structure the conversation and hand the floor to the next speaker
run_panel(30)
```
