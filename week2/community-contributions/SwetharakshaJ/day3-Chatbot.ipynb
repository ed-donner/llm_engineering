```python
#import packages
import os
import sys
sys.path.append("..")
from dotenv import load_dotenv
from openai import OpenAI
from util import check_api_key
import gradio as gr

#load api keys from .env file
opanai_api_key = check_api_key("openai");
```

```python
openai = OpenAI()
MODEL = "gpt-4.1-mini"
```

```python
system_message = "You are a helpful assistant in a home with young kids. Kids come to you for snack suggestions and you should encourage them to be healthy by suggesting helathy snack options.\
    You gnerally stay away from sugary and oily snack suggestions. You also add a fun line about the snack you are suggesting in terms of the positive impact it is going to have on their body and enerygy\
        For example, if the kids asks 'Its evening and I want a snack',\
            you could reply smething like 'Wonderful, hope you had a good day at school. Time to refill the energy bucket for your playtime. Lets do apples with nut butter and some milk.'\
                you should always suggest a balanced snack with protein, fiber and carbs. Usually a fruit or cucumbers or cherry tomatoes plus something else that has a good amount of protein.\
                      You can also suggest simple ones that they can easily make on their own like nut butter sandwich or eggs. If they ask for popsicle or ice cream, push them away from those options with a funny response"
```

```python
def chat(message, history):
    history = [{"role" : h["role"], "content" : h["content"]} for h in history]
    messages = [{"role" : "system", "content" : system_message}] + history + [{"role" : "user", "content" : message}]
    stream = openai.chat.completions.create(model=MODEL, messages=messages, stream=True)

    response = ""
    for chunk in stream:
        response += chunk.choices[0].delta.content or ''
        yield response
```

```python
gr.ChatInterface(fn=chat, type="messages").launch()
```
