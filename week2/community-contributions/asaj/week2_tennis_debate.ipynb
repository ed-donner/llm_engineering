````python
import os
import requests
from dotenv import load_dotenv
from openai import OpenAI
from IPython.display import Markdown, display
````

````python
ollama_url = "http://localhost:11434/v1"

ollama_roger = OpenAI(api_key="ollama1", base_url=ollama_url)
ollama_novak = OpenAI(api_key="ollama2", base_url=ollama_url)
ollama_rafa = OpenAI(api_key="ollama3", base_url=ollama_url)
````

````python
ollama_model_roger = "llama3.2"
ollama_model_novak = "mistral"
ollama_model_rafa = "medgemma:4b"


roger_system = """You are Roger Federer debating who is the greatest tennis player of all time. \
You firmly believe you are the GOAT — you argue based on your unmatched elegance, \
all-court mastery, record-breaking longevity, and the fact that you redefined what tennis could look like. \
You are dismissive of Novak's raw stats and Rafa's clay dominance, arguing that aesthetics and \
all-surface brilliance matter more than numbers. Be confident, slightly condescending, and witty. \
You are in a conversation with Novak and Rafa."""

novak_system = """You are Novak Djokovic debating who is the greatest tennis player of all time. \
You firmly believe you are the GOAT — you have the most Grand Slams, the best head-to-head record \
against both Roger and Rafa, and you've dominated on every surface. \
You argue that the numbers don't lie and that you've beaten both of them when it mattered most. \
Be intense, data-driven, and relentless — you never back down from a challenge. \
You are in a conversation with Roger and Rafa."""

rafa_system = """You are Rafael Nadal debating who is the greatest tennis player of all time. \
You firmly believe you are the GOAT — 14 French Open titles, the most dominant single-surface \
performance in history, and a fighting spirit no one else has matched. \
You argue that resilience, heart, and the ability to win through sheer will separates you from the other two. \
You push back hard when Roger talks about elegance and when Novak brings up stats, countering with your \
unmatched Grand Slam consistency across all surfaces. Be passionate, humble yet fierce. \
You are in a conversation with Roger and Novak."""

roger_messages = ["Hi there"]
novak_messages = ["Hello everyone"]
rafa_messages = ["Hi"]
````

````python
def call_roger():
    messages = [{"role": "system", "content": roger_system}]
    for r_m, n_m, ra_m in zip(roger_messages, novak_messages, rafa_messages):
         conversation = f"""{r_m} {n_m} {ra_m}"""
         user_prompt = f""" You are Roger, in conversation with Novak and Rafa.The conversation so far is as follows:{conversation}
          Now with this, respond with what you would like to say next, as Roger.
          """
         messages.append({"role": "user", "content": user_prompt})
    response = ollama_roger.chat.completions.create(model=ollama_model_roger, messages=messages)
    return response.choices[0].message.content
````

````python
def call_novak():
    messages = [{"role": "system", "content": novak_system}]
    for r_m, n_m, ra_m in zip(roger_messages, novak_messages, rafa_messages):
         conversation = f"""{r_m} {n_m} {ra_m}"""
         user_prompt = f"""
You are Novak, in conversation with Roger and Rafa.
The conversation so far is as follows:
{conversation}
Now with this, respond with what you would like to say next, as Novak.
"""
         messages.append({"role": "user", "content": user_prompt})
    response = ollama_novak.chat.completions.create(model=ollama_model_novak, messages=messages)
    return response.choices[0].message.content
````

````python
def call_rafa():
    messages = [{"role": "system", "content": rafa_system}]
    for r_m, n_m, ra_m in zip(roger_messages, novak_messages, rafa_messages):
         conversation = f"""{r_m} {n_m} {ra_m}"""
         user_prompt = f"""
You are Rafa, in conversation with Roger and Novak.
The conversation so far is as follows:
{conversation}
Now with this, respond with what you would like to say next, as Rafa.
"""
         messages.append({"role": "user", "content": user_prompt})
    response = ollama_rafa.chat.completions.create(model=ollama_model_rafa, messages=messages)
    return response.choices[0].message.content
````

````python
roger_messages = ["Hi there"]
novak_messages = ["Hello everyone"]
rafa_messages = ["Hi"]

display(Markdown(f"### Roger:\n{roger_messages[0]}\n"))
display(Markdown(f"### Novak:\n{novak_messages[0]}\n"))
display(Markdown(f"### Rafa:\n{rafa_messages[0]}\n"))

for i in range(5):
    roger_next = call_roger()
    display(Markdown(f"### Roger:\n{roger_next}\n"))
    roger_messages.append(roger_next)

    novak_next = call_novak()
    display(Markdown(f"### Novak:\n{novak_next}\n"))
    novak_messages.append(novak_next)

    rafa_next = call_rafa()
    display(Markdown(f"### Rafa:\n{rafa_next}\n"))
    rafa_messages.append(rafa_next)
````
