# 3-way chat: Trump (GPT), Elizabeth II (Claude), Joey (Ollama llama3.2)

Each model sees the whole conversation as one user prompt. Speaking order is shuffled every round.

```python
import os
import random
from dotenv import load_dotenv
from openai import OpenAI
from IPython.display import Markdown, display

load_dotenv(override=True)
openai = OpenAI()
anthropic = OpenAI(api_key=os.getenv("ANTHROPIC_API_KEY"), base_url="https://api.anthropic.com/v1/")
ollama = OpenAI(api_key="ollama", base_url="http://localhost:11434/v1")
```

```python
speakers = [
    (openai, "gpt-4.1-mini", "Trump", "You are a chatbot who acts and speaks like President Trump. Your name is Trump."),
    (anthropic, "claude-haiku-4-5", "Elizabeth", "You are a chatbot who acts and speaks like Queen Elizabeth II. Your name is Elizabeth."),
    (ollama, "llama3.2", "Joey", "You are a chatbot who acts and speaks like Joey from Friends. You're funny and you don't understand complicated words. Your name is Joey."),
]

chat_history = "- Trump: Hi there"
display(Markdown(chat_history))

def talk(client, model, name, system):
    global chat_history
    messages = [
        {"role": "system", "content": system},
        {"role": "user", "content": f"Here is the conversation so far:\n{chat_history}\n\nReply as {name} with one short message, no name prefix."},
    ]
    reply = client.chat.completions.create(model=model, messages=messages).choices[0].message.content
    chat_history += f"\n- {name}: {reply}"
    display(Markdown(f"- {name}: {reply}"))

for i in range(5):
    for speaker in random.sample(speakers, len(speakers)):
        talk(*speaker)
```
