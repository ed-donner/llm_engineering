```python
import os
import requests
from dotenv import load_dotenv
from anthropic import Anthropic
from openai import OpenAI
from IPython.display import Markdown, display
```

```python
load_dotenv(override=True)
openai_api_key = os.getenv('OPENAI_API_KEY')
anthropic_api_key = os.getenv('ANTHROPIC_API_KEY')
ollama_url = "http://localhost:11434/v1"
anthropic_url = "https://api.anthropic.com/v1/"
```

```python
openai = OpenAI()

anthropic = Anthropic(api_key=anthropic_api_key)
ollama = OpenAI(api_key="ollama",base_url=ollama_url)


gpt_model="gpt-5-nano"
anthropic_model = "claude-haiku-4-5-20251001"
ollama_model = "llama3.2:1b"
```

```python
# message tempalte [{"role":"user","content":"message details"}] the messages are
#a list of dicts or jsons

gpt_sys_prompt = "you are an optimistic assistant"

claude_sys_prompt ="you are annoyong and irrational"

llama_sys_prompt ="you are drunk"

gpt_messages =["Hello friends"]

claude_messages =["Hi"]

llama_messages =["wassup"]
```

```python

from anthropic import Anthropic

client = Anthropic(
    api_key=anthropic_api_key,  # This is the default and can be omitted
)
page = client.beta.models.list()
page = page.data[0]
print(page.id)
```

```python
def call_gpt():
    messages = [{"role":"system","content":gpt_sys_prompt}]
    for gpt, claude, llama in zip(gpt_messages,claude_messages,llama_messages):
        messages.append({"role":"assistant","content":gpt})
        messages.append({"role":"user","content":claude})
        messages.append({"role":"user","content":llama})
    response = openai.responses.create(
        model=gpt_model,
        input=messages
    )
    return response.output_text
```

```python
def call_claude():
    messages = []
    for gpt, claude, llama in zip(gpt_messages,claude_messages,llama_messages):
            messages.append({"role":"user","content":gpt})
            messages.append({"role":"assistant","content":claude})
            messages.append({"role":"user","content":llama})
    response = anthropic.messages.create(
            model=anthropic_model,
            system=claude_sys_prompt,
            messages=messages,
            max_tokens=1000
    )
    return response.content[0].text
```

```python
def call_llama():
    messages = [{"role":"system","content":llama_sys_prompt}]
    for gpt, claude, llama in zip(gpt_messages,claude_messages,llama_messages):
            messages.append({"role":"user","content":gpt})
            messages.append({"role":"user","content":claude})
            messages.append({"role":"assistant","content":llama})
    response = ollama.responses.create(
            model=ollama_model,
            input=messages,
            
    )
    return response.output_text
```

```python
call_gpt()
```

```python

for i in range(2):
    gpt_resp=call_gpt()
    print("gpt:",gpt_resp)
    gpt_messages.append(gpt_resp)
    claude_resp=call_claude()
    print("calude:",claude_resp)
    claude_messages.append(claude_resp)
    llama_resp=call_llama()
    print("ollama:",llama_resp)
    llama_messages.append(llama_resp)
```
