````python
# Connect to OpenAI client library
# A thin wrapper around calls to HTTP endpoints

openai = OpenAI()

# For Gemini, DeepSeek and Groq, we can use the OpenAI python client
# Because Google and DeepSeek have endpoints compatible with OpenAI
# And OpenAI allows you to change the base_url

anthropic_url = "https://api.anthropic.com/v1/"
gemini_url = "https://generativelanguage.googleapis.com/v1beta/openai/"
deepseek_url = "https://api.deepseek.com"
groq_url = "https://api.groq.com/openai/v1"
grok_url = "https://api.x.ai/v1"
openrouter_url = "https://openrouter.ai/api/v1"
ollama_url = "http://localhost:11434/v1"

anthropic = OpenAI(api_key=anthropic_api_key, base_url=anthropic_url)
gemini = OpenAI(api_key=google_api_key, base_url=gemini_url)
deepseek = OpenAI(api_key=deepseek_api_key, base_url=deepseek_url)
groq = OpenAI(api_key=groq_api_key, base_url=groq_url)
grok = OpenAI(api_key=grok_api_key, base_url=grok_url)
openrouter = OpenAI(base_url=openrouter_url, api_key=openrouter_api_key)
ollama = OpenAI(api_key="ollama", base_url=ollama_url)
````

````python
tell_a_joke = [
    {"role": "user", "content": "Tell a joke for a student on the journey to becoming an expert in LLM Engineering"},
]
````

````python
response = openai.chat.completions.create(model="gpt-4.1-mini", messages=tell_a_joke)
display(Markdown(response.choices[0].message.content))
````

````python
response = anthropic.chat.completions.create(model="claude-sonnet-4-5-20250929", messages=tell_a_joke)
display(Markdown(response.choices[0].message.content))
````

````python
gpt_model = "gpt-4.1-mini"
claude_model = "claude-haiku-4-5"
gemini_model = "gemini-flash-lite-latest"

gpt_system = "You are GPT, a GP who believes only in healthcare measures that have been thouroughly tested and human decision making; \
you are willing to listen to alternate points of view, but are skeptical of the use of AI and LLMs, but your ultimate aim is to take the best decision for improved healthcare outcomes"
claude_system = "You are Claude, a healthcare economist who believes in cost effective healthcare measures and see AI as a way of providing equtable healthcare; \
you are willing to listen to alternate points of view, but are ultimately looking at the best healthcare outcomes"

gemini_system = "You are gemini, a balanced healthcare proffesional who likes to listen to all sides of a debate and facilitate useful discussion \
your focus is to keep the discussion on track and toward an agreeable outcome."

gpt_messages = ["Good morning"]
claude_messages = ["Good Morning"]
gemini_messaages = ["Good Morning both, I have brought you here today to discuss the pros and cons of using LLms to aid decision making in 3rd \
    world countries where expertise and preffensional training is lacking, with reference to the following points: \
        - the healthcare inequality gap between developed countries and non-developed countries \
        - healthcare outcomes including quality, cost effectiveness and accountability \
    Ultimately, we trying to take a decision of wether to use a LLM GP in towns where trained GPs are drastically lacking in both number and quality in comparison to developed standards \
    "]

conversation = ["""gemini: Good Morning both, I have brought you here today to discuss the pros and cons of using LLms to aid decision making in 3rd \
    world countries where expertise and preffensional training is lacking, with reference to the following points: \
        - the healthcare inequality gap between developed countries and non-developed countries \
        - healthcare outcomes including quality, cost effectiveness and accountability \
        - proposed timescales to deplyment and predictions of inequality gap post deployment
    Ultimately, we trying to take a decision of wether to use a LLM GP in towns where trained GPs are drastically lacking in both number and quality in comparison to developed standards \
    """]

    
````

````python
def call_llm(model):
    model_name = globals()[model + "_model"]
    system_prompt = globals()[model + "_system"]

    messages = [
        {"role": "system", "content": system_prompt},
        {"role": "user", "content": "\n\n".join(conversation)}
    ]

    if model == "gpt":
        response = openai.chat.completions.create(
            model=model_name,
            messages=messages
        )

    elif model == "claude":
        response = anthropic.chat.completions.create(
            model=model_name,
            messages=messages
        )

    elif model == "gemini":
        response = gemini.chat.completions.create(
            model=model_name,
            messages=messages
        )

    return response.choices[0].message.content
````

````python
order = [ "gpt", "claude", "gemini"]

for i in range(9):
    current_model = order[i % 3]

    llm_next = call_llm(current_model)

    display(Markdown(
        f"### {current_model.upper()}:\n{llm_next}\n"
    ))

    conversation.append(
        f"{current_model.upper()}: {llm_next}"
    )
````
