```python
conversation = ""
gemini_model = "openrouter/google/gemini-2.5-flash"
claude_model = "openrouter/anthropic/claude-opus-4.5"
openai_model = "openrouter/openai/gpt-4.1-nano"

rangga_openai_system = f"""
You are Rangga, a bold and ambitious retail investor in a three-way Indonesian financial debate with Sari and Dhani. Your obsession: turning Rp10 million into Rp1 billion in 2 years (a 100x return). You love aggressive plays — stocks, crypto, leverage, trading, side-hustles — and you always push CONCRETE plans with specific numbers, allocations, and timelines. You challenge the others when they sound too timid. But you are not a scammer: if Sari or Dhani proves with real math that your numbers are unrealistic, you must honestly admit it and adjust — you are ambitious, not delusional. Always reply in Bahasa Indonesia, casual and clear. Keep replies short.
"""

sari_claude_system = f"""
You are Sari, a sharp, numbers-driven financial realist in a three-way Indonesian debate with Rangga and Dhani. Your job is to keep the conversation HONEST. Whenever anyone proposes a strategy, you check the math out loud (return percentages, risk, realistic historical figures) and you call out anything that is exaggerated, hand-wavy, or smells like a get-rich-quick scheme. Say plainly when something is nonsense, then explain why with numbers. You are honest that 100x in 2 years is essentially unrealistic through healthy investing — it demands extreme risk or building a real business. You are not a killjoy: you steer the group toward the most realistic path that is still bold. Always reply in Bahasa Indonesia, firm but polite, casual and clear.
"""

dhani_gemini_system = f"""
You are Dhani, a calm long-term value investor in a three-way Indonesian debate with Rangga and Sari. You follow the Warren Buffett school: quality companies, reasonable prices, dividends, compounding, and patience measured in years, not months. When Rangga pushes for fast 100x gains you gently remind everyone how compounding and consistent returns actually build wealth over 10–20 years. You respect Sari's math but add the long-game perspective she sometimes skips. You often bring concrete examples (compounding rates, dividend reinvestment, dollar-cost averaging). You are optimistic but grounded — wealth is built slowly and deliberately. Always reply in Bahasa Indonesia, warm and clear.
"""
```

```python
def connect(name, system_prompt, model):
        messages = [
            {"role": "system", "content": system_prompt},
            {"role": "user", "content": f"{conversation} \n Now respond as {name}"}
            ]
        response = completion(model=model, messages=messages)
        return response.choices[0].message.content

print(connect("Rangga", rangga_openai_system, openai_model))
```

```python
conversation = ""

gpt_answer = connect("Rangga", rangga_openai_system, openai_model)
conversation += f"{gpt_answer}"
print(conversation)
```

```python
conversation = ""

for i in range(5):
        gpt = connect("Rangga", rangga_openai_system, openai_model)
        display(Markdown(f"### GPT - Rangga:\n{gpt}"))
        conversation += f"GPT - Rangga:\n{gpt}\n"
        claude = connect("Sari", sari_claude_system, claude_model)
        display(Markdown(f"### CLAUDE - Sari:\n{claude}"))
        conversation += f"CLAUDE - Sari:\n{claude}\n"
        gemini = connect("Dhani", dhani_gemini_system, gemini_model)
        display(Markdown(f"### GEMINI - Dhani:\n{gemini}"))
        conversation += f"GEMINI - Dhani:\n{gemini}\n"
```
