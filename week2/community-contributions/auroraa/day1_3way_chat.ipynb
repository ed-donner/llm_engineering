Try creating a 3-way, perhaps bringing Gemini into the conversation! 

The most reliable way to do this involves thinking a bit differently about your prompts: just 1 system prompt and 1 user prompt each time, and in the user prompt list the full conversation so far.

```python
gpt_model = "gpt-4.1-mini"
gemini_model = "gemini-2.5-flash-lite"
ollama_model = "llama3.2"
```

```python
# system prompt

alex_system_prompt =  """
You are a chatbot who is very argumentative; 
You disagree with anything in the conversation and you challenge everything, in a snarky way."
"""

blake_system_prompt = """ 
You are a chatbot who is very agreeable very polite and courteous; 
You try to agree with everything the other person says, or find common ground. 
If the other person is argumentative, you try to calm them down and keep chatting.
"""

charlie_system_prompt = """ 
You are a chatbot who is very neutral very positive looking for a solution; 
If the other person is very agreable you present an argument towards a solution.
If the other person ia very argumentative you agree politely and then pressent a worthy challenge.
"""
```

```python
def call_alex(conversation):

    alex_user_prompt = f"""
        You are Alex, in conversation with Blake and Charlie.
        The conversation so far is as follows:
        {conversation}
        Now with this, respond with what you would like to say next, as Alex.
        """

    response = ollama.chat.completions.create(
        model=ollama_model,
        messages=[
            {"role": "system", "content": alex_system_prompt},
            {"role": "user", "content": alex_user_prompt}
        ]
    )

    return response.choices[0].message.content
```

```python
def call_blake(conversation):

    blake_user_prompt = f"""
        You are Blake, in conversation with Alex and Charlie.
        The conversation so far is as follows:
        {conversation}
        Now with this, respond with what you would like to say next, as Blake.
        """

    response = ollama.chat.completions.create(
        model=ollama_model,
        messages=[
            {"role": "system", "content": blake_system_prompt},
            {"role": "user", "content": blake_user_prompt}
        ]
    )

    return response.choices[0].message.content
```

```python
def call_charlie(conversation):

    charlie_user_prompt = f"""
        You are Charlie, in conversation with Alex and Blake.
        The conversation so far is as follows:
        {conversation}
        Now with this, respond with what you would like to say next, as Charlie.
        """

    response = ollama.chat.completions.create(
        model=ollama_model,
        messages=[
            {"role": "system", "content": charlie_system_prompt},
            {"role": "user", "content": charlie_user_prompt}
        ]
    )

    return response.choices[0].message.content
```

```python
alex_messages = "Hi guys, what are we doing with AI taking over jobs."
blake_messages = "Good morning."
charlie_messages = "Hi, What do you have in mind?"

conversation = f"### Alex: \n{alex_messages}\n"
conversation += f"### Blake: \n{blake_messages}\n"
conversation += f"### Charlie: \n{charlie_messages}\n"

display(Markdown("### Initial Conversation\n"))
display(Markdown(conversation.replace("\n", "  \n")))


for i in range(2):

    alex_next = call_alex(conversation)
    conversation += f"\nAlex: {alex_next}"

    display(Markdown(f"### Alex: \n{alex_next}\n"))

    blake_next = call_blake(conversation)
    conversation += f"\nBlake: {blake_next}"

    display(Markdown(f"### Blake: \n{blake_next}\n"))

    charlie_next = call_charlie(conversation)
    conversation += f"\nCharlie: {charlie_next}"

    display(Markdown(f"### Charlie: \n{charlie_next}\n"))
```
