# Week 2 Day 5: Create DB for Food Service Menu

```python
import sqlite3
import pandas as pd

# Connect to the local database
conn = sqlite3.connect("foodservice_menu.db")
cursor = conn.cursor()

# Drop the old table to apply the new schema
cursor.execute("DROP TABLE IF EXISTS menu_items")

# Create the updated menu table including descriptions
cursor.execute("""
CREATE TABLE menu_items (
    id INTEGER PRIMARY KEY AUTOINCREMENT,
    food_service TEXT NOT NULL,
    category TEXT NOT NULL,
    item_name TEXT NOT NULL,
    description TEXT,
    price REAL NOT NULL
)
""")

# Insert typical menu data with descriptions and single prices
sample_data = [
    ("food truck", "Appetizers", "Garlic Bread", "Toasted baguette slices topped with garlic butter and herbs.", 5.25),
    ("food truck", "Appetizers", "Buffalo Wings", "Spicy chicken wings served with celery sticks and blue cheese dip.", 10.75),
    ("food truck", "Main Course", "Tacos", "A platter of 3 tacos with your choice of protein, served with a side of rice and beans.", 14.00),
    ("food truck", "Main Course", "Margherita Pizza", "Fresh mozzarella, tomato sauce, and basil leaves on a crispy crust.", 14.25),
    ("food truck", "Main Course", "Grilled Salmon", "Atlantic salmon fillet served with asparagus and lemon butter sauce.", 21.99),
    ("food truck", "Desserts", "Chocolate Lava Cake", "Warm chocolate cake with a molten center, served with vanilla ice cream.", 7.75),
    ("food truck", "Beverages", "Fresh Soda", "Chilled, sparkling fountain soda with natural flavorings.", 3.00),
    ("restaurant", "Appetizers", "Garlic Bread", "Toasted baguette slices topped with garlic butter and herbs.", 5.25),
    ("restaurant", "Appetizers", "Buffalo Wings", "Spicy chicken wings served with celery sticks and blue cheese dip.", 20.75),
    ("restaurant", "Main Course", "Tacos", "A platter of 3 tacos with your choice of protein, served with a side of rice and beans.", 14.00),
    ("restaurant", "Main Course", "Meditteranean Pizza", "Fresh mozzarella, tomato sauce, and basil leaves on a crispy crust.", 24.25),
    ("restaurant", "Main Course", "Grilled Salmon", "Atlantic salmon fillet served with asparagus and lemon butter sauce.", 31.99),
    ("restaurant", "Desserts", "Chocolate Lava Cake", "Warm chocolate cake with a molten center, served with vanilla ice cream.", 17.75),
    ("restaurant", "Beverages", "Fresh Soda", "Chilled, sparkling fountain soda with natural flavorings.", 13.00)
]

cursor.executemany("""
INSERT INTO menu_items (food_service, category, item_name, description, price)
VALUES (?, ?, ?, ?, ?)
""", sample_data)

conn.commit()

# Query and display the data cleanly in your notebook
query = "SELECT food_service, category, item_name, description, price FROM menu_items"
df = pd.read_sql_query(query, conn)

# Close connection
conn.close()

# Format the price column as currency
df['price'] = df['price'].map('${:,.2f}'.format)

# Display the final table
df
```
