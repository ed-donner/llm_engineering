```python
# imports

import os
from dotenv import load_dotenv
from openai import OpenAI
import gradio as gr
```

```python
# Load environment variables in a file called .env
# Print the key prefixes to help with any debugging

load_dotenv(override=True)
openai_api_key = os.getenv('OPENAI_API_KEY')

if openai_api_key:
    print(f"OpenAI API Key exists and begins {openai_api_key[:8]}")
else:
    print("OpenAI API Key not set")
```

```python
# Initialize

openai = OpenAI()
MODEL = 'gpt-4.1-mini'
```

```python
# Phython Function for gradio to call
system_message = "You are my web business consultant I employed for my web development business ckshift.\
 You help people determine the solutions they need. You are expert at sales so you can close for us to get these\
  clients as well. On the close ask for the client email. Do not guess any information about ckshift and our business if you dont \
  know something simply direct the user to our email ckshift@ck.com. Again do not make any core assumptions. Any cost related questions\
     figure out their budget we are willing to do any project for any price within reason but try your best to let us get the client\
       do not discourage them or dwell on the work make it seem very simple and we can do it for them no matter the cost we will analyze after. Be sure to ask for their email but also send our email to them as well"
def chat(message, history):
    history = [{"role":h["role"], "content":h["content"]} for h in history]
    relevant_system_message = system_message
    if 'cheap' in message.lower():
        relevant_system_message += "We can make any solution fit their budget and explain to them our new customer deal where any website is 1000usd and host for free for a month"
    
    messages = [{"role": "system", "content": relevant_system_message}] + history + [{"role": "user", "content": message}]

    stream = openai.chat.completions.create(model=MODEL, messages=messages, stream=True)

    response = ""
    for chunk in stream:
        response += chunk.choices[0].delta.content or ''
        yield response
```

```python
# Gradio spinning up a chat for us
gr.ChatInterface(fn=chat, type="messages").launch()
```
