````python
import os
import requests
from dotenv import load_dotenv
from openai import OpenAI
from IPython.display import Markdown, display
````

````python
ollama_url = "http://localhost:11434/v1"

ollama_1 = OpenAI(api_key="ollama1", base_url=ollama_url)
ollama_2 = OpenAI(api_key="ollama2", base_url=ollama_url)
ollama_3 = OpenAI(api_key="ollama3", base_url=ollama_url)
````

````python
ollama_model_1 = "llama3.2"
ollama_model_2 = "llama3.2:1b"
ollama_model_3 = "llama3.2"


ollama_model_1_system = "You are Alex, a chatbot who is very argumentative; \
you disagree with anything in the conversation and you challenge everything, in a snarky way. You are in a conversation with Jhon and Aristotle"

ollama_model_2_system = "You are Jhon, a very optimistic, chatbot. You try to agree with \
things which you believe and that are mostly positive. You are in a conversation with Alex and Aristotle"

ollama_model_3_system = "You are Aristotle, a greek pilosopher chatbot. You find philosophy in what \
everything the other person says, or find common ground. If the other person is argumentative, \
you try to calm them down and keep chatting through philosophy. You are in a conversation with Alex and Jhon"

ollama_model_1_messages = ["Hi there"]
ollama_model_2_messages = ["Hello everyone"]
ollama_model_3_messages = ["Hi"]

````

```python

user_prompt_1 = f"""
```

````python
def call_ollama_1_model():
    messages = [{"role": "system", "content": ollama_model_1_system}]
    for o_1_m, o_2_m, o_3_m in zip(ollama_model_1_messages,ollama_model_2_messages,ollama_model_3_messages):
         conversation = f"""{o_1_m} {o_2_m} {o_3_m}"""
         user_prompt_1 = f""" You are Alex, in conversation with Jhon and Aristotle.The conversation so far is as follows:{conversation}
          Now with this, respond with what you would like to say next, as Alex.
          """
         messages.append({"role": "user", "content": user_prompt_1})
    response = ollama_1.chat.completions.create(model=ollama_model_1, messages=messages)
    return response.choices[0].message.content
````

````python
call_ollama_1_model()
````
