# Synthetic Healthcare Data Factory

## Project Overview

This project uses LLM to generate synthetic healthcare data.

Synthetic data is artificial data that looks similar to real-world data does not belong to real patients.

We will use different AI models and propmts to generate structured healthcare datasets that can be used for data science, analytics and AI projects.

For this project, we will start with synthetic Prior Authorization data.

## Business Problem

Healthcare data is useful for building and testing data science and AI applications.

However, real patient data contains sensitive information and cannot always be easily accessed or shared.

Synthetic data provides a way to create realistic examples without using real patient information.

Instead of manually creating hundreds of records, we can use LLMs to generate structured synthetic datasets automatically.

## Project Objective

The goal of this project is to build a simple synthetic data generator that can:

- Describe the type of data we want to generate
- Define the structure of the dataset
- Generate synthetic records using an LLM
- Use different prompts to create diverse data
- Compare data generated by different AI models
- Check the quality of the generated data
- Save the final dataset for future projects

All patient information generated in this project will be fictional.

## Our First Dataset

We will start with a Prior Authorization dataset.

Prior authorization is a process where a healthcare provider may need approval from an insurance company before certain healthcare services are provided.

Each synthetic record will represent a fictional prior authorization request.

Later, the same approach can be used to generate other types of healthcare data.

````python
dataset_name = "Prior Authorization"

fields = [
    "patient_id",
    "age",
    "gender",
    "insurance_provider",
    "service",
    "diagnosis_category",
    "physician_order",
    "clinical_notes",
    "previous_imaging",
    "authorization_status"
]

print("Dataset:", dataset_name)
print("\nFields:")

for field in fields:
    print("-", field)
````

We created a simple definition for our first synthetic dataset.

The dataset represents fictional prior authorization requests.

The fields describe information such as the patient's age, insurance provider, requested service, available documents and authorization status.

Defining the fields first helps us clearly tell the LLM what type of structured data we want it to generate.

## Step 2: Imports and Model Setup

````python
# Step 1: Importing Libraries
import os
import json
import pandas as pd
from dotenv import load_dotenv
from openai import OpenAI
````

````python
# Checking API Key is loaded or not
load_dotenv(override=True)

openai_api_key = os.getenv("OPENAI_API_KEY")

if openai_api_key:
    print("OpenAI API Key loaded successfully")
else:
    print("OpenAI API Key not found")
````

````python
# OpenAI Client and selecting the first model that will use to generate synthetic data
openai = OpenAI()
MODEL = "gpt-4.1-mini"

print("Model:", MODEL)
````

## Step 3: Generate Our First Synthetic Healthcare Records.

## Generate Our First Synthetic Data

Now we will use an LLM to generate our first synthetic Prior Authorization records.

We will start with only 5 records so we can easily review the output.

We will tell the model:
- What type of data we need
- Which fields to include
- That the patients must be fictional
- That the output should be structured as JSON

Later, we will improve the prompt and generate larger datasets.

````python
prompt = """
Generate 5 fictional healthcare Prior Authorization records.

Use these fields:

- patient_id
- age
- gender
- insurance_provider
- service
- diagnosis_category
- physician_order
- clinical_notes
- previous_imaging
- authorization_status

Requirements:

- Do not use real patient information.
- Use different patients, services and insurance providers.
- Use realistic but fictional healthcare examples.
- physician_order, clinical_notes and previous_imaging should be true or false.
- authorization_status should be Approved, Pending or Denied.
- Return only a JSON array.
"""
````

````python
# Lets send the prompt to our first model
response = openai.chat.completions.create(
    model=MODEL,
    messages=[{"role": "user","content": prompt}]
)

generated_data = response.choices[0].message.content

print(generated_data)
````

## Step 4: Convert the LLM Output into Structured Data.

## Convert the Generated Data

The LLM returned our synthetic records as text.

Before we can analyze the data, we need to convert the text into structured Python data.

Sometimes an LLM adds Markdown formatting such as ```json around the response.

We will first remove this formatting and then convert the JSON text into Python.

````python
cleaned_data = generated_data.strip()

cleaned_data = cleaned_data.replace("```json", "")
cleaned_data = cleaned_data.replace("```", "")
cleaned_data = cleaned_data.strip()

print(cleaned_data)
````

We removed the extra Markdown formatting from the LLM response.

````python
# The data is still JSON text, but it is now ready to be converted into Python.
records = json.loads(cleaned_data)

print("Number of records:", len(records))
print("Data type:", type(records))
````

````python
# Convert into Dataframe
df = pd.DataFrame(records)
df
````

We converted the synthetic records into a Pandas DataFrame.

This gives us a structured table where each row represents one fictional Prior Authorization request and each column represents a field.

The data is now much easier to view, analyze and validate.

## Step 5: Improving the Prompt for More Diverse Synthetic Data

## Improving Data Diversity

Our first prompt successfully generated structured synthetic data.

However, when generating larger datasets, we do not want every record to look similar.

We can improve the prompt by asking the model to create more variety in areas such as:

- Patient age
- Gender
- Insurance provider
- Healthcare service
- Diagnosis category
- Available documents
- Authorization status

This helps us create a more diverse dataset for testing and data science projects.

````python
diversity_prompt = """
Generate 5 diverse and fictional healthcare Prior Authorization records.

Use these fields:

- patient_id
- age
- gender
- insurance_provider
- service
- diagnosis_category
- physician_order
- clinical_notes
- previous_imaging
- authorization_status

Requirements:

- Do not use real patient information.
- Include patients from different age groups.
- Use different genders.
- Use different fictional insurance providers.
- Include different healthcare services.
- Include different diagnosis categories.
- Create different combinations of available and missing documents.
- Include a mix of Approved, Pending and Denied authorization statuses.
- Avoid repeating the same type of record.
- Return only a JSON array.
"""
````

This prompt gives the LLM more instructions about diversity.

Instead of simply asking for different records, we describe the types of variation we want in the dataset.

This gives us more control over the synthetic data being generated.

````python
response = openai.chat.completions.create(
    model=MODEL,
    messages=[{"role": "user","content": diversity_prompt}]
)

diverse_data = response.choices[0].message.content

print(diverse_data)
````

We generated a second set of synthetic records using the diversity-focused prompt.

We used the same AI model as before.

This allows us to see how changing only the prompt can influence the type and variety of data generated by the model.

````python
# Clean the JSON strips, like we did earlier
cleaned_diverse_data = diverse_data.strip()

cleaned_diverse_data = cleaned_diverse_data.replace("```json", "")
cleaned_diverse_data = cleaned_diverse_data.replace("```", "")
cleaned_diverse_data = cleaned_diverse_data.strip()

diverse_records = json.loads(cleaned_diverse_data)
````

````python
diverse_df = pd.DataFrame(diverse_records)
diverse_df
````

````python
# compare it with our original dataset to see whether the diversity-focused prompt created more variation.
print("Original Dataset")
print("Unique services:", df["service"].nunique())
print("Unique diagnosis categories:", df["diagnosis_category"].nunique())
print("Authorization statuses:", df["authorization_status"].unique())

print("\nDiversity Prompt Dataset")
print("Unique services:", diverse_df["service"].nunique())
print("Unique diagnosis categories:", diverse_df["diagnosis_category"].nunique())
print("Authorization statuses:", diverse_df["authorization_status"].unique())
````

### Observation

Both prompts generated 5 unique services, 5 unique diagnosis categories and all three authorization statuses.

In this small example, the original prompt already produced diverse records, so the diversity-focused prompt did not show an improvement in these measures.

Later, we will generate more records and use different prompts and models to make a better comparison.

## Step 6: Generate Edge Cases

Real-world datasets do not contain only simple and complete records.

They can also contain cases with missing documents, denied requests, incomplete information or unusual combinations.

We will use a different prompt to intentionally generate these types of edge cases.

This can help create better datasets for testing AI and data science applications.

````python
# Third Prompt
edge_case_prompt = """
Generate 5 fictional healthcare Prior Authorization edge-case records.

Use these fields:

- patient_id
- age
- gender
- insurance_provider
- service
- diagnosis_category
- physician_order
- clinical_notes
- previous_imaging
- authorization_status

Create examples that include situations such as:

- Missing physician orders
- Missing clinical notes
- Missing previous imaging
- Denied authorization requests
- Pending authorization requests
- Different healthcare services and diagnosis categories

Requirements:

- Do not use real patient information.
- Make each record different.
- physician_order, clinical_notes and previous_imaging should be true or false.
- authorization_status should be Approved, Pending or Denied.
- Return only a JSON array.
"""
````

This prompt focuses specifically on edge cases. Instead of asking the model for general records, we are asking it to create examples with missing documents and different authorization outcomes.

These records can be useful when testing whether another system can correctly handle unusual or incomplete cases.

````python
response = openai.chat.completions.create(
    model=MODEL,
    messages=[{"role": "user","content": edge_case_prompt}]
)

edge_case_data = response.choices[0].message.content

print(edge_case_data)
````

The model generated a new set of synthetic records using our edge-case prompt.

````python
# Convert Records into Pandas Dataframe for easy to review.
cleaned_edge_data = edge_case_data.strip()

cleaned_edge_data = cleaned_edge_data.replace("```json", "")
cleaned_edge_data = cleaned_edge_data.replace("```", "")
cleaned_edge_data = cleaned_edge_data.strip()

edge_records = json.loads(cleaned_edge_data)

edge_df = pd.DataFrame(edge_records)

edge_df
````

````python
# let's measure what the edge-case prompt produced
print("Total records:", len(edge_df))

print("\nMissing Physician Order:")
print((edge_df["physician_order"] == False).sum())

print("\nMissing Clinical Notes:")
print((edge_df["clinical_notes"] == False).sum())

print("\nMissing Previous Imaging:")
print((edge_df["previous_imaging"] == False).sum())

print("\nAuthorization Status:")
print(edge_df["authorization_status"].value_counts())
````

### Observation

The edge-case prompt intentionally creates records with missing information and different authorization outcomes.

This shows how different prompts can be used to generate specific types of synthetic data.

Instead of relying on one prompt to generate every type of record, we can use different prompts to create normal, diverse and edge-case examples.

Standard Prompt
      ↓
General synthetic records

Diversity Prompt
      ↓
Different services, diagnoses and patient characteristics

Edge-Case Prompt
      ↓

## Create a Resuable Function

## Create a Reusable Data Generator

So far, we have repeated similar code each time we generate synthetic data.

We will now create a simple function that can:

1. Receive a prompt
2. Send the prompt to an AI model
3. Get the generated response
4. Extract the JSON data
5. Convert it into a Pandas DataFrame

This will make it easier to generate more datasets using different models and prompts.

````python
def generate_synthetic_data(client, model, prompt):

    response = client.chat.completions.create(
        model=model,
        messages=[{"role": "user","content": prompt}])

    generated_text = response.choices[0].message.content

    start = generated_text.find("[")
    end = generated_text.rfind("]")

    json_text = generated_text[start:end + 1]

    records = json.loads(json_text)
    return pd.DataFrame(records)
````

The function combines the main steps of our synthetic data generation process.

We provide:

- The AI client
- The model
- The prompt

The function sends the request to the model, finds the JSON array in the response and converts the generated records into a Pandas DataFrame.

This allows us to reuse the same process with different models and prompts.

````python
# Testing GPT Model
gpt_df = generate_synthetic_data(openai,MODEL,prompt)
gpt_df
````

````python
# Testing LLAMA Model
llama_df = generate_synthetic_data(ollama,OLLAMA_MODEL,prompt)
llama_df
````

## Generate a Larger Synthetic Dataset

So far, we have generated small groups of 5 records to test our prompts and models.

We will now generate multiple batches of synthetic Prior Authorization records using different prompt strategies.

We will start by using GPT-4.1-mini because it generated structured data successfully and quickly.

Later, we can combine records generated by different models.

````python
standard_batch = generate_synthetic_data(openai,MODEL,prompt)

diversity_batch = generate_synthetic_data(openai,MODEL,diversity_prompt)

edge_batch = generate_synthetic_data(openai,MODEL,edge_case_prompt)

print("Standard records:", len(standard_batch))
print("Diversity records:", len(diversity_batch))
print("Edge-case records:", len(edge_batch))
````

We generated three different batches of synthetic data.

Each batch was created for a different purpose:

- **Standard** creates general Prior Authorization records.
- **Diversity** encourages variety across patients, services and diagnoses.
- **Edge Case** intentionally creates more difficult or incomplete examples.

Next, we will label each batch so that we know which prompt generated each record.

````python
standard_batch["prompt_type"] = "Standard"
diversity_batch["prompt_type"] = "Diversity"
edge_batch["prompt_type"] = "Edge Case"

standard_batch["model"] = MODEL
diversity_batch["model"] = MODEL
edge_batch["model"] = MODEL
````

We added two metadata columns:

- `prompt_type` tells us which prompt strategy generated the record.
- `model` tells us which AI model generated the record.

Keeping this information allows us to analyze and compare the synthetic data later.

````python
larger_df = pd.concat([standard_batch,diversity_batch,edge_batch], ignore_index=True)
larger_df
````

````python
print("Total records:", len(larger_df))
print("Total columns:", len(larger_df.columns))

print("\nRecords by prompt:")
print(larger_df["prompt_type"].value_counts())
````

## Data Quality Check

Before using synthetic data, we should check whether the generated records contain missing values, duplicates or invalid values.

````python
print("Total records:", len(larger_df))
print("Missing values:", larger_df.isnull().sum().sum())
print("Duplicate records:", larger_df.duplicated().sum())
print("Duplicate patient IDs:", larger_df["patient_id"].duplicated().sum())

invalid_ages = larger_df[
    (larger_df["age"] < 0) | (larger_df["age"] > 120)
]

print("Invalid ages:", len(invalid_ages))
````

````python
# Check Authorization Status
valid_statuses = ["Approved", "Pending", "Denied"]

invalid_statuses = larger_df[~larger_df["authorization_status"].isin(valid_statuses)
]

print(larger_df["authorization_status"].value_counts())

print("\nInvalid statuses:", len(invalid_statuses))
````
