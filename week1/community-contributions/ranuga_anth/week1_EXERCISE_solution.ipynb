```python
# imports
import os
import json
from dotenv import load_dotenv
from IPython.display import Markdown, display, update_display
from openai import OpenAI
```

```python
# constants

MODEL_GPT = 'gpt-4o-mini'
MODEL_LLAMA = 'llama3.2:1b'
OLLAMA_BASE_URL = "http://localhost:11434/v1"

openai = OpenAI()
ollama = OpenAI(base_url=OLLAMA_BASE_URL, api_key="ollama")
```

```python
# set up environment
load_dotenv()
api_key = os.getenv('OPENAI_API_KEY')

if api_key and api_key.startswith('sk-proj-') and len(api_key)>10:
    print("API key looks good so far")
else:
    print("There might be a problem with your API key? Please check the api key again!")
```

```python
user_question = input("Please enter your technical question: ")
```

```python
# here is the question; type over this to ask something new

system_prompt = """
You are a helpful assistant that analyzes the technical question of the user and provide a 
comprehensive, detailed, accurate, well structured, and easy-to-understand answer. 
You should provide the answer in a step-by-step manner, breaking down complex concepts into simpler terms. 
Use examples, analogies and simple scenarios where appropriate to enhance understanding. 
Ensure that your response is thorough and addresses all aspects of the user's question.
"""
```

```python
user_prompt = f"""
You are looking at the following technical question of the user: 
{user_question}
"""
```

```python
# Get gpt-4o-mini to answer, with streaming

def stream_gpt_answer_to_technical_question(system_prompt, user_prompt):
    stream = openai.chat.completions.create(
        model=MODEL_GPT,
        messages=[
            {"role": "system", "content": system_prompt},
            {"role": "user", "content": user_prompt}
        ],
        stream=True
    )
    reponse = ""
    display_handle = display(Markdown(""), display_id=True)
    for chunk in stream:
        reponse += chunk.choices[0].delta.content or ""
        update_display(Markdown(reponse), display_id=display_handle.display_id)
```

```python
stream_gpt_answer_to_technical_question(system_prompt, user_prompt)
```

```python
# Get Llama 3.2 to answer

def stream_llama_answer_to_technical_question(system_prompt, user_prompt):
    stream = ollama.chat.completions.create(
        model=MODEL_LLAMA,
        messages=[
            {"role": "system", "content": system_prompt},
            {"role": "user", "content": user_prompt}
        ],
        stream=True
    )
    reponse = ""
    display_handle = display(Markdown(""), display_id=True)
    for chunk in stream:
        reponse += chunk.choices[0].delta.content or ""
        update_display(Markdown(reponse), display_id=display_handle.display_id)
```

```python
stream_llama_answer_to_technical_question(system_prompt, user_prompt)
```
