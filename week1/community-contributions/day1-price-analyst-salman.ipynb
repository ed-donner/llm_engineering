# Day 1 Exercise - Price Analyst

A small variation on the Day 1 exercise: instead of summarizing a website, we build a system prompt that turns GPT into a "price analyst" - give it an asset (e.g. `BTC`, `LINKUSDT`, `GOLD`, `SILVER`) and it responds with the price.

```python
# imports

import os
from dotenv import load_dotenv
from openai import OpenAI
```

```python
# Load environment variables and connect to OpenAI

load_dotenv(override=True)
api_key = os.getenv('OPENAI_API_KEY')

if not api_key:
    print("No API key was found - please check your .env file")
else:
    print("API key found and looks good so far!")

openai = OpenAI()
```

```python
# Step 1: Create your prompts

system_prompt = """
You are a price analyst. Whenever I give you the name of an asset,
respond with its current price. The asset could be a cryptocurrency
(e.g. BTC, LINKUSDT), a precious metal (e.g. GOLD, SILVER), a stock,
or any other tradeable asset.
"""

asset = "BTC"

user_prompt = f"""
Tell me the price of {asset}
"""
```

```python
# Step 2: Make the messages list

messages = [
    {"role": "system", "content": system_prompt},
    {"role": "user", "content": user_prompt}
]
```

```python
# Step 3: Call OpenAI

response = openai.chat.completions.create(model="gpt-4.1-mini", messages=messages)
```

```python
# Step 4: print the result

print(response.choices[0].message.content)
```
