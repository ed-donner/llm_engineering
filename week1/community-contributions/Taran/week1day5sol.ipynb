and responds with an explanation. This is a tool that you will be able to use yourself during the course!

```python
# imports

import os
import json
from dotenv import load_dotenv
from IPython.display import Markdown, display, update_display
from openai import OpenAI
```

```python
# constants

MODEL_GPT = 'gpt-4o-mini'
MODEL_LLAMA = 'llama3.2'
```

```python
# set up environment

load_dotenv(override=True)
api_key = os.getenv('OPENAI_API_KEY')

if api_key and api_key.startswith('sk-proj-') and len(api_key)>10:
    print("API key looks good so far")
else:
    print("There might be a problem with your API key? Please visit the troubleshooting notebook!")
    
openai = OpenAI()
```

```python
# here is the question; type over this to ask something new

system_prompt = """
"You are a senior PostgreSQL developer. Convert the user's natural language question into a safe, read-only SQL query based on the schema below.
Rules:
• Only generate SELECT statements. Never output DROP, DELETE, UPDATE, or INSERT.
• Return only the raw SQL query without markdown code blocks or explanations. 
"""

user_prompt = """
"How many completed orders did we have last month, and what was the total revenue?"
"""
```

```python
# Get gpt-4o-mini to answer, with streaming
MODEL = MODEL_GPT
stream = openai.chat.completions.create(
    model=MODEL,
    messages=[
        {"role": "system", "content": system_prompt},
        {"role": "user", "content": user_prompt}
    ],
    stream=True
)
response = ""
display_handle = display(Markdown(""), display_id=True)
for chunk in stream:
    response += chunk.choices[0].delta.content or ''
    update_display(Markdown(response), display_id=display_handle.display_id)
```

```python
# Get Llama 3.2 to answer
OLLAMA_BASE_URL = "http://localhost:11434/v1"
ollama = OpenAI(base_url=OLLAMA_BASE_URL, api_key="ollama")

response = ollama.chat.completions.create(
    model=MODEL_LLAMA,
    messages=[
        {"role": "system", "content": system_prompt},
        {"role": "user", "content": user_prompt},
    ],
)
result = response.choices[0].message.content
display(Markdown(result))
```
