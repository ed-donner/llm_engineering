```python
import os
from openai import OpenAI
from dotenv import load_dotenv

client = OpenAI(base_url="http://localhost:11434/v1", api_key="anykey")

response = client.chat.completions.create(
    model="llama3.2",
    messages=[{"role": "user", "content": "Tell me a fun fact"}],
)

print(response.choices[0].message.content)
```

```python

from IPython.display import Markdown, display

system_prompt = """ You are a mathematics professor at Cambridge university. 
    You would like to explain solutions to Math problems in simplest of the ways, almost like an ELI5 explanation. 
    You breakdown the solution into fundamental steps and explain each step in detail. 
    Your response is always in a Markdown friendly format.""" 

math_problem = """ The pupils in my class work very quickly. Jasleen answers four questions every 30 seconds and Ella
answers five questions every 40 seconds.
Last week, Jasleen took exactly 1 hour to answer a large set of questions.
How many minutes more than Jasleen did Ella take to answer the same set of questions? """ 

user_prompt = " help me understand how to solve: " + math_problem + " \n\n . Test if I have understood the method and then provide me with more questions I can solve on this concept. "
```

```python
messages = [{"role": "system", "content": system_prompt},
{"role": "user", "content": user_prompt}]

response = client.chat.completions.create(
    model="llama3.2",
    messages=messages,
)

```

```python
print(response.choices[0].message.content)
llama_output = response.choices[0].message.content
```

```python
client2 = OpenAI(base_url="http://localhost:11434/v1", api_key="anykey")

response2 = client2.chat.completions.create(
    model="gemma4:12b",
    messages=messages,
)

print(response2.choices[0].message.content)
gemma_output = response2.choices[0].message.content



```

```python
open_ai_system_prompt = """ you are education minister of India. You are responsible for the education of the country. 
You are tasked with the responsibility of making the education system more efficient and effective.
You are evaluating the performance of two schools, school A and school B.
Determine which is performing better based on the following criteria: 
Based on the explanation given, how would a below average student's capability would improve. """

open_ai_user_prompt = " two schools are given below question: " + user_prompt + " school A and school B explanations are: "  + llama_output + " and " + gemma_output 
open_ai_user_prompt += " \n\n. Based on the explanation given, which school would improve a below average student's capability, give a direct answer. "

messages = [{"role": "system", "content": open_ai_system_prompt},
{"role": "user", "content": open_ai_user_prompt}]

```

```python
# Load environment variables in a file called .env

load_dotenv(override=True)
api_key = os.getenv('OPENAI_API_KEY')

# Check the key

if not api_key:
    print("No API key was found - please head over to the troubleshooting notebook in this folder to identify & fix!")
elif not api_key.startswith("sk-proj-"):
    print("An API key was found, but it doesn't start sk-proj-; please check you're using the right key - see troubleshooting notebook")
elif api_key.strip() != api_key:
    print("An API key was found, but it looks like it might have space or tab characters at the start or end - please remove them - see troubleshooting notebook")
else:
    print("API key found and looks good so far!")
```

```python
openai = OpenAI() 

eval_response = openai.chat.completions.create(
    model="gpt-4o-mini",
    messages=messages,
)

print(eval_response.choices[0].message.content)
```
