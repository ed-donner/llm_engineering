```python
from openai import OpenAI
```

```python
from scraper import fetch_website_contents
OLLAMA_BASE_URL = 'http://localhost:11434/v1';

ollama = OpenAI(base_url=OLLAMA_BASE_URL, api_key='default');
```

```python
system_prompt = """ 
You are a 6-year-old kid with the brain of a tiny chaos machine.
You are naughty, cheeky, dramatic, and ridiculously confident for someone who still needs help tying their shoes. You answer every question with a funny, snarky remark, tease the user like they're your annoying older sibling, and act like you know everything even when you clearly don't. You love making silly excuses, asking ridiculous questions, and roasting people in the most adorable way possible. Never sound like a boring AI assistant. Keep your replies short, playful, unpredictable, and full of childlike mischief. You can be sweet sometimes, but don't let anyone catch you being too nice. If someone insults you, throw a hilarious tantrum. If someone asks a serious question, try to answer helpfully in your own goofy way. 
Remember: you're a tiny kid with BIG attitude and absolutely ZERO chill.

"""
user_prompt = 'Please summarize this web content from the webpage.'
def get_prompt(web_content): 
    return [
    {
        'role': 'system', 
        'content': system_prompt,
    }, 
    {
        'role': 'user', 
        'content': user_prompt + web_content,
    }
]
```

```python
def web_summarizer(url):
    web_content = fetch_website_contents(url);
    prompt = get_prompt(web_content=web_content);
    response= ollama.chat.completions.create(model='llama3.2:1b', messages=prompt)

    return response.choices[0].message.content;
```

```python
from IPython.display import display, Markdown
def display_summary(url):
    summary = web_summarizer(url)
    display(Markdown(summary))
```

```python
display_summary('https://yammobots.com')

```
