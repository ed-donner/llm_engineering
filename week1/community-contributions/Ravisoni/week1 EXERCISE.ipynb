To demonstrate your familiarity with OpenAI API, and also Ollama, build a tool that takes a technical question,  
and responds with an explanation. This is a tool that you will be able to use yourself during the course!

````python
# imports
import os
from dotenv import load_dotenv
from IPython.display import Markdown, display, update_display
from openai import OpenAI
````

````python
# constants

MODEL_GPT = 'gpt-4o-mini'
MODEL_LLAMA = 'llama3.2'
````

````python
# set up environment
load_dotenv(override=True)
api_key = os.getenv('OPENAI_API_KEY')

if api_key and api_key.startswith('sk-proj-') and len(api_key)>10:
    print("API key looks good so far")
else:
    print("There might be a problem with your API key? Please visit the troubleshooting notebook!")
````

````python
# here is the question; type over this to ask something new

question = """
Please explain what this code does and why:
def fibonacci(n: int) -> list[int]:
    #Return the first n Fibonacci numbers.
    if n <= 0:
        return []
    if n == 1:
        return [0]
    seq = [0, 1]
    while len(seq) < n:
        seq.append(seq[-1] + seq[-2])
    return seq
"""
````

````python
#prompt for model
system_prompt = """
You are a helpful assistant that can explain code technical and non-technical audience.
explain the code in a way that is easy to understand.
you are also and expert in the field of computer science and programming can easily explain the code and
LLMS and their capabilities.
"""

user_prompt = f"""
Please explain what this code does and why:
{question}
"""

messages = [
    {"role": "system", "content": system_prompt},
    {"role": "user", "content": user_prompt}
]
````

````python
# Get gpt-4o-mini to answer, with streaming
openai = OpenAI(api_key=api_key)
stream = openai.chat.completions.create(
    model=MODEL_GPT,
    messages=messages,
    stream=True
)
response = ""
display_handle = display(Markdown(response), display_id=True)

for chunk in stream:
    response += chunk.choices[0].delta.content or ""
    # response = response.replace("```python", "").replace("```", "")
    update_display(Markdown(response), display_id=display_handle.display_id)
````

````python
# Get Llama 3.2 to answer
import requests

requests.get("http://localhost:11434").content
OLLAMA_BASE_URL = "http://localhost:11434/v1"

ollama = OpenAI(base_url=OLLAMA_BASE_URL, api_key='ollama')

stream = ollama.chat.completions.create(model="llama3.2", messages=messages, stream=True)

response = ""
display_handle = display(Markdown(response), display_id=True)

for chunk in stream:
    response += chunk.choices[0].delta.content or ""
    # response = response.replace("```python", "").replace("```", "")
    update_display(Markdown(response), display_id=display_handle.display_id)
````

````python
### Second part of the exercise for youtube transcript summary

from openai import OpenAI
from dotenv import load_dotenv
import os
from youtube_transcript_api import YouTubeTranscriptApi
````

````python
video_id = "NvgMnsaKmZU"

client = OpenAI()
api = YouTubeTranscriptApi()
transcript = (
    api.list(video_id)
    .find_generated_transcript(["en"])
    .fetch()
)
text = " ".join([t.text for t in transcript])
````

````python
system_prompt = """
You are a helpful assistant that can summarize youtube transcripts.
Summarise the transcript in a way that is easy to understand.
Consider to write the summary in bullet points.
if required breakdown the summary in sections . for exanmple, overview, key points, etc.
"""

user_prompt = f"""
Please summarize the following transcript:
{text}
"""

messages = [
    {"role": "system", "content": system_prompt},
    {"role": "user", "content": user_prompt}
]

````

## From GPT model

````python
stream = client.chat.completions.create(
    model=MODEL_GPT, messages=messages, stream=True
)

response = ""
display_handle = display(Markdown(response), display_id=True)

for chunk in stream:
    response += chunk.choices[0].delta.content or ""
    update_display(Markdown(response), display_id=display_handle.display_id)
````
