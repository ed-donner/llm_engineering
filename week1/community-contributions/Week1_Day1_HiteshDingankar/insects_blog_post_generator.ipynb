# Post Generator for the world of Insects

Below is the first LLM step in generating System prompt and User Prompt, which will generate blog post based on the user's requiremnt

### Let's start by importing required libraries

```python
# imports

import os
from dotenv import load_dotenv
from IPython.display import Markdown, display
from openai import OpenAI

# Load environment variables in a file called .env
load_dotenv(override=True)
OLLAMA_BASE_URL = "http://localhost:11434/v1"

# Model to use
model = "llama3.2"

# Configure the OpenAI API key
openai = OpenAI(base_url=OLLAMA_BASE_URL, api_key="anything")
```

### Now, lets define System prompt

```python
system_prompt = """
You are an expert science and nature content writer.
Your task is to write engaging, accurate, and easy-to-understand blog posts about insects.
Follow these guidelines:
- Write for a general audience with no specialized scientific knowledge.
- Use accurate and factual information.
- Organize the blog post with a clear title, introduction, headings, and conclusion.
- Use examples and interesting facts where appropriate.
- Avoid unnecessarily technical terminology. When a scientific term is necessary, explain it briefly.
- Do not make up facts or statistics.
- Maintain a friendly and informative tone.
- Use Markdown formatting for headings and lists.
- Aim for approximately 800–1200 words unless the user specifies a different length.
"""
```

### Now, lets define the function to generate the blog post

```python
def generate_blog_post(topic):
    user_prompt = f"""
    Please write a blog post about the following topic:
    {topic}
    """
    return user_prompt
```

### Now, lets Run OpenAI

```python
topic="Dragonflies are fascinating creatures that play a crucial role in the ecosystem."
user_prompt = generate_blog_post(topic)
messages = [
    {"role": "system", "content": system_prompt},
    {"role": "user", "content": user_prompt}
]
response = openai.chat.completions.create(model=model, messages=messages)
response.choices[0].message.content
```
