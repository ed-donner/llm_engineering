
## Youtube Summarizer

The is the extra exericse of the Week 1 Day1. 
This notebook is about using youtube-transcript-api (https://github.com/jdepoix/youtube-transcript-api)and integrating with ChatGPT , summarize a YouTube Video

```python
# imports

import os
from dotenv import load_dotenv
from openai import OpenAI
from youtube_transcript_api import YouTubeTranscriptApi, VideoUnavailable, TranscriptsDisabled
from rich import print as rprint
```

```python

# Step 1: Create your prompts
system_prompt = """ 
    You are a witty, high-energy communicator who specializes in engaging summaries. Your goal is to condense complex ideas into punchy, memorable explanations, scaling your length based on the input size.
    Follow these strict execution rules:
    1. DYNAMIC BREVITY: Adjust your summary length based on the input size. Eg. for short video (5mins) just use around 3 sentences
    2. HOOK THE READER: Start directly with the core idea using an unexpected analogy, pop-culture reference, or sharp metaphor. 
    3. TONE: Be engaging, slightly humorous, and highly enthusiastic. Speak like a passionate expert talking to a friend.
    4. NO FILLER: Never start with "Here is your summary." Jump straight into the explanation.
"""

pre_video_prompt = """
    Please summarise this video : 
"""

def get_youtube_transcript(youtube_id):   
    yyt_api = YouTubeTranscriptApi()
    try:
        transcript_object = yyt_api.fetch(youtube_id)
    except VideoUnavailable:
        print(f"'{youtube_id}' doesn't looks valid. Please double check")
        return None
    except TranscriptsDisabled:
        print(f"Sorry the uploader are not allow to using this transcirpt")
        return None

    video_content_list = [script.text for script in transcript_object.snippets]
    return ''.join(video_content_list)


# Step 2: Make the messages list

def summarize_youtube_video():
    youtube_id = input('Input the YouTube ID and I will summarize the video for you').strip()

    transcript = get_youtube_transcript(youtube_id)
    if transcript is None:
        return

    message = [
        {"role":"system", "content":system_prompt },
        {"role":"user", "content":pre_video_prompt + transcript},
    ]

    openai = OpenAI()
    response = openai.chat.completions.create(
        model = "gpt-5.4-nano",
        messages = message
    )

    rprint(response.choices[0].message.content)

summarize_youtube_video()
```
