# Email Subject Line Generator

A simple commercial use case: given the body of an email, suggest a concise, professional subject line.

This is the kind of feature that could be built into a commercial email tool like Gmail or Outlook.

```python
import os
from dotenv import load_dotenv
from openai import OpenAI
from IPython.display import Markdown, display
```

```python
load_dotenv(override=True)
openai = OpenAI()
```

```python
system_prompt = """You are an expert email assistant. Given the body of an email,
suggest a concise, professional subject line that captures the key point.
Respond with only the subject line, nothing else."""
```

```python
def suggest_subject(email_body):
    messages = [
        {"role": "system", "content": system_prompt},
        {"role": "user", "content": email_body}
    ]
    response = openai.chat.completions.create(
        model="gpt-4.1-nano",
        messages=messages
    )
    return response.choices[0].message.content
```

## Example 1: Budget reallocation email

```python
email_1 = """
Dear Team,

I wanted to follow up on our discussion from last Friday's meeting regarding the Q3 budget allocation.
After reviewing the numbers with the finance department, we've identified an opportunity to reallocate
approximately $50,000 from the marketing budget to the R&D department. This would allow us to accelerate
the development timeline for the new product launch by roughly two weeks.

Could we schedule a brief call this week to discuss the implications and get everyone's approval?
I'm available Tuesday afternoon or Wednesday morning.

Best regards,
Sarah
"""

subject = suggest_subject(email_1)
print(f"Suggested subject: {subject}")
```

## Example 2: Meeting reschedule email

```python
email_2 = """
Hi Mark,

Unfortunately, I need to reschedule our Thursday product demo. The client just informed us that
their CTO won't be available until next week. I've checked with the team and we can do either
Monday at 2pm or Tuesday at 10am next week.

Please let me know which works better for your side, and I'll confirm with the client.

Thanks,
Priya
"""

subject = suggest_subject(email_2)
print(f"Suggested subject: {subject}")
```

## Example 3: Project status update

```python
email_3 = """
Hello everyone,

Quick update on Project Phoenix: we've completed the database migration ahead of schedule.
All 2.3 million records have been transferred and validated. Performance tests show a 40%
improvement in query response times compared to the old system.

Next steps: we'll begin the API integration phase starting Monday. I'll send a detailed
timeline by end of day Friday.

Cheers,
Alex
"""

subject = suggest_subject(email_3)
print(f"Suggested subject: {subject}")
```

## Batch mode: process multiple emails at once

```python
emails = [email_1, email_2, email_3]

for i, email in enumerate(emails, 1):
    subject = suggest_subject(email)
    print(f"Email {i} -> {subject}")
```
