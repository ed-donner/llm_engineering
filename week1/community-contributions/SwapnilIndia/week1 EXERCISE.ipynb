```python
# imports
from openai import OpenAI
from dotenv import load_dotenv, find_dotenv
import os
```

```python
# constants
client = OpenAI()

MODEL_GPT = 'gpt-4o-mini'
MODEL_LLAMA = 'llama3.2'
```

```python
# set up environment
# Load .env from the project root (this notebook is nested)

load_dotenv(find_dotenv(), override=True)
api_key = os.getenv('OPENAI_API_KEY')

# Check the key

if not api_key:
    print("No API key was found - please head over to the troubleshooting notebook in this folder to identify & fix!")
elif not api_key.startswith("sk-proj-"):
    print("An API key was found, but it doesn't start sk-proj-; please check you're using the right key - see troubleshooting notebook")
elif api_key.strip() != api_key:
    print("An API key was found, but it looks like it might have space or tab characters at the start or end - please remove them - see troubleshooting notebook")
else:
    print("API key found and looks good so far!")
```

```python
# here is the question; type over this to ask something new

question = """
Please explain what this code does and why:
yield from {book.get("author") for book in books if book.get("author")}
"""
system_Prompt=" do deep analysis of question asked and answer in proper format wherever required make bold and identation should be redable answer in both English and Marathi from Vidharbha Region of Maharashtra India"
```

```python
def get_user_prompt(question) :
    user_prompt="do deep analysis of question and answer it in most easist way like explaining to 5 year old child answer should be short"
    user_prompt+=question
    return user_prompt
```

```python
# Get gpt-4o-mini to answer, with streaming
def getAnswerToQuestion(Que):
    response= client.chat.completions.create(
        model=MODEL_GPT,
        messages=[
           { "role":"system" ,"content":system_Prompt},
            {"role":"user", "content":get_user_prompt (Que)}

        ]
    )
    return response.choices[0].message.content
```

```python
getAnswerToQuestion('Why is melody is chocolaty')
```

```python
client2 =OpenAI(
    base_url="http://localhost:11434/v1",
    api_key="ollama"
)
```

```python
def getAotherModelUserPrompt(param):
    userPrompt="this user prompt param will take everything and convert it into Kannad language and give output in kannada language"
    userPrompt=userPrompt+getAnswerToQuestion(userPrompt)
    return userPrompt
```

```python
# Get Llama 3.2 to answer
def usingAnotherModel(param):
    response=client2.chat.completions.create(
        model="llama3.2:1b",
        messages=[
            { "role":"system", "content":system_Prompt },
            { "role":"user" ,"content":getAotherModelUserPrompt(param) }
        ]
    )
    return response.choices[0].message.content
    
```

```python
usingAnotherModel("what is 2+2 is")
```
