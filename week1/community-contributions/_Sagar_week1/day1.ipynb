# Week 1 - Day 1 Exercise: Email Subject Line Generator

Takes an email body and suggests an appropriate short subject line.

```python
import os
from dotenv import load_dotenv
from openai import OpenAI

load_dotenv(override=True)
openai = OpenAI()
```

```python
# Step 1: Create your prompts

def getEmailBody():
    return "Hi,I came across the Backend Golang Developer opening at Proofpoint (R14189 Link) and believe my experience aligns well with the role. I'm currently working as a backend-heavy Software Engineer with experience in Go, PostgreSQL, Redis, AWS, and distributed systems. I've worked on backend services, APIs, messaging systems, and production infrastructure.I'm very interested in this opportunity and would love to be considered for the position. I've attached my resume for your reference. Looking forward to hearing from you. Thanks!"

system_prompt = "I want you to work like email subject maker you will get the email and you need to suggest an appropiate short subject for that email"
user_prompt = f"Hi, this is the email: \n{getEmailBody()}"

# Step 2: Make the messages list

messages = [
    {"role": "system", "content": system_prompt},
    {"role": "user", "content": user_prompt}
]

# Step 3: Call OpenAI
response = openai.chat.completions.create(
    model = "gpt-6-luna",
    reasoning_effort = "none",
    messages = messages
)

# Step 4: print the result
print(response.choices[0].message.content)
```
