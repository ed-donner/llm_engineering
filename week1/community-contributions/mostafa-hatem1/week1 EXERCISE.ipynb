```python
# imports
import os
import json
from dotenv import load_dotenv
from IPython.display import Markdown, display, update_display
from openai import OpenAI
```

```python
# constants

MODEL_GEMINI = 'gemini-3.5-flash-lite'
MODEL_LLAMA = 'llama3.2'
```

```python
# set up environment

GEMINI_BASE_URL = "https://generativelanguage.googleapis.com/v1beta/openai/"
LLAMA_BASE_URL = "http://localhost:11434/v1"
load_dotenv(override=True)

google_api_key = os.getenv("GOOGLE_API_KEY")

if not google_api_key:
    print("No API key was found - please be sure to add your key to the .env file, and save the file!")
elif not google_api_key.startswith(("AIz", "AQ.")):
    print("An API key was found, but it doesn't start with AIz or AQ.")
else:
    print("API key found and looks good so far!")

gemini = OpenAI(base_url=GEMINI_BASE_URL, api_key=google_api_key)

ollama = OpenAI(base_url=LLAMA_BASE_URL, api_key="ollama")
```

```python
# here is the question; type over this to ask something new

question = """
Please explain the differences between Gemini 3.5 and LLaMA 3.2 in terms of architecture, performance, and typical use cases.
"""
system_prompt = """
You are a helpful and knowledgeable technical assistant for students learning to code.
When given a technical question or a snippet of code, you explain it thoroughly and clearly.

Your explanations should:
- State up front, in one sentence, what the code/concept does overall.
- Break down the key parts step by step (syntax, mechanics, any non-obvious behavior).
- Explain *why* it's written this way, not just *what* it does.
- Use concrete, simple language — avoid unnecessary jargon, and define any term you must use.
- Format your response in markdown, using code blocks for any code you reference.
"""
example_question = """
Please explain what this code does and why:
squares = [x**2 for x in range(10) if x % 2 == 0]
"""

example_answer = """
This line builds a list of the squares of all even numbers from 0 to 9, using a **list comprehension**.

**Step by step:**
- `range(10)` produces the numbers 0 through 9.
- `if x % 2 == 0` filters that range down to only the even numbers (0, 2, 4, 6, 8).
- `x**2` squares each of those surviving numbers.
- The square brackets `[...]` collect all the results into a new list.

**Why write it this way:** list comprehensions let you filter and transform a sequence in a single, readable expression, instead of writing a multi-line `for` loop with `.append()` calls. It's considered more "Pythonic" — concise and declarative rather than imperative.

Result: `squares = [0, 4, 16, 36, 64]`
"""

```

```python
def get_answer_gemini(question, system_prompt, example_question, example_answer):
    stream = gemini.chat.completions.create(
        model=MODEL_GEMINI,
        messages=[
            {"role": "system", "content": system_prompt},
            {"role": "user", "content": example_question},
            {"role": "assistant", "content": example_answer},
            {"role": "user", "content": question}
        ],
        stream=True
    )
    response = ""
    display_handle = display(Markdown(""), display_id=True)
    for chunk in stream:
        response += chunk.choices[0].delta.content or ''
        update_display(Markdown(response), display_id=display_handle.display_id)
    return response
```

```python
gemini_answer = get_answer_gemini(question, system_prompt, example_question, example_answer)
```
