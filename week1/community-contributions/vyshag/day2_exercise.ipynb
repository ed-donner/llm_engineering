```python


import requests
OLLAMA_BASE_URL = "http://localhost:11434/v1"

ollama = OpenAI(base_url=OLLAMA_BASE_URL, api_key='ollama')


requests.get("http://localhost:11434").content

system_prompt = "please analyse the given content and suggest the ideas with a title and a small description of it."
user_prompt = """
     i need you to extract the most useful services offered in this website and their real world oppurtunities.
"""

# See how this function creates exactly the format above

def messages_for(website):
    return [
        {"role": "system", "content": system_prompt},
        {"role": "user", "content": user_prompt + website}
    ]

# A function to display this nicely in the output, using markdown

def display_summary(url):
    summary = summarize(url)
    display(Markdown(summary))
# And now: call the OpenAI API. You will get very familiar with this!

def summarize(url):
    website = fetch_website_contents(url)
    response = ollama.chat.completions.create(
        model = "llama3.2:1b",
        messages = messages_for(website)
    )
    return response.choices[0].message.content



# messages = [{"role":"system","content":system_prompt},{"role":"user","content":user_prompt }] # fill this in

# # Step 3: Call OpenAI
# response =client.chat.completions.create(model = "openai/gpt-oss-20b",
#         messages = messages)

# Step 4: print the result
# print(response.choices[0].message.content)
display_summary("https://www.brototype.com/")
```
