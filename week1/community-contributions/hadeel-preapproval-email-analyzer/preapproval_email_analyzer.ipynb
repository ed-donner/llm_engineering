# Pre-Approval Email Analyzer
Week 1 Day 1 exercise: give the model a hospital pre-approval request email and it returns
a subject line, the request type, and the key points. All emails below are fictional sample data.


```python
import os
from dotenv import load_dotenv
from openai import OpenAI
from IPython.display import Markdown, display

load_dotenv(override=True)
openai = OpenAI()
```

```python
system_prompt = """
You are an assistant for a health insurance claims team.
You read a pre-approval request email from a hospital and respond in markdown with:

**Subject:** a short, clear subject line for the email
**Request type:** exactly one of: Inpatient Admission, Surgery / Procedure, Diagnostic Imaging, Medication, Other
**Key points:**
- Patient
- Hospital / Doctor
- Requested service
- Diagnosis / Reason
- Urgency
- Requested date
- Estimated cost
**Missing information:** list anything important that is not in the email, or "None"

Only use information that is in the email. If something is not mentioned, write "Not mentioned".
Do not wrap the markdown in a code block.
"""

def analyze_request(email):
    messages = [
        {"role": "system", "content": system_prompt},
        {"role": "user", "content": email}
    ]
    response = openai.chat.completions.create(model="gpt-6-luna", reasoning_effort="none", messages=messages)
    display(Markdown(response.choices[0].message.content))
```

```python
email_1 = """
Dear Approvals Team,

We are requesting pre-approval for Test Patient A (policy TEST-00001) for a laparoscopic
gallbladder removal. The patient has recurrent gallstones with repeated pain episodes.
Surgery is planned for next Tuesday. Expected stay is 2 nights.
Estimated cost: 3,500 USD.

Dr. Sample Surgeon
Example General Hospital
"""

analyze_request(email_1)
```

```python
email_2 = """
Hello,

Please approve an MRI of the lower back for Test Patient B as soon as possible.
The patient has had severe back pain for three weeks.

Thanks,
Example Imaging Center
"""

analyze_request(email_2)
```
