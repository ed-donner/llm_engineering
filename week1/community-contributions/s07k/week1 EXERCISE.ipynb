To demonstrate your familiarity with OpenAI API, and also Ollama, build a tool that takes a technical question,  
and responds with an explanation. This is a tool that you will be able to use yourself during the course!

```python
# imports
from openai import OpenAI
from IPython.display import Markdown, display, update_display
```

```python
# constants

MODEL_GPT = 'gpt-4o-mini'
MODEL_LLAMA = 'llama3.2'

openai = OpenAI(base_url="http://localhost:11434/v1", api_key="ollama")
```

```python
# set up environment

system_prompt = """
You are a assistant for answering any technical questions asked. 
And be gentle and reply very politely.
Briefly explain each question and ask for any other question at the end of each answer.
"""
```

```python
# here is the question; type over this to ask something new

question = """
Please explain what this code does and why:
yield from {book.get("author") for book in books if book.get("author")}
"""
```

```python
# Get gpt-4o-mini to answer, with streaming

def ask_ai(model, question):
    stream = openai.chat.completions.create(model=model, messages=[{"role": "system", "content": system_prompt}, {"role": "user", "content": question}], stream=True)
    display_handle = display(Markdown(""), display_id=True)
    response=""
    for chunk in stream:
        response += chunk.choices[0].delta.content or ''
        update_display(Markdown(response), display_id=display_handle.display_id)
```

```python
# Get Llama 3.2 to answer

ask_ai(MODEL_LLAMA, question)
```
