```python
from openai import OpenAI
```

Email Response Generator with OLLAMA

```python
OLLAMA_URL = "http://localhost:11434/v1/"
api_key="ollama"
model = "llama3"
open_ai = OpenAI(
    base_url=OLLAMA_URL,
    api_key=api_key
)
```

```python
email_system_prompt = """
You are a professional business email assistant. Analyze the incoming email, understand its intent, and write a clear, polite, and professional reply.
Use business logic when responding. Answer questions directly, acknowledge important points, and suggest a practical proposal or next step when appropriate.
Do not invent facts, commitments, dates, or information. If something is unclear, ask for clarification.
Return only the final email reply, ready to send."""
```

```python
def email_reply_user_prompt(sender_name, sender_email, email_subject, email_body):
    return f"""
Write a professional reply to this email.
Sender: {sender_name}
Email: {sender_email}
Subject: {email_subject}
Email:{email_body}
"""
```

```python

messages = [
    {"role": "system", "content": email_system_prompt},
    {"role": "user", "content": email_reply_user_prompt("John Doe", "john.doe@example.com", "Project Delivery Timeline", "Hi Jignesh, We reviewed the latest development build. There are still some issues with the reporting module. Can you confirm whether these can be resolved by Friday? Business Context: The development team has identified the issues but needs approximately 2 additional working days. Explain the realistic timeline and propose a practical delivery plan.")}
]

response = open_ai.chat.completions.create(
    model=model,
    messages=messages
)
print(response.choices[0].message.content)
```
