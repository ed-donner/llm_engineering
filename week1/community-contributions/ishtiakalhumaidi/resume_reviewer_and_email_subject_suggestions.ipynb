### PDF extractor function

```python
import re

def extract_pdf_text(pdf_path):
    reader = PdfReader(pdf_path)
    pages = []

    for page in reader.pages:
        text = page.extract_text()
        if text:
            pages.append(text)

    text = "\n".join(pages)

    text = re.sub(r'(?<!\n)\n(?!\n)', ' ', text)
    
    text = re.sub(r"[ \t]+", " ", text)

    text = re.sub(r"\n{3,}", "\n\n", text)

    return text.strip()
```

```python
resume_text = extract_pdf_text(r"C:\Users\ishti\Downloads\Ishtiak's Resume (1).pdf")

print(resume_text)
```

```python

system_prompt = """
You are an experienced technical recruiter reviewing software engineering resumes.

Analyze the resume and provide:
- Strengths
- Weaknesses
- Missing information
- Technical skill assessment
- Project quality
- Suggestions for improvement

Be specific and honest. Do not invent information.
Respond in markdown. do not wrap the markdown in a code block.
"""
```

```python
user_prompt = f"""
Review this resume:\n\n{resume_text}
"""
```

```python
def messages_for(system_prompt, user_prompt):
    return [
        {"role": "system", "content": system_prompt},
        {"role": "user", "content": user_prompt}
    ]
```

```python
messages_for(system_prompt, user_prompt)
```

```python
def resume_analyzer(system_prompt, user_prompt):

    response = ollama.chat.completions.create(
        model="llama3.2:3b",
        messages=messages_for(system_prompt, user_prompt)
    )

    return response.choices[0].message.content
```

```python
resume_analyzer(system_prompt, user_prompt)
```

```python
def display_resume_analysis(system_prompt, user_prompt):
    analysis = resume_analyzer(system_prompt, user_prompt)
    display(Markdown(analysis))
```

```python
display_resume_analysis(system_prompt, user_prompt)
```
