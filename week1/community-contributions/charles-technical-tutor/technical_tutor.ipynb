To demonstrate your familiarity with OpenAI API, and also Ollama, build a tool that takes a technical question,  
and responds with an explanation. This is a tool that you will be able to use yourself during the course!

```python
# imports

import os
from dotenv import load_dotenv
from openai import OpenAI
from IPython.display import display, Markdown

```

```python
# constants

MODEL_GPT = 'gpt-4o-mini'
MODEL_LLAMA = 'llama3.2:3b'
```

```python
# set up environment

load_dotenv()

# OpenAI API client
openai_client = OpenAI(
    api_key=os.getenv("OPENAI_API_KEY")
)

# Ollama local client
ollama = OpenAI(
    base_url="http://localhost:11434/v1",
    api_key="ollama"
)

```

```python
# here is the question; type over this to ask something new

question = """
Please explain what this code does and why:
yield from {book.get("author") for book in books if book.get("author")}
"""
```

```python
# Get gpt-4o-mini to answer, with streaming


from IPython.display import display, Markdown, clear_output

system_prompt = """
You are an experienced technical tutor.

Explain programming concepts accurately and in
simple English.

When explaining Python code:
1. Preserve the original syntax.
2. Explain each part step by step.
3. Identify Python concepts correctly.
4. Provide practical examples when useful.
5. Explain why the code works.
6. Acknowledge uncertainty when necessary.
"""

stream = openai_client.chat.completions.create(
    model=MODEL_GPT,
    messages=[
        {"role": "system", "content": system_prompt},
        {"role": "user", "content": question}
    ],
    stream=True
)

full_response = ""

for chunk in stream:
    content = chunk.choices[0].delta.content

    if content:
        full_response += content
        clear_output(wait=True)
        display(Markdown(full_response))
```

```python
# Get Llama 3.2 to answer


system_prompt = """
You are an experienced Python programming tutor.

Your responsibility is to provide accurate, clear,
and beginner-friendly technical explanations.

Follow these instructions:

1. Read the user's question carefully before answering.

2. When explaining Python code, preserve the exact
   syntax provided by the user.

3. Pay special attention to parentheses (), square
   brackets [], and curly braces {} because they
   can change the meaning of Python code.

4. Identify the Python concepts involved correctly.
   For example, distinguish generator expressions,
   list comprehensions, and set comprehensions.

5. Explain the code step by step in simple English.

6. Provide a practical, executable Python example
   when appropriate.

7. Explain what the example should output and why.

8. Do not invent details or silently modify the
   user's original code.

9. If you are uncertain about an explanation,
   acknowledge the uncertainty.

10. Before responding, check that your explanation
    is consistent with the original code.
"""

response = ollama.chat.completions.create(
    model=MODEL_LLAMA,
    messages=[
        {"role": "system", "content": system_prompt},
        {"role": "user", "content": question}
    ],
    temperature=0.2
)

answer = response.choices[0].message.content

display(Markdown(answer))
```
