```python
from openai import OpenAI

client = OpenAI()
```

```python
# Step 1: Create your prompts

system_prompt = """
You are a technical assistant for software developers.

Analyze the provided text and return:
1. A short summary
2. Important technical concepts
3. Technologies mentioned
4. Key takeaways for a software developer

Respond in Markdown.
"""

user_prompt = """
Web scraping is the process of extracting information from websites.

Traditional scraping can use Python libraries such as requests
and BeautifulSoup. However, some modern websites generate their
content dynamically using JavaScript.

For these websites, browser automation tools such as Selenium
or Playwright can be used. These tools launch or control a browser,
allow JavaScript to execute, and then allow developers to query
the rendered page.
"""
```

```python
# Step 2: Make the messages list

messages = [
    {"role": "system", "content": system_prompt},
    {"role": "user", "content": user_prompt}
]
```

```python
# Step 3: Call OpenAI

response = client.chat.completions.create(
    model="gpt-4o-mini",
    messages=messages
)
```

```python
# Step 4: Print the result

print(response.choices[0].message.content)
```
