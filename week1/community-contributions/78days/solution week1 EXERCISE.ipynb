To demonstrate your familiarity with OpenAI API, and also Ollama, build a tool that takes a technical question,  
and responds with an explanation. This is a tool that you will be able to use yourself during the course!

```python
# imports
from openai import OpenAI
import ollama
import os
```

```python
# constants

MODEL_GPT = 'gemini-3.5-flash'
MODEL_LLAMA = 'llama3.2:1b'
```

```python
# set up environment
api_key = os.environ.get("GEMINI_API_KEY")
client = OpenAI(base_url="https://generativelanguage.googleapis.com/v1beta/openai/" , api_key=api_key)
```

```python
# here is the question; type over this to ask something new

question = """
Give me complete roadmap to transition into AI/ML engineering role in big tech and startups.
"""
```

```python
# Get gpt-4o-mini to answer, with streaming
response = client.chat.completions.create(
    model=MODEL_GPT,
    messages=[{"role": "user", "content": question}],
    stream=True
)
```

```python
for chunk in response:
    print(chunk.choices[0].delta.content or "", end="")


```
