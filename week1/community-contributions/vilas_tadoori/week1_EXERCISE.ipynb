```python
# imports
import os
import json
from dotenv import load_dotenv
from IPython.display import Markdown, display, update_display
from scraper import fetch_website_links, fetch_website_contents
from openai import OpenAI
import time
```

```python
# constants

MODEL_GPT = 'gpt-4o-mini'
MODEL_LLAMA = 'llama3.2'
OLLAMA_BASE_URL = "http://localhost:11434/v1"
```

```python
# set up environment
load_dotenv(override=True)
api_key = os.getenv('OPENAI_API_KEY')

if api_key and api_key.startswith('sk-proj-') and len(api_key)>10:
    print("API key looks good so far")
else:
    print("There might be a problem with your API key? Please visit the troubleshooting notebook!")
    
MODEL = MODEL_GPT
openai = OpenAI()
```

```python
link_system_prompt = """
You are a clear technical tutor.
Please explain step by step
Use simpele Language
Use Markdown
if the question is baout code, explain what it does and why?
"""
```

```python
# here is the question; type over this to ask something new

question = """
Please give me an example of writing a step by step binary tree in python ??
"""
```

```python
def messages_for(question):
    return [
        {"role": "system", "content": link_system_prompt},
        {"role": "user", "content": question}
    ]
```

```python
messages = messages_for(question)
stream = openai.chat.completions.create(
    model=MODEL,
    messages=messages,
    stream=True
)
response = ""
display_handle = display(Markdown(""), display_id=True )

for event in stream:
    response += event.choices[0].delta.content or ""
    time.sleep(0.05)
    update_display(Markdown(response), display_id=display_handle.display_id)
```
