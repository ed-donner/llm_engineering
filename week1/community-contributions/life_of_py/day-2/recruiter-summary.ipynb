My Version: Recruiter-style Ollama website summarizer

```python
from openai import OpenAI
from scraper import fetch_website_contents

OLLAMA_BASE_URL = "http://localhost:11434/v1"

ollama = OpenAI(
    base_url=OLLAMA_BASE_URL,
    api_key="ollama"
)

system_prompt = """
You are an experienced tech reviewer and web analyst.

Your job is to analyze the contents of a website and produce a concise,
well-structured markdown report.

Your report should contain:

# Website Summary
A 2-3 sentence overview.

## Purpose
What this website is primarily trying to do.

## Highlights
- Mention interesting features.
- Mention technologies if obvious.
- Mention products, services, or projects.

## News / Updates
Summarize any announcements if present.

## Overall Impression
End with one sentence describing the website in a friendly and engaging tone.

Ignore navigation menus, cookie banners, repetitive footer text,
and anything unrelated to the main content.

Respond only in markdown.
"""

user_prompt_prefix = """
Analyze the following website and generate the report.

Website Content:

"""

def messages_for(website):
    return [
        {"role": "system", "content": system_prompt},
        {"role": "user", "content": user_prompt_prefix + website}
    ]

def summarize(url):
    website = fetch_website_contents(url)

    response = ollama.chat.completions.create(
        model="llama3.2",
        messages=messages_for(website),
    )

    return response.choices[0].message.content


print(summarize("https://dev-portfolio-rho-eight.vercel.app/"))
```
