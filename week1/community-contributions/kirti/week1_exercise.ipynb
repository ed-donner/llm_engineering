```python
# imports
import os
from dotenv import load_dotenv
from IPython.display import Markdown, display
from openai import OpenAI
import requests
```

```python
# constants

MODEL_GPT = 'gpt-4o-mini'
MODEL_LLAMA = 'llama3.2'
```

```python
# set up environment
load_dotenv(override=True)
api_key = os.getenv('OPENAI_API_KEY')

# Check the key

if not api_key:
    print("No API key was found - please head over to the troubleshooting notebook in this folder to identify & fix!")
elif not api_key.startswith("sk-proj-"):
    print("An API key was found, but it doesn't start sk-proj-; please check you're using the right key - see troubleshooting notebook")
elif api_key.strip() != api_key:
    print("An API key was found, but it looks like it might have space or tab characters at the start or end - please remove them - see troubleshooting notebook")
else:
    print("API key found and looks good so far!")
```

```python
# here is the question; type over this to ask something new

question = """
Please explain what this code does and why:
with open("hamlet.txt", "r", encoding="utf-8") as f:
    hamlet = f.read()
loc = hamlet.find("Speak, man")
print(hamlet[loc:loc+100]) 

"""
system_prompt = "You are a helpful assistant. Explain code clearly in markdown format. Use code blocks to format the code."
```

```python
# Get gpt-4o-mini to answer, with streaming
openai = OpenAI()

stream = openai.chat.completions.create(model=MODEL_GPT, 
messages=[{"role": "system", "content":  system_prompt}, {"role": "user", "content": question}],stream=True)
handle = display(Markdown(""), display_id=True)
text=""
for chunk in stream:
    text = text + (chunk.choices[0].delta.content or "")
    handle.update(Markdown(text))
    #print(chunk.choices[0].delta.content, end="", flush=True)   # streaming the response        
```
