and responds with an explanation. This is a tool that you will be able to use yourself during the course!

```python
# imports
import os
from dotenv import load_dotenv
from IPython.display import Markdown,display,update_display
from openai import OpenAI
```

```python
# constants

MODEL_GPT = 'gpt-4o-mini'
MODEL_LLAMA = 'llama3.2'

OLLAMA_BASE_URL = "http://localhost:11434/v1"
```

```python
# set up environment
load_dotenv(override=True)
api_key = os.getenv('OPENAI_API_KEY')
```

```python
# here is the question; type over this to ask something new

question = """
Please explain what this code does and why:
yield from {book.get("author") for book in books if book.get("author")}
"""
```

```python

system_prompt = """
You are an assistant who answers STEM questions only and nothing else.
If the user asks a non STEM question, simply respond saying you only resolve STEM questions.
Mention the technical area of the question.
The answer should be short, clear and simple. Do not ramble.
If a code snippet is provided, explain the code snippet. Then proceed to explain each line of code.
Do not provide any example code. The user is expecting an explanation.
Respond in markdown. Do not wrap the markdown in a code block - respond just with the markdown.

The format of your answer should be:

* Technical area:

* Explanation:
"""
```

```python
# Get gpt-4o-mini to answer, with streaming

openai = OpenAI()

def stream_answer_openai(question):
    stream = openai.chat.completions.create(
        model = MODEL_GPT,
        messages = [
            {"role":"system","content":system_prompt},
            {"role":"user","content":question}
        ],
        stream = True   
    )
    response = ""
    display_handle = display(Markdown(""),display_id=True)
    for chunk in stream:
        response += chunk.choices[0].delta.content or ''
        update_display(Markdown(response), display_id=display_handle.display_id)
```

```python
stream_answer_openai(question)
```

```python
question = "Which is the largest city in the world?"

stream_answer_openai(question)
```
