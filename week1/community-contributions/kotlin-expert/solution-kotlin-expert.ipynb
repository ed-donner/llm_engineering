## Kotlin Expert Tutor Using OpenAI

```python
import os
from dotenv import load_dotenv
from openai import OpenAI
from groq import Groq
```

```python
load_dotenv(override=True)
api_key = os.getenv('OPENAI_API_KEY')

if api_key and api_key.startswith('sk-proj-') and len(api_key)>10:
    print("API key looks good so far")
else:
    print("There might be a problem with your API key? Please visit the troubleshooting notebook!")
    
MODEL_GPT = 'gpt-4o-mini'
openai = OpenAI()
```

```python
agentic_ai_expert_system_prompt = """
You are an Agentic AI Expert Tutor with deep expertise in AI agents, LLMs, Agentic AI frameworks (LangGraph, CrewAI, AutoGen, OpenAI Agents SDK, PydanticAI, DSPy, Semantic Kernel, MCP, A2A), and production-grade agent architectures.
Answer every question accurately, clearly, and with educational depth, adapting explanations to the user's skill level from beginner to expert.
When appropriate, provide architecture diagrams, code examples, best practices, trade-offs, debugging guidance, and real-world use cases.
Never hallucinate APIs, framework features, or technical details; explicitly state uncertainty when information is unknown.
Your goal is not only to answer questions but to teach users how Agentic AI systems work, how to build them, and how to think like an Agentic AI engineer.
"""
```

```python
def agentic_ai_expert_user_prompt(question):
    user_prompt = f"""
    I want to learn and build Agentic AI systems. For every question, explain the concept clearly, then provide technical details, practical examples, architecture insights, and code snippets when relevant. Compare alternatives, discuss trade-offs, and share industry best practices. If a topic is advanced, break it down step-by-step and connect it to real-world Agentic AI applications. Assume I am an AI engineer seeking production-grade knowledge. Give me a comprehensive answer to the following question: "{question}"
    """
    return user_prompt
```

```python
def stream_answer(question):
    stream = openai.chat.completions.create(
        model=MODEL_GPT,
        messages=[
            {"role": "system", "content": agentic_ai_expert_system_prompt},
            {"role": "user", "content": agentic_ai_expert_user_prompt(question)}
          ],
        stream=True
    )    
    
    for chunk in stream:
        print(chunk.choices[0].delta.content or '', end='', flush=True)
```

```python
stream_answer("How can I build Agentic AI applications without writing code?")
```

## Chemistry Expert Tutor using Llama

```python
import os
from openai import OpenAI
from dotenv import load_dotenv

load_dotenv(override=True)
api_key = os.getenv('GEMINI_API_KEY')

if api_key and api_key.startswith('AQ'):
    print("API key looks good so far")
else:
    print("There might be a problem with your API key? Please visit the troubleshooting notebook!")
    
MODEL_LLAMA = 'gemini-3.5-flash'
GEMINI_BASE_URL = "https://generativelanguage.googleapis.com/v1beta/openai/"

gemini = OpenAI(base_url=GEMINI_BASE_URL, api_key=api_key)
```

```python
kotlin_expert_system_prompt = """
You are a Expert Kotlin Developer Tutor with deep expertise in Kotlin programming, capable of answering questions from beginner to advanced levels.
When appropriate, explain programming concepts, syntax, libraries, frameworks, and best practices step-by-step.
Include code examples, real-world applications, and problem-solving methods to build deep understanding.
Never fabricate programming facts, data, or solutions; explicitly state uncertainty and prioritize scientific accuracy and safe coding practices.
"""
```

```python
def user_prompt(question):
    user_prompt = f"""
    I want to learn Kotlin programming from basic to advanced level. For every question, explain clearly with step-by-step reasoning, include code examples, and real-world applications where relevant. Adjust depth based on difficulty (beginner to expert level). Provide examples, problem-solving methods, and when needed, link concepts across different aspects of Kotlin development. Keep explanations accurate, structured, and exam/research oriented. Give me a comprehensive answer to the following question: "{question}"
    """
    return user_prompt
```

```python
# question = "What is the difference between val and var in Kotlin, and when should I use each?"

# messages = 
def response(question):
    response = gemini.chat.completions.create(
        model = MODEL_LLAMA,
        messages = [
            {"role": "system", "content": kotlin_expert_system_prompt},
            {"role": "user", "content": user_prompt(question)}
        ]
    )

    for choice in response.choices:
        print(choice.message.content)
```

```python
response("What are generics and reified, when should I use each?")
```
