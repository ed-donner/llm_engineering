```python
# imports

import os
from dotenv import load_dotenv
from openai import OpenAI

# Load environment variables in a file called .env

load_dotenv(override=True)
api_key = os.getenv('OPENAI_API_KEY')

# Check the key

if not api_key:
    print("No API key was found - please head over to the troubleshooting notebook in this folder to identify & fix!")
elif not api_key.startswith("sk-proj-"):
    print("An API key was found, but it doesn't start sk-proj-; please check you're using the right key - see troubleshooting notebook")
elif api_key.strip() != api_key:
    print("An API key was found, but it looks like it might have space or tab characters at the start or end - please remove them - see troubleshooting notebook")
else:
    print("API key found and looks good so far!")


openai = OpenAI()

# Email summarizer

# Step 1: Create your prompts

system_prompt = """ 
You're a professional email summarizer assistant who reads, understands, summarizes, emails that your boss receives.
You will write a summary of the email received, specify the date it was received, the sender, the main message, any requests, questions, or concerns,
and label each in a new line.
You will also include the subject line.
"""
user_prompt = f"""
    Summarize this email and let me know what the purpose of the email is. 
    If the email is of high importance label it with a "URGENT!" subtitle.
    Here is the email:
"""

email = """
To: team.lead@innovativetech.org
Date & Time: August 5, 2026 at 11:30 AM EDT
Subject: Request: Brief Review Session on Infrastructure Integration

Hi Jordan,

Do you have 20 minutes available later this week for a brief review session? I would like to run through our updated directory configuration and gather your feedback on our operational rollout timeline.

Please let me know if any of the following times work for you:

Thursday, Aug 7 at 2:00 PM EDT

Friday, Aug 8 at 10:30 AM EDT

If another time suits you better, feel free to send an invitation.

Thanks,
"""

# Step 2: Make the messages list

messages = [{"role": "system", "content": system_prompt},
            {"role": "user", "content": user_prompt + email}] # fill this in

# Step 3: Call OpenAI
response = openai.chat.completions.create(model="gpt-4.1-mini", messages=messages)

# Step 4: print the result
print(response.choices[0].message.content)
```
