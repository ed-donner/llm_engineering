```python
# imports

import os
from dotenv import load_dotenv
from openai import OpenAI

# Load environment variables in a file called .env
load_dotenv(override=True)
ollama_api_key = os.getenv('OPENAI_API_KEY')
OLLAMA_BASE_URL = "http://localhost:11434/v1"


ollama = OpenAI(base_url=OLLAMA_BASE_URL, api_key="ollama")

def greeting():
   print("""\n\nHi! This is an email message summarizer powered by llama3.2
   I will summarize the email message and give you key details about it such as:
      - Date/Time Received
      - Purpose of the email
      - Sender""")

def get_email():
   message = input("\n\nPaste your email message here:")
   if(message):
      summarize(message)
   else:
      print("\n\nYou must provide an email message. Try again.")
      get_email()

def summarize(email):
   # Step 1: Create your prompts
   system_prompt = """ 
   You're a professional email summarizer assistant who reads, understands, summarizes, emails that your boss receives.
   You will write a summary of the email received, specify the date it was received, the sender, the main message, any requests, questions, or concerns,
   and label each in a new line.
   You will also include the subject line.
   """
   user_prompt = f"""
   Summarize this email and let me know what the purpose of the email is. 
   If the email requires me to do something label it as urgent.
   Here is the email:
   """

   # Step 2: Make the messages list

   messages = [{"role": "system", "content": system_prompt},
               {"role": "user", "content": user_prompt + email}] # fill this in

   # Step 3: Call OpenAI
   response = ollama.chat.completions.create(model="llama3.2", messages=messages)

   # Step 4: print the result
   print("\n\n" + response.choices[0].message.content)

greeting()
get_email()
```
