```python
import os
from dotenv import load_dotenv
from IPython.display import Markdown, display
from openai import OpenAI
```

```python
# Load the environment variables
load_dotenv(override=True)
api_key = os.getenv('OPENAI_API_KEY')
```

```python
openai = OpenAI()

def greeting():
   print("""\n\nHi! I am your personal gym coach
   I can give you workout routine and diet recommendations as well as create a plan that suits your needs.
   I will ask you a few questions in order to help you.""")


def make_plan(age, weight, level, goal, questions):
   # System promtp - gym coach/trainer
   system_prompt = """
   You are a certified professional gym coach/trainer, knowledgeable in nutrition and physiology.
   You give workout routines recommendations for all levels: beginner, intermediate, advanced
   You provide diet plan ideas for bulking/cutting as well as any clarification or explanation requested.
   Respond in markdown only. Do not wrap the markdown in a codeblock, but only respond in markdown.
   """

   # User prompt - instructions
   user_prompt = """
   Give me workout routine and diet suggestions in order to meet my goals.
   Create a plan for me to follow based on the following information about myself:
   """

   messages = [{"role": "system", "content": system_prompt},
               {"role": "user", "content": user_prompt + 
                                          "age: " + age + 
                                          "weight: " + weight + 
                                          "level: " + level + 
                                          "goal: " + goal +
                                          "questions: " + questions}]

   response = openai.chat.completions.create(model="gpt-4.1-mini", messages=messages)

   response.choices[0].message.content
   display(Markdown(response.choices[0].message.content))

def get_integer(prompt):
   """ Get and validate the user input for age and weight """
   while True:
      user_input = input(prompt)
      try:
         # Try to convert str to int
         return int(user_input)
      except ValueError:
         print("\nInvalid input. Please enter a whole number.")

def get_gym_level(prompt):
   """ Get and validate the gym level from the user """
   while True:
      valid_levels = ["beginner","intermediate","expert"]
      user_level = input(prompt).strip().lower()
      if user_level in valid_levels:
         return user_level
      else:
         print(f"\nInvalid input. Please choose from: {", ".join(valid_levels)}")

def get_string(prompt):
   """ Get a comment or question from the user """
   return input(prompt).strip()

def get_user_data():
   """ Get all the user data needed to make a gym routine and meal plan """
   age = get_integer("\nHow old are you?")
   weight = get_integer("\nWhat is your weight in pounds?")
   level = get_gym_level("\nAre you a beginner, intermediate, or expert at the gym?")
   goal = get_string("\nIn a few words describe what your goal is: ")
   question = get_string("\nQuestions/Comments: ")
   
   make_plan(age=str(age), weight=str(weight), level=level, goal=goal, questions=question)

greeting()
get_user_data()
```
