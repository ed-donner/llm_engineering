```python
import os
from dotenv import load_dotenv
from openai import OpenAI
```

```python
# Load the environment variables
load_dotenv(override=True)
api_key = os.getenv('OPENAI_API_KEY')
```

```python
openai = OpenAI()

def greeting():
   print("""\n\nHi! I am your personal gym coach
   I can give you workout routine and diet recommendations as well as create a plan that suits your needs.
   I will ask you a few questions in order to help you.""")

def questionnaire():
   try:
      age = input("\nHow old are you? ")
      int(age)

      weight = input("\nWhat is your weight in pounds? ")
      int(weight)
   except ValueError:
      print("\nAge/Weight must be a number")

   try:
      level = input("\nDo you consider yourself a begginer, intermediate, or expert at the gym? ")
      str(level)
   except ValueError:
      print("\nPlease ")

   goal = input("In a few words describe what you goal is (weight-loss, muscle gain, etc): ")
   questions = input("Additional questions: comments: ")

   make_plan(age=age, weight=weight, level=level, goal=goal, questions=questions)

def make_plan(age, weight, level, goal, questions):
   # System promtp - gym coach/trainer
   system_prompt = """
   You are a certified professional gym coach/trainer, knowledgeable in nutrition and physiology.
   You give workout routines recommendations for all levels: beginner, intermediate, advanced
   You provide diet plan ideas for bulking/cutting as well as any clarification or explanation requested.
   Respond in markdown only. Do not wrap the markdown in a codeblock, but only respond in markdown.
   """

   # User prompt - instructions
   user_prompt = """
   Give me workout routine and diet suggestions in order to meet my goals.
   Create a plan for me to follow based on the following information about myself:
   """

   messages = [{"role": "system", "content": system_prompt},
               {"role": "user", "content": user_prompt  + 
                                          "age: "       + age + 
                                          "weight: "    + weight + 
                                          "level: "     + level + 
                                          "goal: "      + goal +
                                          "questions: " + questions}]

   response = openai.chat.completions.create(model="gpt-4.1-mini", messages=messages)

   response.choices[0].message.content
   print("\n\n" + response.choices[0].message.content)

greeting()
questionnaire()
```
