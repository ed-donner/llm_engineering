# Scraper timeout drop-in

Official `week1/scraper.py` has no `timeout` on `requests.get`, so a slow host hangs Day 5. This notebook uses the drop-in scraper in this folder (`timeout=10`).

```python
from scraper import fetch_website_contents, fetch_website_links, REQUEST_TIMEOUT
import requests

print("timeout seconds:", REQUEST_TIMEOUT)
```

## Known-good URL (same as the Day 5 lab)

```python
url = "https://edwarddonner.com"
text = fetch_website_contents(url)
links = fetch_website_links(url)
print(text[:400])
print("\nlink count:", len(links))
```

## Timeout instead of a hang

This address is typically unroutable. Without a timeout, `requests.get` can wait forever. With `timeout=10` you get `requests.exceptions.Timeout` (or a related connection error) in about 10 seconds.

```python
try:
    fetch_website_contents("http://10.255.255.1")
    print("Unexpected: request succeeded")
except (requests.exceptions.Timeout, requests.exceptions.ConnectionError) as e:
    print(type(e).__name__ + ":", e)
```
