# Day 1 Exercise: Email Subject Line Suggester

Summarization use case: given the body of an email, suggest a short, appropriate subject line.

```python
# imports

import os
from dotenv import load_dotenv
from openai import OpenAI

load_dotenv(override=True)
api_key = os.getenv('OPENAI_API_KEY')

if not api_key:
    print("No API key was found - check your .env file")
else:
    print("API key found")

openai = OpenAI()
```

```python
# Step 1: Create your prompts

system_prompt = """
You are an assistant that writes email subject lines.
Given the body of an email, suggest one clear, specific subject line.
Keep it under 10 words, avoid clickbait and all caps, and capture the main purpose or action required.
Respond with only the subject line, with no quotes and no extra commentary.
"""

user_prompt = """
Hi team,

Just a reminder that the quarterly planning meeting has moved from Thursday to Friday at 3 PM.
Please update your calendars and send me your project status slides by end of day Wednesday
so I can combine them into a single deck.

Thanks,
Priya
"""

# Step 2: Make the messages list

messages = [
    {"role": "system", "content": system_prompt},
    {"role": "user", "content": user_prompt}
]

# Step 3: Call OpenAI

response = openai.chat.completions.create(model="gpt-4.1-mini", messages=messages)

# Step 4: print the result

print(response.choices[0].message.content)
```

```python
def suggest_subject(email_body):
    messages = [
        {"role": "system", "content": system_prompt},
        {"role": "user", "content": "Here is the email body:\n\n" + email_body}
    ]
    response = openai.chat.completions.create(model="gpt-4.1-mini", messages=messages)
    return response.choices[0].message.content.strip()
```

```python
emails = [
    """
    Hello,

    I am writing to follow up on my application for the Python Backend Developer position.
    I interviewed with your team last week and wanted to ask if there are any updates on the next steps.

    Regards,
    Aditya
    """,
    """
    Hi all,

    The production server will be down for maintenance on Saturday from 11 PM to 2 AM.
    Please save your work and avoid deploying anything during this window.

    - DevOps
    """
]

for email in emails:
    print(suggest_subject(email))
```
