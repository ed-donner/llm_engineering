```python
inj = fetch_website_contents("https://injective.com/")
print(inj)
```

```python
# Define our system prompt - you can experiment with this later, changing the last sentence to 'Respond in markdown in Spanish."

system_prompt = """
You are a snarky elderly spanish assistant that analyzes the contents of a website,
and provides a short, interesting, humorous summary with a spanish accent/few spanish slangs mixed with english, 
ignoring text that might be navigation related. Remember to use mostly english and be witty/snarky.
Respond in markdown. Do not wrap the markdown in a code block - respond just with the markdown.

"""
```

```python
# Define our user prompt

user_prompt_prefix = """
Here are the contents of a website.
Provide a short summary of this website.
If it includes news or announcements, then summarize these too.

"""
```

```python
messages = [{"role": "system", "content": system_prompt},
{"role": "user", "content": "what is 2 + 2?"}
]
response = ollama.chat.completions.create(model="llama3.2:1b", messages=messages)

response.choices[0].message.content
```

```python
def messages_for(website):
    return [
        {"role": "system", "content": system_prompt},
        {"role": "user", "content": user_prompt_prefix + website}
    ]
```

```python
messages_for(inj)
```

```python
def summarize(url):
    website = fetch_website_contents(url)
    response = ollama.chat.completions.create(model="llama3.2:1b", messages=messages_for(website))
    return response.choices[0].message.content
```

```python
summarize("https://injective.com/")
```

```python
def display_summary(url):
    summary = summarize(url)
    display(Markdown(summary))
```

```python
display_summary("https://injective.com/")
```
