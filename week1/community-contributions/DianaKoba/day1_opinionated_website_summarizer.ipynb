# Week 1 Day 1: Opinionated website summarizer

Variation of the Day 1 lab. The model summarizes a website and then adds a short, strict opinion.

Select the repo `.venv` kernel before running. Load `OPENAI_API_KEY` from the project `.env`.

```python
import os
import sys
from pathlib import Path

from dotenv import load_dotenv
from IPython.display import Markdown, display
from openai import OpenAI

week1_dir = Path("../..").resolve()
sys.path.append(str(week1_dir))
from scraper import fetch_website_contents
```

```python
load_dotenv(week1_dir.parent / ".env", override=True)
load_dotenv(override=True)

api_key = os.getenv("OPENAI_API_KEY")
if not api_key:
    print("No API key was found. Check your .env and the troubleshooting notebook.")
elif not api_key.startswith("sk-proj-"):
    print("An API key was found, but it does not start with sk-proj-.")
elif api_key.strip() != api_key:
    print("An API key was found, but it has space or tab characters at the start or end.")
else:
    print("API key found and looks good so far!")

openai = OpenAI()
```

```python
system_prompt = """
You are a very detail oriented website analyzer with strict opinions.
Respond in markdown. Do not wrap the markdown in a code block - respond just with the markdown.
"""

user_prompt_prefix = """
Summarize the website and add your opinions at the end of it.

Website contents:
"""
```

```python
url = "https://anthropic.com"
website = fetch_website_contents(url)

messages = [
    {"role": "system", "content": system_prompt},
    {"role": "user", "content": user_prompt_prefix + website},
]

response = openai.chat.completions.create(
    model="gpt-4.1-mini",
    messages=messages,
)

print(response.choices[0].message.content)
display(Markdown(response.choices[0].message.content))
```
