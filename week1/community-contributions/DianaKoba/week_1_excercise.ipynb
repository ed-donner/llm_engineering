````python
# imports
# If these fail, please check you're running from an 'activated' environment with (llms) in the command prompt

import os
import sys
from pathlib import Path
from dotenv import load_dotenv
from IPython.display import Markdown, display, update_display
from openai import OpenAI
import ollama

week1_dir = Path("../..").resolve()
sys.path.append(str(week1_dir))
from scraper import fetch_website_links, fetch_website_contents
````

````python
# Initialize and constants

load_dotenv(week1_dir.parent / ".env", override=True)
load_dotenv(override=True)
api_key = os.getenv('OPENAI_API_KEY')

if api_key and api_key.startswith('sk-proj-') and len(api_key)>10:
    print("API key looks good so far")
else:
    print("There might be a problem with your API key? Please visit the troubleshooting notebook!")
    
MODEL_GPT = 'gpt-4o-mini'
MODEL_LLAMA = 'llama3.2'
````

````python
# set up environment
openai = OpenAI()
````

````python
# here is the question; type over this to ask something new

question = """
Please explain what this code does and why:
yield from {book.get("author") for book in books if book.get("author")}
"""
````

````python
system_prompt = """
you are a senior software engineer, expertized in explaining lines of code,
line by line, as short as possible and suitable for all levels,
Respond in markdown without code blocks.
"""

user_prompt = "please give a comprehensive answer to the following question " + question
````

````python
messages = [
        {"role": "system", "content": system_prompt},
        {"role": "user", "content": user_prompt}
    ]
````

````python
def stream_technical_question():
    stream = openai.chat.completions.create(
        model=MODEL_GPT,
        messages=messages,
        stream=True
    )    
    response = ""
    display_handle = display(Markdown(""), display_id=True)
    for chunk in stream:
        response += chunk.choices[0].delta.content or ''
        update_display(Markdown(response), display_id=display_handle.display_id)
````

````python
stream_technical_question()
````

````python
def stream_technical_question_ollama():

    stream = ollama.chat(model=MODEL_LLAMA, messages=messages,  stream=True)

    response = ""
    display_handle = display(Markdown(""), display_id=True)
    for chunk in stream:
        response += chunk.get("message", {}).get("content", "")
        response = response.replace("```","").replace("markdown", "")
        update_display(Markdown(response), display_id=display_handle.display_id)
````
