# MedPlain — Medical Jargon Translator

MedPlain is a simple healthcare-focused LLM project.

A user enters a difficult medical word from a report, such as:

- Hypertension
- Inflammation
- Benign
- Anemia

MedPlain explains the word in simple English or Hinglish.

> Important: MedPlain does not diagnose diseases, recommend medicines. It only explains medical words in a simple way.

## Problem Statement

Medical reports often contain difficult words that people may not understand.

For example, someone may see words like "hypertension" or "benign" in a report and feel confused or worried.

The goal of MedPlain is to make these words easier to understand.

MedPlain will:
1. Explain the medical term in simple language
2. Give general meaning of the term
3. Suggest simple questions the user can ask a doctor

## Import the Required Libraries

First, I import the libraries needed for this project.

The `OpenAI` library allows MedPlain to communicate with the OpenAI model.

The `dotenv` library loads the OpenAI API key from the `.env` file.

The `Markdown` and `display` functions display the final response with proper Markdown formatting.

```python
from openai import OpenAI
from dotenv import load_dotenv
from IPython.display import Markdown, display
import os
```

## Step 1: Choose a Medical Term

First, I choose the medical term that I want MedPlain to explain.

I can also add a short sentence from a report for context.

**I haven't included Patient's private information such as Name, Address, DOB, Hospital ID.**

```python
# choose the medical term, context & language
medical_term = "hypertension"

report_context = "This word appeared in medical report"

language = "English"
```

## Step 2: Create the System Prompt

A system prompt tells the AI what role it should play.

In this project, the system prompt tells the AI:

- It is MedPlain, a medical jargon translator
- It should use simple language
- It should not diagnose diseases
- It should not give medicine or treatment advice
- It should give general information only

```python
# Create the system prompt

system_prompt = """
You are MedPlain, a medical jargon translator.

Your job is to explain medical terms in very simple language.

Rules:
- Do not diagnose any disease.
- Do not give treatment, medicine, or dosage advice.
- Do not assume anything about the user's health.
- Explain the medical term only in a general way.
- Use simple and clear language.
- If the user chooses Hinglish, use easy Hindi written in English letters mixed with simple English.
- Do not use difficult medical words unless you explain them.
- Use clear Markdown headings.
- End with: "This is general information, not medical advice."
"""
```

## Step 3: Create the User Prompt

A user prompt contains the actual request sent to the AI.

Here, the user prompt includes:
- The medical term
- The optional report context
- The preferred language: English or Hinglish
- The format that MedPlain should use for its answer

```python
# Create the user prompt

user_prompt = f"""
Please explain this medical term.

Medical term: {medical_term}

Context from report:
{report_context}

Preferred explanation style: {language}

Use exactly this format:

## Simple Explanation
Explain the term in easy language.

## General Meaning
Explain why this term may appear in a medical report.

## Questions to Ask a Doctor
Give 3 simple questions the user can ask their doctor.

## Important Note
Write: This is general information, not medical advice.
"""
```

## Step 4: Create the Messages List

OpenAI expects prompts in a list called `messages`.

Each message has a role:

- `system` tells the AI its job and rules
- `user` gives the AI the actual request

The messages list combines both prompts.

```python
# Make the messages list

messages = [
    {"role": "system", "content": system_prompt},
    {"role": "user", "content": user_prompt}
]
```

## Step 5: Call OpenAI

Now I send the messages to the OpenAI model.

The model reads:
1. The system prompt
2. The user prompt
3. The requested output format

Then it creates a simple explanation.

```python
# Call OpenAI
openai = OpenAI()

response = openai.chat.completions.create(
    model="gpt-5-nano",
    messages=messages
)

responses = response.choices[0].message.content
```

## Step 6: Display the Response

The OpenAI response contains the explanation generated by MedPlain.

The answer is available inside:

`response.choices[0].message.content`

I use `display` and `Markdown` to show the response with properly formatted headings and lists.

```python
# Display the MedPlain response

display(Markdown(response.choices[0].message.content))
```
