and responds with an explanation. This is a tool that you will be able to use yourself during the course!

```python
# imports
import os, sys, json
from dotenv import load_dotenv
from openai import OpenAI
from IPython.display import Markdown, display, update_display
```

```python
# constants

MODEL_DEEPSEEK = 'deepseek-v4-pro'
MODEL_LLAMA = 'llama3.2'
MODEL_PHI = 'phi3'
# Load the api keys from the .env file
load_dotenv(override=True)
deepseek_api_key = os.environ.get("DEEPSEEK_API_KEY")
ollama_api_key = os.environ.get("OLLAMA_API_KEY")
```

```python
# set up environment
OLLAMA_BASE_URL = "http://localhost:11434/v1"
DEEPSEEK_BASE_URL = "https://api.deepseek.com"

ollama = OpenAI(base_url=OLLAMA_BASE_URL, api_key='ollama')
deepseek = OpenAI(base_url=DEEPSEEK_BASE_URL, api_key=deepseek_api_key)

system_prompt = f"""You are a knowledgeable expert in coding and working with LLMs and different AI tools, models and other technologies. 
You will be providing help and thorough explanation to different questions from a user. You will use easy to understand language  
to explain the concepts and provide useful information and examples. You response should be in one paragraph and codes should be formatted in a code block."""
```

```python
# Get LLM to answer, with streaming
def explain_this_question(question, model):
    print(f"Explaining the question: {question[:50]}... using model: {model}")
    if model == "deepseek":
        stream_rep = deepseek.chat.completions.create(
            model=MODEL_DEEPSEEK,
            messages=[
                {"role": "system", "content": system_prompt},
                {"role": "user", "content": question}
            ],
            stream=True
        )   
    elif model == "llama":
        stream_rep = ollama.chat.completions.create(
            model=MODEL_LLAMA,
            messages=[
                {"role": "system", "content": system_prompt},
                {"role": "user", "content": question}
            ],
            stream=True
        )
    elif model == "phi3":
        stream_rep = ollama.chat.completions.create(
            model=MODEL_PHI,
            messages=[
                {"role": "system", "content": system_prompt},
                {"role": "user", "content": question}
            ],
            stream=True
        )
    response = "" # needed to accumulate the streamed content
    display_handle = display(Markdown(""), display_id=True)
    for chunk in stream_rep:
        response += chunk.choices[0].delta.content or ''
        update_display(Markdown(response), display_id=display_handle.display_id)
```

```python
# Comparing the answers from different models, we can see the differences in their responses.
question = f"""
what is decorators in python and how do they work? please provide examples."""

models = ["deepseek", "llama", "phi3"]

for model in models:
    explain_this_question(question, model)
```
