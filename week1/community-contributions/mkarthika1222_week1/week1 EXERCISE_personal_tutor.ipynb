```python
# imports
import os
import json
from openai import OpenAI
from dotenv import load_dotenv
from IPython.display import Markdown,display,display_markdown,update_display
```

```python
# constants

#MODEL_GPT = 'gpt-4o-mini'
MODEL_GPT='gpt-6-luna'
MODEL_LLAMA ='llama3.2:1b'
```

```python
# set up environment
load_dotenv(override=True)
key=os.getenv("OPENAI_API_KEY")


```

```python
# here is the question; type over this to ask something new
systemprompt="""
You are an expert teacher who can explain the coding probleams easily to a student in a simplistic manner." \
You will be explaining the code line by line and share more 1 or 2 more examples.
"""

question = """
Please explain what this code does and why :
yield from {book.get("author") for book in books if book.get("author")}
"""
```

```python
# Get gpt-4o-mini to answer, with streaming

userinput=input("Enter your question")
print(userinput)
openai=OpenAI()

messages=[
    {"role":"system","content":systemprompt},
    {"role":"user","content":userinput},

]
response=openai.chat.completions.create(
model=MODEL_GPT,
reasoning_effort="none",
messages=messages,
)

output=""
answer=response.choices[0].message.content
Markdown(answer)
```

```python
# Get Llama 3.2 to answer
OLLAMA_BASE_URL = "http://localhost:11434/v1"

llama_base_url=OLLAMA_BASE_URL

ollama=OpenAI(base_url=OLLAMA_BASE_URL,api_key='ollama')
messages=[
    {"role":"system","content":systemprompt},
    {"role":"user","content":question},

]

stream=ollama.chat.completions.create(
model=MODEL_LLAMA,
reasoning_effort="none",
messages=messages,
stream=True
)
output=""
display_handle = display(Markdown(""), display_id=True)

for chunk in stream:
 output+=chunk.choices[0].delta.content or ''
 update_display(Markdown(output),display_id=display_handle.display_id)
 
```
