## Tokenizing with code

```python
from transformers import AutoTokenizer

tokenizer = AutoTokenizer.from_pretrained("NousResearch/Llama-3.2-1B")

text = "Hi my name is Ed and I like banoffee pie"

tokens = tokenizer.encode(text)

print("Tokens:", tokens)
print("Count:", len(tokens))

for token_id in tokens:
    print(token_id, "=", repr(tokenizer.decode([token_id])))
```

```python
import os
from dotenv import load_dotenv

load_dotenv(override=True)
api_key = os.getenv('GEMINI_API_KEY')

if not api_key:
    print("No API key was found - please head over to the troubleshooting notebook in this folder to identify & fix!")
elif not api_key.startswith("AQ."):
    print("An API key was found, but it doesn't start AQ.; please check you're using the right key - see troubleshooting notebook")
else:
    print("API key found and looks good so far!")
```

```python
MODEL = "llama3.2"
API_KEY = "ollama"
OLLAMA_BASE_URL = "http://localhost:11434/v1"
```

```python
from openai import OpenAI

ollama = OpenAI(base_url=OLLAMA_BASE_URL, api_key=API_KEY)
```

```python
messages = [
    {"role": "system", "content": "You are a helpful assistant"},
    {"role": "user", "content": "Hi! I'm Ed!"}
    ]
```

```python
response = ollama.chat.completions.create(model=MODEL, messages=messages)
response.choices[0].message.content
```
