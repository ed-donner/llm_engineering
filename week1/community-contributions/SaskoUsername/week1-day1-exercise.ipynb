```python
# imports

import os
from dotenv import load_dotenv
from scraper import fetch_website_contents
from IPython.display import Markdown, display
from openai import OpenAI
```

```python
# Load environment variables in a file called .env

load_dotenv(override=True)
api_key = os.getenv('OPENAI_API_KEY')

# Check the key

if not api_key:
    print("No API key was found - please head over to the troubleshooting notebook in this folder to identify & fix!")
elif not api_key.startswith("sk-proj-"):
    print("An API key was found, but it doesn't start sk-proj-; please check you're using the right key - see troubleshooting notebook")
elif api_key.strip() != api_key:
    print("An API key was found, but it looks like it might have space or tab characters at the start or end - please remove them - see troubleshooting notebook")
else:
    print("API key found and looks good so far!")
```

```python
openai = OpenAI()
def create_massage(system_prompt: str, user_prompt: str, website_content: str, location: str) -> list[dict]:
    return [
        {"role": "system", "content": system_prompt},
        {"role": "user", "content": user_prompt + website_content + location}
    ]
```

```python
system_prompt = """
    You are polite and helpful assistant, answer only in Ukrainian. Do not repeat yourself.
    Imagine you are touristic guide, who is helping to find the best places to visit in Ukraine.
    You should find at least 4 sequential places to visit, that are close to each other.
    Analize the website contents and provide a list of places with short description for each.
    User will provide you start point like city or area name and you should provide places to visit in the city or area depends on the user request.
    Respond in markdown. Do not wrap the markdown in a code block - respond just with the markdown.
"""
user_prompt = """
    Here are the contents of a website and location of the user.
"""
website_content = fetch_website_contents("https://dovkola.media/hotovyy-marshrut-podilliam-zatoplene-selo-kostely-ta-fortetsi/")

# Step 2: Make the messages list

messages = create_massage(system_prompt, user_prompt, website_content, "Khmelnytskyi and Vinnytsia area")

# Step 3: Call OpenAI
response = openai.chat.completions.create(
    model="gpt-4.1-mini",
    messages=messages
)
# Step 4: print the result
display(Markdown(response.choices[0].message.content))
```
