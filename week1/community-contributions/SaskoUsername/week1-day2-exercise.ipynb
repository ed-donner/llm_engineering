```python
ollama = OpenAI(base_url="http://localhost:11434/v1", api_key='ollama')
def create_massage(system_prompt: str, user_prompt: str, website_content: str, username: str) -> list[dict]:
    return [
        {"role": "system", "content": system_prompt},
        {"role": "user", "content": user_prompt + website_content + " username: " + username}
    ]
```

```python
system_prompt = """
   Imagine that you are personal coach for a player of league of legends game.
   You should analyze provided game with given nickname and provide feedback to the player for specific match.
   Do not analyze full user profile, only for specific match.
   Take into account the player`s role in the match.
   You should consider the player's performance, the player's mistakes, the player's strengths depends on the game statistics and another players in same match.
   Analyze, taking into account the cahmpion the user played.
   Analize the website contents and provide feedback in markdown.
"""
user_prompt = """
    Here is the game with given nickname and match statistics.
"""
website_content = fetch_website_contents("https://op.gg/lol/summoners/euw/Little%20Goblin%20JR-526/matches/ta1XN8oiMOI-1ytgYhRbM3AOXdcSwmLPv-8R0bR8qQI%3D/1795313555000")
# Step 2: Make the messages list

messages = create_massage(system_prompt, user_prompt, website_content, "Little Goblin JR#526")

# # Step 3: Call local ollama
response = ollama.chat.completions.create(
    model="llama3.2",
    messages=messages
)
# # Step 4: print the result
display(Markdown(response.choices[0].message.content))
```
