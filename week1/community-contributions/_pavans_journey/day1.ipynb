```python
#url=""
def summarizepk(url):
    result1=openai.chat.completions.create(
    model="gpt-4.1-mini",
    messages=[{"role":"system","content":
               "summarize website "},{"role":"user","content":fetch_website_contents(url)
    }
    ]
    )
    return result1.choices[0].message.content

```

```python
# A function to display this nicely in the output, using markdown

def display_summary(url):
    summary = summarizepk(url)
    display(Markdown(summary))
```
