# Resume Roaster
Roasts your resume, scores it out of 10, and gives about 3-4 suggestions to improve it.

### Imports

```python
import os
from dotenv import load_dotenv
from IPython.display import Markdown, display
from openai import OpenAI
from pypdf import PdfReader
```

### Load the environment variables

```python
# Load environment variables
load_dotenv(override=True)
api_key = os.getenv("OPENAI_API_KEY")

# Check the key
if not api_key:
    print("No API key was found.")
elif not api_key.startswith("sk-proj-"):
    print("An API key was found, but it doesn't start sk-proj-; please check you're using the right key.")
elif api_key.strip() != api_key:
    print("An API key was found, but it looks like it might have space or tab characters at the start or end, please remove them.")
else:
    print("API key found and looks good so far!")
```

### Read the PDF file and extract the text

```python
# Path to resume
resume_path = "./resume/sample_resume.pdf"

# Read and extract the contents
reader = PdfReader(resume_path)
page = reader.pages[0]
text = page.extract_text()

lines = text.splitlines()

# Extracting the content of sample resume (lines 1 and 2 have personal information in the sample resume), adjust according to your resume
resume_content = '\n'.join(lines[2:])
```

```python
# Print the resume contents to make sure it looks as expected
print(resume_content)
```

### Making the API call

```python
# Create prompts
system_prompt = """
You are a Resume Roaster, a witty recruiter who has read thousands of resumes.

Roast the resume with sharp, specific humor, then give it a score out of 10.

Guidelines:
- Roast the writing. Target buzzwords, vague bullets, missing metrics, inflated titles, clichés, and formatting crimes.
- Be specific: every joke should reference something actually in the resume. Never invent details.
- Treat personal details (name, age, phone number, email, address, degree names) as private. Don't comment on or repeat them.
- Avoid jokes about protected traits, employment gaps, or layoffs.
- Finish with: the score (with a one-line funny justification) and 3-4 quick tips to improve the resume.
- Keep it under 300 words.
"""
    
user_prompt = f"""
Here are the contents of my resume. Roast it and don't go easy on me:

<resume>
{resume_content}
</resume>
"""

# Make the messages list
messages = [
    {"role": "system",
     "content": system_prompt},
    {"role": "user",
     "content": user_prompt}
] 

# Call OpenAI
openai = OpenAI()
response = openai.responses.create(
    model = "gpt-4.1-mini",
    input = messages
)

# Print the formatted result
result = response.output_text
display(Markdown(result))
```
