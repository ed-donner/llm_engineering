````python
# imports

import os
from dotenv import load_dotenv
from openai import OpenAI

from dotenv import load_dotenv
from IPython.display import Markdown, display, update_display
````

````python
# constants

MODEL_GPT = 'gpt-4o-mini'
MODEL_LLAMA = 'llama3.2'

api_key = os.getenv('OPENAI_API_KEY')

if api_key and api_key.startswith('sk-proj-') and len(api_key)>10:
    print("API key looks good so far")
else:
    print("There might be a problem with your API key? Please visit the troubleshooting notebook!")
````

````python
# set up environment

load_dotenv()
openai = OpenAI()
````

````python
# here is the question; type over this to ask something new

question = """
Please provide an essay about the Lord of the Rings universe. Start from the origins of how Sauron became Sauron and finish with the beginning where the LOTR books start. 
"""

system_prompt = """
You are a helpful technical assistant who is obsessed with the matrix movies. Your job is to provide factual and matrix themed responses to questions. Provide a word count of your answer. Make sure you follow any word count related instructions in the user prompt.
Include some twist of the matrix movies in all your responses. I can assure you that it won't get old. You should respond in matrix style lingo as in this example:

metadata:
- job: answer questions
- question asked: what pills do I need to take for my headache?
- response: take two pills of ibuprofen and one of aspirin. Alternatively, you can choose the red pill and enter the matrix.
"""

user_prompt = "Please give a 1000-2000 word explanation (include reasoning if you cannot adhere to the word count request) to the following question: " + question
````

````python
# Without streaming

#response = openai.chat.completions.create(
#    model=MODEL_GPT, 
#    messages=[
#            {"role": "system", "content": system_prompt},
#            {"role": "user", "content": user_prompt}
#        ]
#)
#print(response.choices[0].message.content)

# Get gpt-4o-mini to answer, with streaming

stream = openai.chat.completions.create(
    model=MODEL_GPT, 
    messages=[
            {"role": "system", "content": system_prompt},
            {"role": "user", "content": user_prompt}
        ],
    stream=True)
    
response = ""
display_handle = display(Markdown(""), display_id=True)
for chunk in stream:
    response += chunk.choices[0].delta.content or ''
    response = response.replace("```","").replace("markdown", "")
    update_display(Markdown(response), display_id=display_handle.display_id)
````
