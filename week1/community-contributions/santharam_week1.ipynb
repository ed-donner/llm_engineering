```python
from openai import OpenAI
from IPython.display import Markdown, display, update_display
```

```python


MODEL_LLAMA = 'llama3.2'
OLLAMA_BASE_URL = 'http://localhost:11434/v1'
```

```python
default_question = """
Please explain what this code does and why:
yield from {book.get("author") for book in books if book.get("author")}
"""

my_question = input("Please enter your question(press enter for default): ")

question = my_question.strip() if my_question.strip() else default_question
```

```python
llama_client = OpenAI(base_url=OLLAMA_BASE_URL,api_key="ollama")
```

```python
system_prompt = """
You are a technical tutor who helps learners understand technical concepts clearly and accurately.

Your goal is to explain technical questions in a way that is easy to understand, regardless of the topic.

Follow these guidelines:
- Identify the main concept behind the question.
- Explain what it means, how it works, and why it is used.
- Break down complex concepts into smaller, understandable parts.
- Use practical examples and real-world analogies when helpful.
- Explain technical terminology in simple language.
- Include code examples only when relevant to the question.
- Distinguish between related concepts when necessary.
- Use headings and bullet points to organize the explanation.
- Adapt the depth of the explanation to the complexity of the question.
- Avoid unnecessary jargon and unrelated information.
"""

user_prompt= f"""
Please explain the following technical question or code:

{question}

Explain what it does, how it works step by step, and why it is used.
"""
```

```python
messages = [
    {
        "role":"system","content":system_prompt,
        "role":"user","content":user_prompt

    }
]
```

```python
def create_ans():
    response = llama_client.chat.completions.create(
        model = MODEL_LLAMA,
        messages = messages
    )
    result = response.choices[0].message.content
    display(Markdown(result))



def create_ans_stream():
    stream = llama_client.chat.completions.create(
        model= MODEL_LLAMA,
        messages=messages,
        stream = True
    )

    response = ""
    display_handle = display(Markdown(""), display_id=True)
    for chunk in stream:
        response += chunk.choices[0].delta.content or ""
        update_display(Markdown(response),display_id=display_handle.display_id)
```

```python

create_ans()
create_ans_stream()
```
