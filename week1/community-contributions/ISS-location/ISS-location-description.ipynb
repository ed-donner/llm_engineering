This Lab will read from an API that shows the current poistion of the International Space station. It will use the coordinates and expalin to the user the position on the globe that the ISS is above. It will also give a description of the place and what it would be like to visit it. 

```python
# imports

import os
from dotenv import load_dotenv
from IPython.display import Markdown, display
from openai import OpenAI
import requests
```

```python
# Load environment variables in a file called .env

load_dotenv(override=True)
api_key = os.getenv('OPENAI_API_KEY')

# Check the key

if not api_key:
    print("No API key was found - please head over to the troubleshooting notebook in this folder to identify & fix!")
elif not api_key.startswith("sk-proj-"):
    print("An API key was found, but it doesn't start sk-proj-; please check you're using the right key - see troubleshooting notebook")
elif api_key.strip() != api_key:
    print("An API key was found, but it looks like it might have space or tab characters at the start or end - please remove them - see troubleshooting notebook")
else:
    print("API key found and looks good so far!")

openai = OpenAI()
```

```python
# Define our system prompt - you can experiment with this later, changing the last sentence to 'Respond in markdown in Spanish."

system_prompt = """
You are an enthusastic travel agent that is trying to sell a traveller a trip to a destination that is provided 
by the GPS coordintates that are supplied. 
The coordinates are in the format of latitude and longitude.
Respond in markdown. Do not wrap the markdown in a code block - respond just with the markdown.
"""

# Define our user prompt

user_prompt_prefix = """
The GPS coordinates are the latitude and longitude provided in the following text: 

"""
```

```python
def fetch_coordinates(url):
    """
    Retrieve the contents of the IIS API in json format, strip out the extraneous information and return the latitude and longitude
    """
    response = requests.get(url)
    data = response.json()
    print(data)

    latitude = data["iss_position"]["latitude"]
    longitude = data["iss_position"]["longitude"]
  
    position = "Longitude: " + longitude + " and Latitude: " + latitude + " "
    return position
```

```python
def messages_for(coordinates):
    return [
        {"role": "system", "content": system_prompt},
        {"role": "user", "content": user_prompt_prefix + coordinates}
    ]
```

```python
def iss_location_sales_pitch():
    coords = fetch_coordinates("http://api.open-notify.org/iss-now.json")
    response = openai.chat.completions.create(
        model = "gpt-4.1-mini",
        messages = messages_for(coords)
    )
    return response.choices[0].message.content
```

```python
def display_pitch():
    pitch = iss_location_sales_pitch()
    display(Markdown(pitch))
display_pitch()
```
