```python
import os
from dotenv import load_dotenv
from openai import OpenAI
import pandas as pd
import io
from IPython.display import display, Markdown, update_display
import requests
```

```python
MODEL_GEMINI = "gemini-3.5-flash-lite"
MODEL_OLLAMA = "qwen2.5:0.5b"
```

```python
load_dotenv(override=True)
api_key = os.getenv("GOOGLE_API_KEY")

if not api_key:
    print("No key found. Check the .env location and name.")
else:
    print(f"Key loaded, starts with {api_key[:4]}")

ollama = OpenAI(base_url='http://localhost:11434/v1', api_key='ollama')
requests.get("http://localhost:11434").content

```

```python
gemini = OpenAI(
    base_url="https://generativelanguage.googleapis.com/v1beta/openai/",
    api_key=api_key,
)
```

```python
system_prompt = (
    "You are a helpful technical tutor. When given a technical question or a piece of code, "
    "explain clearly what it does and why, step by step, in Markdown. "
    "Use short examples where they help."
)
user_prompt = """
Please explain what this code does and why:
yield from {book.get("author") for book in books if book.get("author")}
"""
```

```python
messages = [
    {"role" : "user" , "content" : user_prompt},
    {"role" : "system" , "content" : system_prompt}
    
]
```

```python
stream = gemini.chat.completions.create(model=MODEL_GEMINI, messages=messages,stream=True)
response = ""
display_handle = display(Markdown(""), display_id=True)
for chunk in stream:
    response += chunk.choices[0].delta.content or ''
    update_display(Markdown(response), display_id=display_handle.display_id)
```

```python
stream = ollama.chat.completions.create(model=MODEL_OLLAMA, messages=messages, stream=True)

response = ""
display_handle = display(Markdown(""), display_id=True)
for chunk in stream:
    response += chunk.choices[0].delta.content or ''
    update_display(Markdown(response), display_id=display_handle.display_id)
```
