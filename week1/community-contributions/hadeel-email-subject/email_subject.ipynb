# Email Subject Line Generator
Week 1 Day 1 exercise: give the model an email and it suggests a short subject line.


```python
import os
from dotenv import load_dotenv
from openai import OpenAI

load_dotenv(override=True)
openai = OpenAI()
```

```python
# Step 1: Create your prompts
system_prompt = """
You are an assistant that reads an email and suggests a short, clear subject line for it.
Reply with the subject line only.
"""
user_prompt = """
Hi team,
The production deployment planned for Thursday is moved to Sunday 10 AM
because QA found a blocker in the payment flow. Please don't merge to main until then.
Thanks, Hadeel
"""

# Step 2: Make the messages list
messages = [
    {"role": "system", "content": system_prompt},
    {"role": "user", "content": user_prompt}
]

# Step 3: Call OpenAI
response = openai.chat.completions.create(model="gpt-6-luna", reasoning_effort="none", messages=messages)

# Step 4: print the result
print(response.choices[0].message.content)
```
