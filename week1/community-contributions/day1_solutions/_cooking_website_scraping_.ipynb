# Let's make a quick call to a Frontier model to get started!

```python
openai = OpenAI()
```

## Types of prompts

You may know this already - but if not, you will get very familiar with it!

Models like GPT have been trained to receive instructions in a particular way.

They expect to receive:

**A system prompt** that tells them what task they are performing and what tone they should use

**A user prompt** -- the conversation starter that they should reply to

```python
# Define our system prompt"

system_prompt = """
You are a personalised assistant that thoroughly read and understand the contents of a website,
and provides brief ideas about different types of cooking methods, ignoring text that might be navigation related.
The output should be in a simple and easy to understand format, wrapped in markdown instead of code blocks - don't include any code blocks in your response.
"""

# Define our user prompt

user_prompt_prefix = """
Here are the contents of the website and provide a brief summary of the different types of cooking methods.
It includes each type of cooking method, foods suitable to that type, and common process to prepare food in that type.
"""

# Make the call to the model

def messages_for(website):
    return [
        {"role": "system", "content": system_prompt},
        {"role": "user", "content": user_prompt_prefix + website}
    ]
```

## Time to bring it together - the API for OpenAI is very simple!

```python
# And now: call the OpenAI API. You will get very familiar with this!

def summarize(url):
    website = fetch_website_contents(url)
    response = openai.chat.completions.create(
        model = "gpt-6-astra",
        reasoning_effort = "low",
        messages = messages_for(website)
    )
    return response.choices[0].message.content
```

# A function to display this nicely in the output, using markdown

```python
def display_summary(url):
    summary = summarize(url)
    display(Markdown(summary))

display_summary("https://www.webstaurantstore.com/article/454/types-of-cooking-methods.html?srsltid=AYCX9vLsPKOmt5_NNKj-GdTZGMFxxy6idzGteiqw5u9na5z576Q67fon")
```
