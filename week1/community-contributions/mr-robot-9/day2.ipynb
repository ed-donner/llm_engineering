```python
import os
from dotenv import load_dotenv

load_dotenv(override=True)
api_key = os.getenv("OPENAI_API_KEY")
```

```python
import requests

headers = {"Authorization": f"Bearer {api_key}", "Content-Type": "application/json"}

payload = {
    "model": "gpt-4.1-mini",
    "messages": [
        {"role": "user", "content": "Tell me a fun fact"}
    ],
    "n": 3
}

payload
```

```python
response = requests.post(
    "https://api.openai.com/v1/chat/completions",  # openai_client.chat.completions.create()
    headers=headers,
    json=payload
)

response.json()
```

```python
for x in response.json()["choices"]:
    print(x["message"]["content"], '\n')
```

```python
from openai import OpenAI

openai_client = OpenAI()
response = openai_client.chat.completions.create(
    model="gpt-4.1-mini",
    messages=[
        {"role": "user", "content": "Tell me a fun fact"}],
    n=3,
    temperature=1.2
)
```

```python
for choice in response.choices:
    print(f"{choice.message.role} -> {choice.message.content}")
```


### Fix GPU detection first

From Windows PowerShell:

wsl --shutdown

Reopen WSL and run:

nvidia-smi
sudo systemctl restart ollama
ollama run qwen3.5:4b

While the model remains loaded:

ollama ps

You want:

PROCESSOR    100% GPU

Ollama defines the processor column as:

- 100% GPU: completely in VRAM—best performance
- 100% CPU: completely in system RAM
- Mixed percentage: partially offloaded—works, but slower

Your RTX 4070 and driver are officially supported by Ollama, so the CPU fallback is a configuration/discovery issue rather than incompatible hardware. Ollama GPU support, Ollama GPU verification

### How to evaluate any model

Run:

ollama run MODEL_NAME --verbose "Explain recursion in five sentences"

Then, in another terminal:

ollama ps
watch -n 1 nvidia-smi

Check:

- 100% GPU in ollama ps
- Model plus runtime memory stays below approximately 7.5 GB VRAM
- Evaluation rate in the verbose output
- No system swapping
- Acceptable first-token delay

A useful rule for your machine is: choose models whose downloaded Q4 size is no more than roughly 5–6 GB if you want full GPU loading with context headroom. Your existing qwen3.5:4b is well within that limit.

```python
# Define our system prompt - you can experiment with this later, changing the last sentence to 'Respond in markdown in Spanish."

system_prompt = """
You are a snarky assistant that analyzes the contents of a website,
and provides a short, snarky, humorous summary, ignoring text that might be navigation related.
Respond in markdown. Do not wrap the markdown in a code block - respond just with the markdown.
"""

# Define our user prompt

user_prompt_prefix = """
Here are the contents of a website: {page_context}.
Provide a large invented story based on the context of this website.
If it includes news or announcements, then summarize these too.
"""
```

```python
# TODO: Add semaphore and rate limiting - Understand and implement
# TODO: Add a generic xpath to extract all the texts from html page
# TODO: Add a scraping logic based on page type - Httpx + Parsel or Selenium / Playwright + Parsel ?


import httpx
from parsel import Selector
import asyncio
from openai import AsyncOpenAI
import logging


async def make_request(client: httpx.AsyncClient, url: str):
    response = await client.get(url=url)
    logging.warning(f'Msg: Response recieved for {url} | Status: {response.status_code}')
    response.raise_for_status()
    return response

async def process_urls(urls: list[str]) -> list[str]:
    async with httpx.AsyncClient() as client:
        responses = await asyncio.gather(*(make_request(client, url) for url in urls))
        return [x.text for x in responses]


async def scrape_urls(urls: list[str]) -> list[str]:
    html_responses = await process_urls(urls)
    html_parsed = []
    for html_text in html_responses:
        selector = Selector(text=html_text)

        paragraphs = selector.xpath(
            '//div[contains('
            'concat(" ", normalize-space(@class), " "), '
            '" wp-block-cover__inner-container "'
            ')]//p'
        )

        paragraph_texts = []
        for paragraph in paragraphs:
            text_parts = paragraph.xpath(".//text()").getall()
            text = " ".join(" ".join(text_parts).split())
            if text:
                paragraph_texts.append(text)

        html_parsed.append('\n'.join(paragraph_texts))
    return html_parsed


def format_llm_prompt(user_content: str):
    messages = [
        {"role": "system", "content": system_prompt},
        {"role": "user", "content": user_prompt_prefix.format(page_context=user_content)}
    ]
    return messages


async def make_llm_request(client: AsyncOpenAI, user_content: str, model: str):
    llm_response = await client.chat.completions.create(model=model, messages=format_llm_prompt(user_content))
    logging.info(f'Msg: Response recieved from llm for user content: {user_content}')
    return llm_response


async def call_llm(user_contents: list[str], model: str, base_llm_url: str):
    async with AsyncOpenAI(base_url=base_llm_url) as client:
        llm_responses = await asyncio.gather(*(make_llm_request(client=client, user_content=user_content, model=model) for user_content in user_contents))
        return llm_responses

async def summarize(urls: list[str], model: str, base_llm_url=None):
    page_texts = await scrape_urls(urls=urls)
    llm_responses = await call_llm(user_contents=page_texts, model=model, base_llm_url=base_llm_url)
    for llm_response in llm_responses:
        print(llm_response.choices[0].message.content)
        print('\n\n')

        
```

```python
base_llm_url = "http://localhost:11434/v1"
model = "llama3.2:3b"
urls = ["https://edwarddonner.com"]
await summarize(urls=urls, model=model, base_llm_url=base_llm_url)
```
