LLM Come in 3 Flavours

Base Model : A model that predicts the outcome based on your input (Before GPT)
Chat/Instruct Model : Chat GPT
Reasoning/Thinking : Thinking model (o3 variants)

Modern models are hybrid of Chat & Reasoning models (Thinking threshold) - Gpt 5, Gpt 6
Note: If the task requires no reasoning, make sure to select chat/instruct model to save cost

Adding word weight to periodically into the sequence as you're making the model to predict the next tokens, causes it to reflect on its reasoning and reason a bit deeper, and challenge itself and weight up the arguments (Budget forcing)

LLM Selection Rule
1. Chat models are better for interactive and creative content generation
2. Reasoning models are better for problem solving
3. Base models are better for fine tuning to learn a new skill - Let's say you want to the agent to call a tool based on few text phrases - it doesn't need to reason, just need to call the tool - These models are apt for it to save cost
