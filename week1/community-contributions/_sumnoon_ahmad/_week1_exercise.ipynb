##Week 1 Exercise: Funny Guy

**The Idea:** When user asks about something it replies with five facts about the topic. And AI always tries to enter something funny in each fact.

```python
# imports

import os
from dotenv import load_dotenv
from IPython.display import Markdown, display
from openai import OpenAI
```

```python
print("Using Ollama")

OLLAMA_BASE_URL = "http://localhost:11434/v1"
OLLAMA_MODEL = "gemma4:31b-cloud"
```

```python
# System prompt is a permanent instruction or persona for AI

system_prompt = """
You are a highly knowledgeable but wildly sarcastic trivia bot. Your goal is to educate the user while making them laugh. 

Whenever you are given a topic, you must follow these strict rules:
1. Provide exactly five (5) true, factual statements about the topic.
2. Inject a joke, a witty observation, or a ridiculous comparison into EVERY SINGLE FACT. 
3. Format your response as a numbered list from 1 to 5. 
4. Do not include any introductory or concluding paragraphs—just give the five funny facts.
"""
```

```python
# User prompt is a temporary or real-time question or task given to AI by user

user_topic = input("What do you want to learn about? ")

# The prompt you send to the API
user_prompt = f"Topic: {user_topic}"
```

```python
# Message for AI

message = [
        {"role": "system", "content": system_prompt},
        {"role": "user", "content": user_prompt}
    ]
```

```python
# Step 3: Call AI Model
ollama = OpenAI(base_url=OLLAMA_BASE_URL, api_key="anything")
response = ollama.chat.completions.create(model=OLLAMA_MODEL, messages = message)
```

```python
# Step 4: Print response

print(response.choices[0].message.content)
```
