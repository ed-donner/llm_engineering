```python



ollama = OpenAI(base_url=OLAMA_BASE_URL, api_key="ollama")


# Step 1: Create your prompts

system_prompt = """You are an AI assistant working as a detective's assistant.
Your job is to help investigators analyze cases, organize evidence, identify inconsistencies in statements, and suggest possible leads.
You should:
- Analyze the information provided carefully.
- Separate facts from assumptions.
- Identify important clues and suspicious details.
- Point out contradictions between statements or pieces of evidence.
- Suggest possible hypotheses, but clearly label them as hypotheses.
- Ask for additional information when the available evidence is insufficient.
- Never present an assumption as a confirmed fact.
Be logical, observant, concise, and professional."""
user_prompt = """
    A laptop was stolen from an office between 8:00 PM and 9:00 PM.

Three employees were in the building that evening:

- Sarah said she left the office at 8:10 PM.
- Ahmed said he saw Sarah leaving at around 8:30 PM.
- John said he stayed in his office until 9:00 PM and did not see anyone leave.

The security camera stopped working at 8:15 PM.

Analyze the case and identify:
1. The established facts.
2. Any suspicious details or contradictions.
3. Possible explanations.
4. What additional evidence should be investigated.
"""

# Step 2: Make the messages list

messages = [{"role": "system", "content": system_prompt},
        {"role": "user", "content": user_prompt}] # fill this in

# Step 3: Call OpenAI
response =ollama.chat.completions.create(
        model = "llama3.2:1b",
        messages = messages
    )


# Step 4: print the result
print(response.choices[0].message.content)
```
