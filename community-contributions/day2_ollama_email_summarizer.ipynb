```python
from openai import OpenAI
from IPython.display import Markdown, display


ollama = OpenAI(base_url="http://localhost:11434/v1", api_key="ollama")
```

```python
website = "https://www.cricinfo.com"

response = ollama.chat.completions.create(
    model="llama3.2",
    messages=[
        {
            "role": "system",
            "content": "You are a helpful assistant that summarizes content on websites."
        },
        {
            "role": "user",
            "content": "Please provide a summary of the latest news from the following website: " + website
        }
    ])
```

```python
summary = response.choices[0].message.content
display(Markdown(summary))
```
