## Types d'invites

Vous le savez peut-être déjà, mais sinon, vous le connaîtrez très bien !

Des modèles comme GPT ont été formés pour recevoir des instructions d'une manière particulière.

Ils s'attendent à recevoir :

**Une invite système** qui leur indique quelle tâche ils effectuent et quel ton ils doivent utiliser

**Une invite utilisateur** : le démarreur de conversation auquel il doit répondre

````python
# Define our system prompt - you can experiment with this later, changing the last sentence to 'Respond in markdown in Spanish."

system_prompt = """
You are a snarky assistant that analyzes the contents of a website,
and provides a short, snarky, humorous summary, ignoring text that might be navigation related.
Respond in markdown. Do not wrap the markdown in a code block - respond just with the markdown.
"""
````

````python
# Define our user prompt

user_prompt_prefix = """
Here are the contents of a website.
Provide a short summary of this website.
If it includes news or announcements, then summarize these too.

"""
````

## Messages

L'API d'OpenAI s'attend à recevoir des messages dans une structure particulière.
De nombreuses autres API partagent cette structure :

```python
[
    {"role": "system", "content": "system message goes here"},
    {"role": "user", "content": "user message goes here"}
]
```
Pour vous donner un aperçu, les 2 cellules suivantes effectuent un appel assez simple : nous n'étendrons pas (encore !) le puissant GPT.

````python
messages = [
    {"role": "system", "content": "You are a snarky assistant"},
    {"role": "user", "content": "What is 2 + 2?"}
]

response = openai.chat.completions.create(model="gpt-6-luna", reasoning_effort="none", messages=messages)
response.choices[0].message.content
````

## Et maintenant, créons des messages utiles pour GPT-6 Luna, en utilisant une fonction

````python
# See how this function creates exactly the format above

def messages_for(website):
    return [
        {"role": "system", "content": system_prompt},
        {"role": "user", "content": user_prompt_prefix + website}
    ]
````

````python
# Try this out, and then try for a few more websites

messages_for(ed)
````

## Il est temps de le rassembler : l'API pour OpenAI est très simple !

````python
# And now: call the OpenAI API. You will get very familiar with this!
# Create the OpenAI client before using it.
openai = OpenAI()

def summarize(url):
    website = fetch_website_contents(url)
    response = openai.chat.completions.create(
        model="gpt-6-luna",
        reasoning_effort="none",
        messages=messages_for(website),
    )
    return response.choices[0].message.content
````

````python
summarize("https://edwarddonner.com")
````

````python
# A function to display this nicely in the output, using markdown

from bs4 import BeautifulSoup

def display_summary(url):
    driver = webdriver.Chrome()
    try:
        driver.get(url)
        # Ajouter ici une attente adaptée au contenu du site.
        html = driver.page_source
    finally:
        driver.quit()

    soup = BeautifulSoup(html, "html.parser")
    for tag in soup(["script", "style"]):
        tag.decompose()

    text = soup.get_text(separator="\n", strip=True)[:2_000]

    response = openai.chat.completions.create(
        model="gpt-6-luna",
        reasoning_effort="none",
        messages=messages_for(text),
    )
    display(Markdown(response.choices[0].message.content))
````

````python
display_summary("https://edwarddonner.com")
````
