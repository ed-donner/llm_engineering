## Savez-vous ce qu'est un point de terminaison ?

Si ce n'est pas le cas, veuillez consulter le guide Bases techniques dans le dossier des guides.

Et voici un point final qui pourrait vous intéresser...

````python
import requests

headers = {"Authorization": f"Bearer {api_key}", "Content-Type": "application/json"}

payload = {
    "model": "gpt-6-luna",
    "reasoning_effort": "none",
    "messages": [
        {"role": "user", "content": "Tell me a fun fact"}]
}

payload
````

````python
response = requests.post(
    "https://api.openai.com/v1/chat/completions",
    headers=headers,
    json=payload
)

response.json()
````

````python
response.json()["choices"][0]["message"]["content"]
````

# Qu'est-ce que le package openai ?

C'est ce qu'on appelle une bibliothèque client Python.

Ce n'est rien de plus qu'un wrapper pour effectuer cet appel exact au point de terminaison http.

Cela vous permet simplement de travailler avec du bon code Python au lieu de jouer avec des objets json janky.

Mais c'est tout. C’est open source et léger. Certaines personnes pensent qu’il contient le code du modèle OpenAI – ce n’est pas le cas !


````python
# Create OpenAI client

from openai import OpenAI
openai = OpenAI()

response = openai.chat.completions.create(model="gpt-6-luna", reasoning_effort="none", messages=[{"role": "user", "content": "Tell me a fun fact"}])

response.choices[0].message.content

````

## Et puis cette grande chose s'est produite :

L'API Chat Completions d'OpenAI était si populaire que les autres fournisseurs de modèles ont créé des points de terminaison identiques.

Ils sont connus sous le nom de « points de terminaison compatibles OpenAI ».

Par exemple, Google en a créé un ici : https://generativelanguage.googleapis.com/v1beta/openai/

Et OpenAI a décidé d'être gentil : ils ont dit, hé, vous pouvez simplement utiliser la même bibliothèque client que celle que nous avons créée pour GPT. Nous vous permettrons de spécifier une URL de point de terminaison différente et une clé différente, pour utiliser un autre fournisseur.

Vous pouvez donc utiliser :

```python
gemini = OpenAI(base_url="https://generativelanguage.googleapis.com/v1beta/openai/", api_key="AIz....")
gemini.chat.completions.create(...)
```

Et pour être clair - même si OpenAI est dans le code, nous utilisons uniquement cette bibliothèque client Python légère pour appeler le point de terminaison - aucun modèle OpenAI n'est impliqué ici.

Si vous êtes confus, veuillez consulter le Guide 9 dans le dossier Guides !

Et maintenant, essayons !

````python
GEMINI_BASE_URL = "https://generativelanguage.googleapis.com/v1beta/openai/"

load_dotenv(REPO_ROOT / ".env", override=True)

google_api_key = os.getenv("GOOGLE_API_KEY")

if not google_api_key:
    print("No API key was found - please be sure to add your key to the .env file, and save the file! Or you can skip the next 2 cells if you don't want to use Gemini")
elif not google_api_key.startswith(("AIz", "AQ.")):
    print("An API key was found, but it doesn't start with AIz or AQ.")
else:
    print("API key found and looks good so far!")

````

````python
gemini = OpenAI(base_url=GEMINI_BASE_URL, api_key=google_api_key)

response = gemini.chat.completions.create(model="gemini-3.1-flash-lite", messages=[{"role": "user", "content": "Tell me a fun fact"}])

response.choices[0].message.content
````

## Récapitulatif de l'installation d'Ollama

Visitez simplement [ollama.com](https://ollama.com) et installez !

Une fois terminé, le serveur ollama devrait déjà fonctionner localement.  
Si vous visitez :  
[http://localhost:11434/](http://localhost:11434/)

Vous devriez voir le message `Ollama is running`.  

Sinon, ouvrez un nouveau terminal (Mac) ou Powershell (Windows) et entrez `ollama serve`.  
Et dans un autre Terminal (Mac) ou Powershell (Windows), saisissez `ollama pull llama3.2`  
Ensuite, essayez à nouveau [http://localhost:11434/](http://localhost:11434/).

Si Ollama est lent sur votre ordinateur, essayez d'utiliser `llama3.2:1b` comme alternative. Exécutez `ollama pull llama3.2:1b` à partir d'un terminal ou d'un Powershell et modifiez le code de `MODEL = "llama3.2"` en `MODEL = "llama3.2:1b"`.

````python
system_prompt = """
En moins de 250 caractères.
"""

user_prompt_prefix = """
Rédige un résumé concis du contenu du site web fourni. En français.

"""
````

````python
def messages_for(website):
    return [
        {"role": "system", "content": system_prompt},
        {"role": "user", "content": user_prompt_prefix + website}
    ]
````

````python
from scraper import fetch_website_contents

def summarized(url):
    website = fetch_website_contents(url)
    response = ollama.chat.completions.create(
        model="llama3.2",
        messages=messages_for(website),
    )
    return response.choices[0].message.content
````

````python
summarized("https://ollama.com/")
````
