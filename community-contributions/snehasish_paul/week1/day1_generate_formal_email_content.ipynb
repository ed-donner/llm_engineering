```python
import os
from dotenv import load_dotenv
from IPython.display import Markdown, display
from openai import OpenAI
```

```python
load_dotenv(override=True)
api_key = os.getenv('OPENAI_API_KEY')
```

```python
openai = OpenAI()
```

```python
# Step 1: Create your prompts

system_prompt = """
You are an Email writing assistant that helps to draft a formal, consice email body,
and provides a relevant consice email subject.
Respond in markdown. Do not wrap the markdown in a code block - respond just with the markdown.
"""

user_prompt_prefix = """
Here are the contents of a draft email body.
Repharse and provide a short subject of this email body.

"""

email_body = """
Hello Team, This is my first step in learning AI prompts. I am excited to share my first hands-on experience on AI prompt engineering using OpenAI and system and user prompts.
I learnt how to run models locally with Ollama, write code to call OpenAI's frontier models, distinguish between the system and user prompts.
Also, used markdown to format my text output. If interested to leant the same, then to let me know. We can connect and I will be excited to share my experence.
"""

# Step 2: Make the messages list

messages = [
    {"role": "system", "content": system_prompt},
    {"role": "user", "content": user_prompt_prefix + email_body}
]

# Step 3: Call OpenAI
response = openai.chat.completions.create(
    model = "gpt-4.1-mini",
    messages = messages
)

# Step 4: print the result
display(Markdown(response.choices[0].message.content))
```
