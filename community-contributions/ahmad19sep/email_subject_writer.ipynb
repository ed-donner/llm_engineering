```python
import os
from os import system
from pathlib import Path

from dotenv import load_dotenv
from scraper import fetch_website_contents
from IPython.display import display , Markdown
from anthropic import Anthropic

from week1.day1 import client
```

```python
load_dotenv()
api_key=os.getenv("ANTHROPIC_API_KEY",default="")

if not api_key:
    raise ValueError("API_KEY is not set")

client=Anthropic(api_key=api_key ,
                 default_headers={"anthropic-workspace-id": os.getenv("ANTHROPIC_WORKSPACE_ID")}
                 )
```

```python
system_prompt="""
Yo are an expert assitant that analyzes the email deeply and write 2 best subjects properly formated looked professional and to the point not big or small and   each separate by a line break


"""
```

```python
user_prompt="""
Here are the contents of the email.
write the best 2 subjects each separate by a line break of according to the email.

"""
```

```python
def get_user_msg(email):
 return[
    {"role": "user", "content": user_prompt+email}

  ]
```

```python
def subject_writer(email):
    user_msg=get_user_msg(email)
    response=client.messages.create(
    model="claude-sonnet-4-5",
    max_tokens=1024,
    system=system_prompt,
    messages=user_msg
    )
    return response.content[0].text
```
