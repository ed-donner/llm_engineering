```python
import os
from dotenv import load_dotenv
from scraper import fetch_website_contents
from IPython.display import Markdown, display
from openai import OpenAI
```

```python
load_dotenv(override=True)
api_key = os.getenv('GOOGLE_API_KEY')

if not api_key:
    print("Sorry!!! No API key was found")
elif api_key.strip() != api_key:
    print("An API key was found, but it looks like it might have space or tab characters at the start or end - please remove them - see troubleshooting notebook")
else:
    print("API key found and looks good so far!")
```

```python
system_prompt = """
You are a technical recruiter. Analyze careers-page content and
identify software engineering internships.

For each role, return:
- Role title
- Location
- Required skills
- Application deadline
- Match score
- Reason it matches the candidate
"""

user_prompt = """
My skills are TypeScript, Next.js, Node.js, Express, PostgreSQL,
Docker, Python and Java.

Find the five roles that fit me best from this page:

"""
```

```python
def messages_for(website):
    return [
        {"role": "system", "content": system_prompt},
        {"role": "user", "content":  user_prompt + website}
    ]
```

```python
def summarize(url):
    website = fetch_website_contents(url)
    GEMINI_BASE_URL="https://generativelanguage.googleapis.com/v1beta/openai/"
    api_key = os.getenv('GOOGLE_API_KEY')
    gemini = OpenAI(base_url=GEMINI_BASE_URL, api_key=api_key)
    response = gemini.chat.completions.create(
        model = "gemini-3.1-flash-lite",
        messages = messages_for(website)
    )
    return response.choices[0].message.content
```

```python
def display_summary(url):
    summary = summarize(url)
    display(Markdown(summary))
```

```python
display_summary("https://boards.greenhouse.io/flyzipline/jobs/5754039003")
```
