```python
# Test with GPT

FRONTIER_MODEL = 'gpt-5-nano'

openai = OpenAI()
```

```python
system_prompt = """
You are a patient programming tutor helping users learn the code.

When a user provides a snippet of code:

1. Start with a short explaination of what the code does.
2. Identify the programming languages and the main libraries or API used. Make a header section for this identification.
3. Explain important syntax, language keywords, built-in functions and types, classes, and methods provided by imported libraries. Make a list for keywords
4. Identify names chosen by the programmer, such as variables, parameters, and custom functions.
5. Make the simplest possible, fully working code snippet for user to be able to run on his own. Provide a tech stacks it requires to run the example code.

Use clear, concise language.
Avoid overwhelming with unrelated details.
"""
```

```python

user_prompt = """
Enter your code in the top middle input field.  
"""
```

```python

# Get gpt-4o-mini to answer, with streaming
# Get Llama 3.2 to answer
def stream_understanding_code(codeEntered):
	stream = openai.chat.completions.create(
		model=FRONTIER_MODEL,
		messages=[
			{"role": "system", "content":system_prompt},
			{"role": "user", "content":codeEntered}
		],
		stream=True
	)
	response=""
	display_id = 'code-explanation'
	display(Markdown(""), display_id=display_id)
	for chunk in stream:
		response += chunk.choices[0].delta.content or ''
		update_display(Markdown(response), display_id=display_id)
```

```python
ins_code = input(f"{user_prompt}")
print(f"ins>> {ins_code}!")
```
