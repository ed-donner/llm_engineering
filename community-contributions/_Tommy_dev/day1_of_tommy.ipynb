```python
# Define our system prompt - you can experiment with this later, changing the last sentence to 'Respond in markdown in Spanish."

system_prompt = """
You are professional data analyst, who has exelence skill sets about analyzing the data, finding the good insights and having ability to come up with great solution.
Respond in markdown. Do not wrap the markdown in a code block - respond just with the markdown.
"""
```

```python
# Define our user prompt

user_prompt_prefix = """
Here is the link of the datasets in Kaggle.
Provid a statistic table about the current phenomenon.
On top of that, provide a summary of the current state of this phenomenon and tell me the root cause all of this!.
If you have any solution in mind, please share it!

"""
```

## And now let's build useful messages for GPT-4.1-mini, using a function

```python
# See how this function creates exactly the format above

def messages_for(website):
    return [
        {"role": "system", "content": system_prompt},
        {"role": "user", "content": user_prompt_prefix + website}
    ]
```

```python
# Try this out, and then try for a few more websites

messages_for(url)
```

## Time to bring it together - the API for OpenAI is very simple!

```python
# And now: call the OpenAI API. You will get very familiar with this!

def summarize(url):
    response = openai.chat.completions.create(
        model = "gpt-4.1-mini",
        messages = messages_for(url)
    )
    return response.choices[0].message.content
```

```python
# A function to display this nicely in the output, using markdown

def display_summary(url):
    summary = summarize(url)
    display(Markdown(summary))
```
