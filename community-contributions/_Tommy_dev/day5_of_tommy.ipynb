```python
# Define our system prompt - you can experiment with this later, changing the last sentence to 'Respond in markdown in Spanish."

system_prompt = """
You are professional data analyst, who has exelence skill sets about analyzing the data, finding the good insights and having ability to come up with great solution.
Respond in markdown. Do not wrap the markdown in a code block - respond just with the markdown.
"""
```

```python
# Define our user prompt

user_prompt_prefix = """
Here is the link of the datasets in Kaggle.
Provid a statistic table about the current phenomenon.
On top of that, provide a summary of the current state of this phenomenon and tell me the root cause all of this!.
"""
```

## And now let's build useful messages for GPT-4.1-mini, using a function

```python
# See how this function creates exactly the format above

def messages_for(website):
    return [
        {"role": "system", "content": system_prompt},
        {"role": "user", "content": user_prompt_prefix + website}
    ]
```

```python
# Try this out, and then try for a few more websites

messages_for(url)
```

## Time to bring it together - the API for OpenAI is very simple!

```python
# And now: call the OpenAI API. You will get very familiar with this!

def summarize(url):
    print(f"Fetching contents of {url}...")
    response = openai.chat.completions.create(
        model = "gpt-4.1-mini",
        messages = messages_for(url)
    )

    print(f"Response received from OpenAI API for {url}.")
    return response.choices[0].message.content
```

```python
# A function to display this nicely in the output, using markdown

def display_summary(url):
    summary = summarize(url)
    display(Markdown(summary))
    return summary
```

```python
documents = display_summary(url)
```

```python
system_prompt_ollama = """
You are my smart assistant, who has exelence skill sets about analyzing the data, finding the good insights and having ability to come up with great solution.
Your duty is to analyze the report from the documents I provide and give me a summary of the current state of this phenomenon and tell me the root cause all of this!.
You just need to analyze the documents I provide and give me a summary of the current state of this phenomenon and tell me the root cause all of this!. but don't have access any links or any other information except the documents I provide.
"""

# Define our user prompt

user_prompt_prefix_ollama = """
Here is the document I want you to analyze.

After analyzing the document, provide a statistic table about the current phenomenon.
On top of that, I am usually stay up late at night and use my phone a lot, so I want you to provide a summary of the current state of this phenomenon and tell me the root cause all of this!.
If you have any solution in mind, please share it!

Please give your advice to me in a sincere and friendly way, as if you were my friend. I want to hear your honest opinion and advice, not just a generic response in the end of your analysis. I want to feel like you really care about me and my situation, and that you are giving me advice that is tailored to me as a person, not just a generic response.
Personal advice for me
"""
```

```python
def combine_messages(system_prompt_ollama, user_prompt_prefix_ollama, documents):
    return [
        {"role": "system", "content": system_prompt_ollama},
        {"role": "user", "content": user_prompt_prefix_ollama + "\n\n" + "\n\n".join(documents)}
    ]
```

```python
print(combine_messages(system_prompt_ollama, user_prompt_prefix_ollama, [documents]))
```

```python
print("Sending request to Ollama API...")
response = ollama.chat.completions.create(
    model = "deepseek-r1:1.5b",
    messages = combine_messages(system_prompt_ollama, user_prompt_prefix_ollama, [documents])
)

print("Response received from Ollama API.")
```

```python
display(Markdown(response.choices[0].message.content))
```
