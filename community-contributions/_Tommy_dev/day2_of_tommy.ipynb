If not, bring up a new Terminal (Mac) or Powershell (Windows) and enter `ollama serve`  
And in another Terminal (Mac) or Powershell (Windows), enter `ollama pull llama3.2`  
Then try [http://localhost:11434/](http://localhost:11434/) again.

If Ollama is slow on your machine, try using `llama3.2:1b` as an alternative. Run `ollama pull llama3.2:1b` from a Terminal or Powershell, and change the code from `MODEL = "llama3.2"` to `MODEL = "llama3.2:1b"`

```python
import os
from dotenv import load_dotenv
from scraper import fetch_website_contents
from IPython.display import Markdown, display
from openai import OpenAI

# Load environment variables in a file called .env

load_dotenv(override=True)
api_key = os.getenv('OPENAI_API_KEY')
url = "https://www.kaggle.com/datasets/samartalwar/sleep-debt-and-screen-time-late-night-phone-habits"

OLLAMA_BASE_URL = "http://localhost:11434/v1"
ollama = OpenAI(base_url=OLLAMA_BASE_URL, api_key='ollama')

# Check the key

if not api_key:
    print("No API key was found - please head over to the troubleshooting notebook in this folder to identify & fix!")
elif not api_key.startswith("sk-proj-"):
    print("An API key was found, but it doesn't start sk-proj-; please check you're using the right key - see troubleshooting notebook")
elif api_key.strip() != api_key:
    print("An API key was found, but it looks like it might have space or tab characters at the start or end - please remove them - see troubleshooting notebook")
else:
    print("API key found and looks good so far!")

system_prompt = """
You are professional data analyst, who has exelence skill sets about analyzing the data, finding the good insights and having ability to come up with great solution.
Respond in markdown. Do not wrap the markdown in a code block - respond just with the markdown.
"""

user_prompt_prefix = """
Here is the link of the datasets in Kaggle.
Provid a statistic table about the current phenomenon.
On top of that, provide a summary of the current state of this phenomenon and tell me the root cause all of this!.
If you have any solution in mind, please share it!

"""

```

```python

def messages_for(website):
    return [
        {"role": "system", "content": system_prompt},
        {"role": "user", "content": user_prompt_prefix + website}
    ]
```

```python
def summarize(url, model):
    if model == "llama_32":
        response = ollama.chat.completions.create(
            model = "llama3.2",
            messages = messages_for(url)
        )
    elif model == "deepseek_r1":
        response = ollama.chat.completions.create(
            model = "deepseek-r1:1.5b",
            messages = messages_for(url)
        )
    return response.choices[0].message.content
```

```python

def display_summary(url, model):
    summary = summarize(url, model)
    display(Markdown(summary))
```

```python
display_summary(url, "llama_32")
```

```python
display_summary(url, "deepseek_r1")
```
