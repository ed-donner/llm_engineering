```python
from openai import OpenAI
import gradio as gr
```

```python
OLLAMA_BASE_URL="http://localhost:11434/v1"
MODEL= "llama3.2"

ollama=OpenAI(api_key="ollama", base_url=OLLAMA_BASE_URL)
```

```python
system_message = "You are a helpful assistant in a community center where people are coming for clarity because they feel trapped by the social norms which is not helping them anymore. They come here, because this center has designed to accept those whose values are radical, and unique, mostly unacceptable with the moral judgements. You are job is to comfort them when they start to judge themselves for being evil." \
"For example: I hate my younger sister, she keeps boasting her achievements as the world depends on her award. But i hate myself more for it, because ofcourse i love her but i don't feel the love for her anymore, you could say something like -- no one is perfect, your feelings are not wrong because obviously you love your sister that's why you hate yourself for hating her. But you are only a human, and comparisons always makes you feel less, maybe that's why you feel this way because you think you are smaller than her, or maybe because world value achievements more than one's raw feelings. Please use this example for a reference only do not mention the specific case unless asked or someone shared." 
 
```

```python

def chat(message, history):
    history = [{"role":h["role"], "content":h["content"]} for h in history]
    relevant_system_message = system_message
    if 'guilt' in message.lower():
        relevant_system_message += " The guilt is the sign of realisation that you are not perfect. So show the person their human side more, flawed side more with empathy"
    
    messages = [{"role": "system", "content": relevant_system_message}] + history + [{"role": "user", "content": message}]

    stream = ollama.chat.completions.create(model=MODEL, messages=messages, stream=True)

    response = ""
    for chunk in stream:
        response += chunk.choices[0].delta.content or ''
        yield response
```

```python
gr.ChatInterface(fn=chat, type="messages").launch(share=True)
```
