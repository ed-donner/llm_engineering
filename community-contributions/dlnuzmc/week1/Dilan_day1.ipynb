```python
import os
from dotenv import load_dotenv
from scraper import fetch_website_contents
from IPython.display import Markdown, display
from openai import OpenAI


# Step 1: Create your prompts

system_prompt = """You are a data scientist with 40 years of experience. You are capable of instantly explaining even those findings that remain inexplicable by AI. 
When people ask you questions in this field, you provide answers with such simplicity and clarity that even a child between the ages of 0 and 7 could understand 
them. Please answer any questions you are asked in this field in a simple, effective, and clear manner."""

user_prompt = """
    I am a junior data scientist. While I am certainly familiar with the concept of inferential statistics, I am not entirely sure how to apply 
    these methods in my work or how to communicate the results I obtain to my manager. Could you help me with this?
"""

# Step 2: Make the messages list

messages = [{"role": "system", "content": system_prompt}, 
                {"role": "user", "content": user_prompt}]  # fill this in

# Step 3: Call OpenAI
api_key = os.getenv('OPENAI_API_KEY')
openai = OpenAI()
response = openai.chat.completions.create(model="gpt-6-luna", reasoning_effort="none", messages=messages)

# Step 4: print the result
print(response.choices[0].message.content)
```
