```python
import os
from dotenv import load_dotenv
from scraper import fetch_website_contents
from IPython.display import Markdown, display
from openai import OpenAI

# Load environment variables in a file called .env

load_dotenv(override=True)
api_key = os.getenv('GROQ_API_KEY')

# Check the key

if not api_key:
    print("No API key was found - please head over to the troubleshooting notebook in this folder to identify & fix!")
elif not api_key.startswith("sk-proj-"):
    print("An API key was found, but it doesn't start sk-proj-; please check you're using the right key - see troubleshooting notebook")
elif api_key.strip() != api_key:
    print("An API key was found, but it looks like it might have space or tab characters at the start or end - please remove them - see troubleshooting notebook")
else:
    print("API key found and looks good so far!")

from openai import OpenAI

# Connect to Groq using the OpenAI-compatible SDK
client = OpenAI(
    api_key=api_key,
    base_url="https://api.groq.com/openai/v1"
)

system_prompt = "please analyse the given content and suggest the ideas with a title and a small description of it."
user_prompt = """
     i need you to extract the most useful services offered in this website and their real world oppurtunities.
"""

# See how this function creates exactly the format above

def messages_for(website):
    return [
        {"role": "system", "content": system_prompt},
        {"role": "user", "content": user_prompt + website}
    ]

# A function to display this nicely in the output, using markdown

def display_summary(url):
    summary = summarize(url)
    display(Markdown(summary))
# And now: call the OpenAI API. You will get very familiar with this!

def summarize(url):
    website = fetch_website_contents(url)
    response = client.chat.completions.create(
        model = "openai/gpt-oss-20b",
        messages = messages_for(website)
    )
    return response.choices[0].message.content



messages = [{"role":"system","content":system_prompt},{"role":"user","content":user_prompt }] # fill this in

# Step 3: Call OpenAI
response =client.chat.completions.create(model = "openai/gpt-oss-20b",
        messages = messages)

# Step 4: print the result
# print(response.choices[0].message.content)
display_summary("https://www.brototype.com/")
```
