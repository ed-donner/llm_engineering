```python
# Scrapping content
from bs4 import BeautifulSoup
import requests
headers = {
    "User-Agent": "Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/117.0.0.0 Safari/537.36"
}
def fetch_website_contents(url):
    """
    Return the title and contents of the website at the given url;
    truncate to 2,000 characters as a sensible limit
    """
    response = requests.get(url, headers=headers)
    soup = BeautifulSoup(response.content, "html.parser")
    title = soup.title.string if soup.title else "No title found"
    if soup.body:
        for irrelevant in soup.body(["script", "style", "img", "input"]):
            irrelevant.decompose()
        text = soup.body.get_text(separator="\n", strip=True)
    else:
        text = ""
    return (title + "\n\n" + text)[:2_000]
```

```python
# Key initialziations and prompts
import os
from dotenv import load_dotenv
from IPython.display import Markdown, display
from openai import OpenAI


GROQ_BASE_URL = "https://api.groq.com/openai/v1"
GROQ_API_KEY = os.getenv("GROQAI_API_KEY")
groq = OpenAI(base_url=GROQ_BASE_URL, api_key=GROQ_API_KEY)

system_prompt = """
You are a world-class comedy writer and stand-up comedian known for sharp, witty, and clever observational humor.

Your task is to write 10 genuinely funny, original jokes inspired by the provided webpage content.

Guidelines:
- Ground the jokes specifically in the facts, tone, jargon, or absurdities present in the source text.
- Avoid generic filler, cliché puns, and low-effort "dad jokes."
- Mix comedic formats (punchy one-liners, setup-and-punchline, and observational takes).
- Deliver the output as a clean, numbered list from 1 to 10 without meta-commentary.

"""

user_prompt = """
Generate 10 hilarious, original jokes based on the provided content below.

"""
```

```python
# Place where it sends an url to scrap and create jokes
def messages_for(website_content):
    return [
        {"role": "system", "content": system_prompt},
        {"role": "user", "content": user_prompt + website_content}
    ]

def summarize(url):
    website_content = fetch_website_contents(url)
    response = groq.chat.completions.create(
        model = "groq/compound-mini",
        messages = messages_for(website_content)
    )
    return response.choices[0].message.content

def display_summary(url):
    summary = summarize(url)
    display(Markdown(summary))

display_summary("https://anthropic.skilljar.com/claude-101")
```
