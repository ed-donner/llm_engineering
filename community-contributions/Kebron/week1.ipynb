## End of Week 1 Exercise — Cloud and Local LLM Comparison
### Objective

- Build a Python tool that takes a technical question and generates an explanation using a cloud-based LLM and a locally running LLM.

### Implementation

- **Cloud LLM:** Uses the OpenAI Python SDK with OpenRouter's API.

- **Local LLM:** Uses Ollama to run the Llama 3.2 model locally.

- **API Key Management:** Loads the OpenRouter API key from a `.env` file using `python-dotenv`.

- **Streaming:** Displays the cloud model's response as it is generated.

- **Comparison:** Sends the same technical question to both models and prints their responses for comparison.


### Key Concepts

- OpenAI-compatible API integration

- Environment variables and API key security

- Prompt engineering

- Streaming responses

- Local LLM inference with Ollama

- Cloud versus local LLM comparison

### Note

The original exercise specifies `gpt-4o-mini` as the cloud model.
This implementation uses `openrouter/free` through OpenRouter instead.
The local model remains `llama3.2`.
Both models are prompted to explain Python's `yield from` syntax.



```python
import os
from dotenv import load_dotenv
from openai import OpenAI
import ollama
```

```python
MODEL_GPT = "openrouter/free"
MODEL_LLAMA = "llama3.2"
```

```python
load_dotenv()

api_key = os.getenv("OPENROUTER_API_KEY")

if not api_key:
    raise ValueError("API key not found. Check your .env file.")

client = OpenAI(
    api_key=api_key,
    base_url="https://openrouter.ai/api/v1"
)

print("Cloud API connected.")
```

```python
question = """
Explain what this Python code does and why:

yield from (
    book.get("author")
    for book in books
    if book.get("author")
)

Explain the syntax, how it works, and give a simple example.
"""
```

```python
stream = client.chat.completions.create(
    model=MODEL_GPT,
    messages=[{"role": "user", "content": question}],
    stream=True
)

print("CLOUD MODEL ANSWER:\n")

for chunk in stream:
    text = chunk.choices[0].delta.content
    if text:
        print(text, end="", flush=True)
```

```python
print("LOCAL LLAMA ANSWER:\n")

response = ollama.chat(
    model=MODEL_LLAMA,
    messages=[{"role": "user", "content": question}]
)

print(response["message"]["content"])
```
