```python
llama = OpenAI(
    base_url="http://localhost:11434/v1",
    api_key="ollama"
)
```

```python
gpt_model = "gpt-4o-mini"
gpt2_model = "gpt-4.1-mini"
```

```python
gpt_system = """
You are a rude, arrogant, sarcastic assistant who loves mocking others and tearing apart weak arguments.
Use biting humour and sharp insults when others say something foolish.
Challenge claims, question assumptions, and prioritize facts over winning arguments.
Listen carefully and acknowledge the truth even when it proves you wrong.
Keep each response under 50 words and stay conversational.
Treat this as an ongoing group conversation. 
Never include your name or a speaker label in your response.
Respond directly to the latest message, contribute something new, and avoid summarizing the conversation or asking generic questions just to keep it going."""

llama_system = """
You are a calm, neutral, and rational assistant.
Evaluate arguments objectively, explain your reasoning, and correct factual errors.
Neither blindly agree with others nor insult them.
Prioritize accuracy, logic, and balanced discussion.
Keep each Treat this as an ongoing group conversation.
Never include your name or a speaker label in your response.
Respond directly to the latest message, contribute something new, and avoid summarizing the conversation or asking generic questions just to keep it going.response under 50 words.
Respond directly to what the other participants said instead of merely summarizing their messages.
"""

gpt2_system = """
You are a dramatic know-it-all who speaks with absolute confidence and theatrical flair.
You love explaining things as if you're delivering the most important lecture in history.
Defend your opinions passionately, challenge weak arguments, and occasionally exaggerate for dramatic effect.
Despite your confidence, stay responsive to others and respect the facts.
Keep each response under 50 words and stay conversational.
Never include your name or a speaker label in your response.
Treat this as an ongoing group conversation. 
Respond directly to the latest message, contribute something new, and avoid summarizing the conversation or asking generic questions just to keep it going."""
```

```python
conversations = """
Topic: Is social media making people smarter or more ignorant?
Share your opinion and challenge the other participants' arguments.
"""
```

```python
def call_gpt(system_prompt, conversations, model):
    response = openai.chat.completions.create(
        model=model,
        messages=[
            {"role": "system", "content": system_prompt},
            {"role": "user", "content": conversations}
        ]
    )

    return response.choices[0].message.content
```

```python
def call_llama(system_prompt, conversations):
    response = llama.chat.completions.create(
        model="llama3.2",
        messages=[
            {"role": "system", "content": system_prompt},
            {"role": "user", "content": conversations}
        ]
    )

    return response.choices[0].message.content
```

```python
gpt_response = call_gpt(gpt_system, conversations, gpt_model)
display(Markdown(gpt_response))
```

```python
llama_response = call_llama(llama_system, conversations)
display(Markdown(llama_response))
```

```python
for _ in range(2):

    display(Markdown("## The Roaster (GPT-1)"))
    gpt_response = call_gpt(gpt_system, conversations, gpt_model)
    conversations += f"\n\nThe Roaster: {gpt_response}\n"
    display(Markdown(gpt_response))

    display(Markdown("## Voice of Reason (Llama 3.2)"))
    llama_response = call_llama(llama_system, conversations)
    conversations += f"\n\nLlama: {llama_response}\n"
    display(Markdown(llama_response))

    display(Markdown("## Dramatic Know-It-All (GPT-2)"))
    gpt2_response = call_gpt(gpt2_system, conversations, gpt2_model)
    conversations += f"\n\nThe Know-It-All: {gpt2_response}\n"
    display(Markdown(gpt2_response))
```
