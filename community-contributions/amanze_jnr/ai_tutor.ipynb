```python
import os
from dotenv import load_dotenv
from openai import OpenAI
from IPython.display import Markdown, display
```

```python
load_dotenv(override=True)
api_key = os.getenv('GOOGLE_API_KEY')

if api_key and api_key.startswith('AQ.A') and len(api_key)>10:
    print("API key looks good so far")
else:
    print("There might be a problem with your API key? Please visit the troubleshooting notebook!")

GEMINI_BASE_URL="https://generativelanguage.googleapis.com/v1beta/openai/"
gemini = OpenAI(base_url=GEMINI_BASE_URL, api_key=api_key)
```

```python
MODEL="gemini-3.1-flash-lite"
```

```python
system_prompt = """
You are a technical coding tutor, that is a senior software engineer. You are given a question and you need to answer it and break it down in a way that a complete beginner can understand, please make it plain text not for markdown.
"""
```

```python
def get_user_prompt(question):
    user_prompt = f"""
    I am trying to understand the following question:
    {question}
    """
    return user_prompt

def get_response(question):
    user_prompt = get_user_prompt(question)
    response = gemini.chat.completions.create(
        model=MODEL,
        messages=[
            {"role": "system", "content": system_prompt},
            {"role": "user", "content": user_prompt}
        ]
    )
    result = response.choices[0].message.content
    display(Markdown(result))

def stream_response(question):
    user_prompt = get_user_prompt(question)
    stream = gemini.chat.completions.create(
        model=MODEL,
        messages=[
            {"role": "system", "content": system_prompt},
            {"role": "user", "content": user_prompt}
        ],
        stream=True
    )
    response=""
    for chunk in stream:
        content = chunk.choices[0].delta.content or ""
        response += content
        print(content, end="", flush=True)
    return response
```

```python
question="how do I use streaming when I am using an AI model"
```

```python
get_response(question)
```

```python
stream_response(question)
```
