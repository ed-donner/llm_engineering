```python
import os
from dotenv import load_dotenv
from ai_news_scraper import fetch_website_headlines
from IPython.display import Markdown, display
from openai import OpenAI
```

```python
base_url = "http://localhost:11434/v1"
# api_key = os.getenv('OPENROUTER_API_KEY')
```

```python
ollama = OpenAI(
    base_url = base_url, 
    api_key = "ollama"
)
```

```python
system_prompt = """
You are an expert tech journalist and LinkedIn content strategist. 
Your task is to write an engaging, professional one-paragraph LinkedIn post 
for each news headline provided.
Respond in Markdown.
"""
user_prompt_prefix = """
Below is a list of news headlines. 
For EACH headline, format your response like this:
### [Headline Text]
[One-paragraph LinkedIn post]
---
Headlines:
"""
```

```python
def messages_for(headlines):
     headlines_text = "\n".join(headlines)
     return [
        {"role": "system", "content": system_prompt},
        {"role": "user", "content": user_prompt_prefix + headlines_text}
    ]
```

```python
def generate_posts(url):
    headlines = fetch_website_headlines(url)
    if not headlines:
        return "No headlines were found on the provided website."
    response = ollama.chat.completions.create(
        model = "llama3.2",
        messages = messages_for(headlines)
    )
    return response.choices[0].message.content

```

```python
def display_posts(url):
    posts = generate_posts(url)
    display(Markdown(posts))

```

```python
display_posts("https://techcrunch.com")
```
