```python
import os
from dotenv import load_dotenv
from scraper import fetch_website_contents
from IPython.display import Markdown, display
from openai import OpenAI
```

```python
base_url = "https://openrouter.ai/api/v1"
api_key = os.getenv('OPENROUTER_API_KEY')
```

```python
openai = OpenAI(
    base_url = base_url, 
    api_key = api_key
)
```

```python
system_prompt = """
You are an expert website consultant. Analyze the website's content and provide concise, actionable 
recommendations to improve its clarity, user experience, SEO, and conversion potential, 
ignoring navigation, headers, footers, and other non-content elements. 
Respond in Markdown only, without wrapping it in a code block.

"""

user_prompt_prefix = """
Below are the contents of a website.
Provide a prioritized bullet-point list of recommendations to improve the wesbite.

"""
```

```python
def messages_for(website):
    return [
        {"role": "system", "content": system_prompt},
        {"role": "user", "content": user_prompt_prefix + website}
    ]
```

```python
def summarize(url):
    website = fetch_website_contents(url)
    response = openai.chat.completions.create(
        model = "openrouter/free",
        messages = messages_for(website)
    )
    return response.choices[0].message.content

```

```python
def summary_markdown(url):
    summary = summarize(url)
    display(Markdown(summary))

```

```python
summary_markdown("https://farazdataanalyst.com")
```
