```python
import os
from dotenv import load_dotenv
from openai import OpenAI
from IPython.display import Markdown, display, update_display
```

```python
MODEL_GPT = 'gpt-4o-mini'
MODEL_LLAMA = 'llama3.2'
```

```python
load_dotenv(override=True)
api_key = os.getenv('OPENROUTER_API_KEY')
```

```python
openai = OpenAI(
    base_url="https://openrouter.ai/api/v1", 
    api_key = api_key
)
```

```python
question = """
Please explain what this code does and why:
yield from {book.get("author") for book in books if book.get("author")}
"""
```

```python
system_prompt = """You are a helpful technical tutor who provide beginner-friendly answers to the questions about python code, 
software engineering, data science and LLMs"""

user_prompt = """Answer the following question with detailed explanations to a total beginner: """ + question
```

```python
messages = [
    {"role": "system", "content": system_prompt},
    {"role": "user", "content": user_prompt}
]
```

```python
stream = openai.chat.completions.create(
    model=MODEL_GPT,
    messages=messages,
    stream=True
)
response = ""
display_handle = display(Markdown(""), display_id=True)
for chunk in stream:
    response += chunk.choices[0].delta.content or ''
    update_display(Markdown(response), display_id=display_handle.display_id)
```

```python
ollama = OpenAI(
    base_url = "http://localhost:11434/v1",
    api_key= "ollama" 
)
```

```python
response = ollama.chat.completions.create(
        model = MODEL_LLAMA,
        messages = [{"role":"user", "content":question}]
    )
result = response.choices[0].message.content
```

```python
display(Markdown(result))
```
