# End of week 1 exercise

To demonstrate your familiarity with OpenAI API, and also Ollama, build a tool that takes a technical question,  
and responds with an explanation. This is a tool that you will be able to use yourself during the course!

````python
# imports

from dotenv import load_dotenv
from IPython.display import Markdown, display, update_display
from openai import OpenAI
````

````python
# constants

MODEL_GPT = 'gpt-4o-mini'
MODEL_LLAMA = 'llama3.2'
OLLAMA_BASE_URL = 'http://localhost:11434/v1'
````

````python
# set up environment

load_dotenv(override=True)
openai = OpenAI(base_url="https://eu.api.openai.com/v1")
ollama = OpenAI(base_url=OLLAMA_BASE_URL, api_key="ollama")
````

````python
# here is the question; type over this to ask something new

question = """
Please explain what this code does and why:
yield from {book.get("author") for book in books if book.get("author")}
"""
````

````python
# prompts + messages

system_prompt = (
    "You are a helpful technical tutor who answers questions about python code, "
    "software engineering, data science and LLMs. Be clear and concrete."
)
user_prompt = "Please give a detailed explanation to the following question:\n\n" + question

messages = [
    {"role": "system", "content": system_prompt},
    {"role": "user", "content": user_prompt},
]
````

````python
# Get gpt-4o-mini to answer, with streaming

stream = openai.chat.completions.create(
    model=MODEL_GPT,
    messages=messages,
    stream=True,
)

response = ""
display_handle = display(Markdown(""), display_id=True)
for chunk in stream:
    response += chunk.choices[0].delta.content or ""
    cleaned = response.replace("```", "").replace("markdown", "")
    update_display(Markdown(cleaned), display_id=display_handle.display_id)
````

````python
# Get Llama 3.2 to answer

response = ollama.chat.completions.create(
    model=MODEL_LLAMA,
    messages=messages,
)
display(Markdown(response.choices[0].message.content))
````
