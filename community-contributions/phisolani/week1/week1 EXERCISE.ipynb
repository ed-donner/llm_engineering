To demonstrate your familiarity with OpenAI API, and also Ollama, build a tool that takes a technical question,  
and responds with an explanation. This is a tool that you will be able to use yourself during the course!

```python
# imports
from pathlib import Path

from IPython.display import Markdown, display
from openai import OpenAI
```

```python
# constants
MODEL = "qwen2.5-coder:1.5b"
SYSTEM_PROMPT = """
You are a patient programming tutor.
Explain the technical question clearly and accurately for a learner.
Use Markdown with short sections: What it does, How it works, and Example.
Identify the exact Python construct used, such as a set comprehension, list comprehension, or generator expression.
Explain important behavior such as ordering, duplicates, laziness, and errors when relevant.
Do not assume facts that are not present in the question.
"""
```

```python
# set up Ollama through its OpenAI-compatible API
ollama_client = OpenAI(
    base_url="http://localhost:11434/v1",
    api_key="ollama",
 )
```

```python
# type a technical question here
question = """
Please explain what this Python code does and why:

yield from {
    book.get("author")
    for book in books
    if book.get("author")
}
"""
```

```python
# ask the Ollama model for an explanation
response = ollama_client.chat.completions.create(
    model=MODEL,
    messages=[
        {"role": "system", "content": SYSTEM_PROMPT},
        {"role": "user", "content": question},
    ],
    temperature=0.2,
 )

explanation = response.choices[0].message.content.strip()
```

```python
# display and save the explanation as Markdown
display(Markdown(f"## Ollama explanation\n\n{explanation}"))
Path("technical_explanation.md").write_text(explanation + "\n", encoding="utf-8")
print("Saved explanation to technical_explanation.md")
```
