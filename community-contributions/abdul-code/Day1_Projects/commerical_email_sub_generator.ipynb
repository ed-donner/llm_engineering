# Commercial Email Subject Generator

This project uses an LLM to generate a suitable subject line for a commercial email based on its content.

## Libraries

The following libraries are used in this project for working with the OpenAI API, displaying Markdown-formatted output, and managing environment variables.

```python
import os
from dotenv import load_dotenv
from IPython.display import Markdown, display
from openai import OpenAI
openai = OpenAI()
```

## Message Configuration

The LLM will receive a system message that defines its role and a user message containing the email content for which a suitable subject line should be generated.

```python
system_prompt = """
You are a professional commercial email assistant.
Your task is to analyze the provided email content and suggest a clear, concise, and professional subject line.
The subject should accurately reflect the main purpose of the email and should not include information that is not present in the email.
"""
user_prompt_prefix = """
Suggest a suitable subject line for the following commercial email:
"""
def messages_for(email_content):
    return [
        {"role": "system", "content": system_prompt},
        {"role": "user", "content": user_prompt_prefix + email_content}
    ]

```

## Email Subject Generation Function

The following function takes the email content as input and uses the LLM with the defined system and user prompts to generate a suitable, professional email subject line.

```python
def generate_subject_line(email_content):
    response = openai.chat.completions.create(
        model = "gpt-5.6-luna",
        messages = messages_for(email_content)
    )
    return response.choices[0].message.content.strip()


```

## display function to make output appealing

```python
def display_subject_line(email_content):
    subject_line = generate_subject_line(email_content)
    display(Markdown(f"**Generated Subject Line:**\n\n{subject_line}"))
```

## Generate Email Subject

Now we call the LLM with the system and user prompts to generate a suitable subject line for the provided email.

```python
email_content = """
Hi Sarah,

I hope you're doing well.

I'm reaching out to introduce our new AI-powered customer support platform, which helps businesses automate repetitive support tasks, reduce response times, and provide 24/7 assistance to customers.

We'd love to schedule a short 20-minute demo to show you how the platform works and discuss how it could fit your team's workflow.

Please let me know if you're available sometime this week.

Best regards,
Abdul Samad Tariq
Associate Software Engineer
"""
display_subject_line(email_content)
```
