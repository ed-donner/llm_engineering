# Portfolio Website → Professional Urdu Summary

This project takes a portfolio website URL and uses an LLM to generate a professional summary in Urdu.

The summary covers:

- Professional Summary
- Education
- Skills
- Work Experience

## Objective

The objective of this project is to extract relevant professional information from a real portfolio website and use an LLM to summarize it in a clear and professional Urdu format.

## Workflow

The application follows this workflow:

Portfolio URL
→ Extract Website Content
→ Provide Content to LLM
→ Generate Structured Urdu Summary

## Input

The application takes a publicly accessible portfolio website URL as input.

The content retrieved from the website will be used as the source material for generating the summary.

## Webpage Content Extraction

The following function takes a portfolio URL and extracts the relevant text content from the webpage.

This extracted content will later be provided to the LLM for summarization.

```python
# Standard headers to fetch a website
headers = {
    "User-Agent": "Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/117.0.0.0 Safari/537.36"
}

def fetch_website_contents(url):
    """
    Return the title and contents of the website at the given url;
    truncate to 2,000 characters as a sensible limit
    """
    try:
        response = requests.get(url, headers=headers)
        soup = BeautifulSoup(response.content, "html.parser")
        title = soup.title.string if soup.title else "No title found"
        if soup.body:
            for irrelevant in soup.body(["script", "style", "img", "input"]):
                irrelevant.decompose()
            text = soup.body.get_text(separator="\n", strip=True)
        else:
            text = ""
    except Exception as e:
        title = "No title found"
        text = "No content found on this page"
    return (title + "\n\n" + text)[:2000]




```

## Prompt Design

We use two prompts:

- **System Prompt:** Defines the role and instructions for the LLM.
- **User Prompt:** Provides the extracted portfolio content and requests the required Urdu summary.

```python
system_prompt = """
You are a professional recruitment specialist analyzing a candidate's portfolio.

Your task is to carefully examine the provided portfolio content and identify the candidate's professional information, including:

- Professional Summary
- Education
- Work Experience, including organizations and roles
- Technical and professional Skills

Only use information explicitly available in the provided portfolio. Do not assume, infer, or invent any information.

Organize the extracted information clearly and provide the final summary in professional Urdu.
"""
user_prompt_prefix = """
Analyze the following portfolio content and create a professional Urdu summary of the candidate.

Extract and organize the information under these sections:
1. Professional Summary
2. Education
3. Work Experience
4. Skills

Portfolio Content:
"""

# See how this function creates exactly the format above

def messages_for(website):
    return [
        {"role": "system", "content": system_prompt},
        {"role": "user", "content": user_prompt_prefix + website}
    ]
```

## Summarization Function

The following function combines the system instructions with the portfolio content and sends them to the LLM.

It returns a structured professional summary in Urdu containing the candidate's professional summary, education, work experience, and skills.

```python
def summarize(url):
    website_data = fetch_website_contents(url)
    response = openai.chat.completions.create(
        model = "gpt-5.6-luna",
        messages = messages_for(website_data)
    )
    return response.choices[0].message.content
```

```python
# A function to display this nicely in the output, using markdown

def display_summary(url):
    summary = summarize(url)
    display(Markdown(summary))
```

## Final Result

The LLM identifies the important professional information, including:

Finally, the extracted information is organized into a clear and professional summary in Urdu.
