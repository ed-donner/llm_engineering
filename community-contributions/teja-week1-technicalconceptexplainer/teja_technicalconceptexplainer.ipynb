````python
# User prompt for the technical concept explainer. This prompt is used to ask the assistant to explain a specific technical concept or piece of code, following the guidelines set in the system prompt.
user_prompt = """Please explain what this code does and why:
```python
from transformers import AutoTokenizer, AutoModelForSequenceClassification
tokenizer = AutoTokenizer.from_pretrained("bert-base-uncased")
model = AutoModelForSequenceClassification.from_pretrained("bert-base-uncased")
text = "Transformers are amazing!"
inputs = tokenizer(text, return_tensors="pt")
outputs = model(**inputs)
```
"""
````

````python
# Creating the messages list that will be sent to the Ollama API. This list includes the system prompt and the user prompt, which together provide context for the assistant to generate a response.
messages = [
    {"role": "system", "content": system_prompt},
    {"role": "user", "content": user_prompt}
]
````

````python
# Function to stream the response from the Ollama API. This function takes the messages list as input, sends it to the Ollama API for processing, and streams the response back in real-time. The response is displayed in a Jupyter notebook cell using Markdown formatting.
def stream_response(messages):
    stream = ollama.chat.completions.create(
        model="deepseek-r1:1.5b",
        messages=messages,
        stream=True
    )    
    response = ""
    display_handle = display(Markdown(""), display_id=True)
    for chunk in stream:
        response += chunk.choices[0].delta.content or ''
        update_display(Markdown(response), display_id=display_handle.display_id)
````

````python
stream_response(messages)
````
