```python
import os
import sys
from scrape import fetch_website_contents

from dotenv import load_dotenv
from openai import OpenAI
from IPython.display import Markdown, display
```

```python
# Load variables from the .env file
load_dotenv()

api_key = os.getenv("OPENAI_API_KEY")
```

```python
# Create the OpenAI client
openai = OpenAI()
```

```python
# Fetch website content
website = fetch_website_contents("https://www.wikipedia.org")

print(website[:2000])
```

```python
system_prompt = """

You are an assistant that analyzes website content
and produces a short, useful summary.

Ignore navigation, menus, and irrelevant website text.

Respond in Markdown.

"""
```

```python
user_prompt_prefix = """

Here is the content of a website:

Provide a short summary of this website in Markdown.

If it includes news or announcements,
summarize those too.

"""
```

```python
def message_for(website):

    return [
        {
            "role": "system",
            "content": system_prompt
        },
        {
            "role": "user",
            "content": user_prompt_prefix + website
        }
    ]
```

```python
message_for(website)
```

```python
def summarize(url):

    website = fetch_website_contents(url)

    response = openai.chat.completions.create(
        model="gpt-4.1-mini",
        messages=message_for(website)
    )

    return response.choices[0].message.content
```

```python
def summarize(url):

    website = fetch_website_contents(url)

    response = openai.chat.completions.create(
        model="gpt-4.1-mini",
        messages=message_for(website),
        stream=True
    )

    return response
```

```python
def display_summary(url):

    response = summarize(url)

    for chunk in response:
        content = chunk.choices[0].delta.content

        if content:
            print(content, end="")
```

```python
display_summary("https://www.wikipedia.org")
```
