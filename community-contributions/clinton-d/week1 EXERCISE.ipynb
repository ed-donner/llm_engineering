```python
import os
import json
from dotenv import load_dotenv
from IPython.display import Markdown, display, update_display
from openai import OpenAI
```

```python
# constants

MODEL_GPT = 'gpt-4o-mini'
MODEL_LLAMA = 'gpt-oss'
OLLAMA_BASE_URL = "http://localhost:11434/v1"
```

```python
# set up environment
load_dotenv(override=True)
api_key = os.getenv('OPENAI_API_KEY')

if api_key and api_key.startswith('sk-proj-') and len(api_key)>10:
    print("API key looks good so far")
else:
    print("There might be a problem with your API key? Please visit the troubleshooting notebook!")
```

```python
# here is the question; type over this to ask something new
system_prompt = """
You are a experienced coding tutor whos job is to explain the code provided. Your students are begineer python programmers. Break down the code line by line and explain what each line does and why. Use official resources only. If no suitable official resource exists, return an empty resources array.
"""


# user_prompt = """
# Explain the following
# yield from {book.get("author") for book in books if book.get("author")}
# """

def connect_to_model(model_name: str):
    if model_name == MODEL_LLAMA:
        return OpenAI(base_url=OLLAMA_BASE_URL, api_key="ollama")
    return OpenAI()


def educate_me(MODEL_NAME, QUERY):
    openai = connect_to_model(MODEL_NAME)
    user_prompt = f"""
    Explain the following\n
    {QUERY}
    """
    stream = openai.chat.completions.create(
        model=MODEL_NAME,
        messages=[
            {"role": "system", "content": system_prompt},
            {"role": "user", "content": user_prompt}
            ],
        stream=True
    )    
    response = ""
    display_handle = display(Markdown(""), display_id=True)
    for chunk in stream:
        response += chunk.choices[0].delta.content or ''
        update_display(Markdown(response), display_id=display_handle.display_id)
```

```python
# Get gpt-4o-mini to answer, with streaming
educate_me(MODEL_GPT, 'yield from {book.get("author") for book in books if book.get("author")}')
```

```python
# Get Llama 3.2 to answer
educate_me(MODEL_LLAMA, 'yield from {book.get("author") for book in books if book.get("author")}')
```
